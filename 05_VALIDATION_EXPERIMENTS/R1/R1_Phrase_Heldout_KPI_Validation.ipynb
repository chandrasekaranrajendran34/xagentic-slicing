```
import numpy as np
import pandas as pd
from pathlib import Path

SEEDS = [42, 43, 44, 45, 46]

CLASS_SPECIFIC = df[df["uc_kind"].isin(["known", "unseen"])].copy()
GENERIC = df[df["uc_kind"] == "generic"].copy()

def build_family_split(df, seed, test_fraction=0.30):
    rng = np.random.default_rng(seed)

    train_families = []
    test_families = []

    # Split phrase/use-case families independently within
    # each slice and vocabulary category.
    family_table = (
        df[df["uc_kind"].isin(["known", "unseen"])]
        [["slice", "uc_kind", "use_case"]]
        .drop_duplicates()
    )

    for (slice_name, kind), grp in family_table.groupby(
        ["slice", "uc_kind"]
    ):
        families = sorted(grp["use_case"].tolist())
        families = np.array(families, dtype=object)

        rng.shuffle(families)

        n_test = max(1, int(round(len(families) * test_fraction)))

        test = families[:n_test]
        train = families[n_test:]

        for fam in train:
            train_families.append({
                "slice": slice_name,
                "uc_kind": kind,
                "use_case": fam
            })

        for fam in test:
            test_families.append({
                "slice": slice_name,
                "uc_kind": kind,
                "use_case": fam
            })

    train_families = pd.DataFrame(train_families)
    test_families = pd.DataFrame(test_families)

    return train_families, test_families
```

```
all_split_rows = []

for seed in SEEDS:

    train_fam, test_fam = build_family_split(df, seed)

    train_set = set(
        zip(
            train_fam["slice"],
            train_fam["uc_kind"],
            train_fam["use_case"]
        )
    )

    test_set = set(
        zip(
            test_fam["slice"],
            test_fam["uc_kind"],
            test_fam["use_case"]
        )
    )

    overlap = train_set & test_set

    assert len(overlap) == 0, (
        f"LEAKAGE DETECTED seed={seed}: {overlap}"
    )

    train_fam["partition"] = "train"
    test_fam["partition"] = "test"

    manifest_seed = pd.concat(
        [train_fam, test_fam],
        ignore_index=True
    )

    manifest_seed["seed"] = seed

    manifest_seed.to_csv(
        R1 / f"splits/phrase_family_split_seed{seed}.csv",
        index=False
    )

    all_split_rows.append(manifest_seed)

    print(
        f"Seed {seed}: "
        f"train families={len(train_fam)}, "
        f"test families={len(test_fam)}, "
        f"overlap={len(overlap)}"
    )

all_splits = pd.concat(all_split_rows, ignore_index=True)

display(all_splits.head(30))
```

```
import pandas as pd
import numpy as np
import re
from pathlib import Path

# ============================================================
# R1 DATASET CONSTRUCTION
# ============================================================

R1 = Path("/content/revision_experiments/R1_phrase_generalization")

df = pd.read_csv(
    ROOT / "02_E5_QWEN37_N100_SEED52" / "corpus.csv"
).copy()

KPI_COLS = [
    "lat_true",
    "thr_true",
    "rel_true",
    "den_true",
    "mob_true",
]

# ------------------------------------------------------------
# Helper: replace the use-case phrase with neutral wording.
#
# Important:
# We preserve the ORIGINAL sentence/template and KPI wording.
# Only the use-case lexical cue is removed.
# ------------------------------------------------------------

NEUTRAL_PHRASE = "network service"

def make_generic_kpi_text(row):
    text = str(row["text"])
    uc = str(row["use_case"])

    # Exact replacement because synth_corpus inserted use_case literally.
    new_text, n = re.subn(
        re.escape(uc),
        NEUTRAL_PHRASE,
        text,
        count=1,
        flags=re.IGNORECASE,
    )

    if n != 1:
        raise ValueError(
            f"Could not replace use_case for id={row['id']}: "
            f"{uc!r} in {text!r}"
        )

    return new_text


# ------------------------------------------------------------
# KPI-only text.
#
# Use only KPIs that were actually present in the original
# template/text. Do NOT reveal hidden latent KPI values.
#
# We therefore derive KPI-only wording from the observable
# natural-language declaration rather than dumping all *_true
# columns into the text.
# ------------------------------------------------------------

def extract_observable_kpi_fragments(text):

    fragments = []

    patterns = [
        # latency
        (
            "latency",
            r"(?i)(?:latency|delay)[^.;,]*?"
            r"([0-9][0-9,]*\.?[0-9]*)\s*ms"
        ),

        # reversed latency expression, e.g. "2.2 ms delay"
        (
            "latency_rev",
            r"(?i)([0-9][0-9,]*\.?[0-9]*)\s*ms"
            r"[^.;,]*?(?:latency|delay)"
        ),

        # throughput / rate
        (
            "throughput",
            r"(?i)([0-9][0-9,]*\.?[0-9]*)\s*Mbps"
        ),

        # reliability / availability
        (
            "reliability",
            r"(?i)([0-9][0-9,]*\.?[0-9]*)\s*"
            r"(?:%|percent)[^.;,]*?"
            r"(?:reliability|availability)"
        ),

        # reversed reliability wording
        (
            "reliability_rev",
            r"(?i)(?:reliability|availability)[^.;,]*?"
            r"([0-9][0-9,]*\.?[0-9]*)\s*(?:%|percent)"
        ),

        # density
        (
            "density",
            r"(?i)([0-9][0-9,]*\.?[0-9]*)\s*"
            r"(?:devices|terminals)[^.;,]*?(?:km2|km²|square km)"
        ),

        # mobility
        (
            "mobility",
            r"(?i)([0-9][0-9,]*\.?[0-9]*)\s*km/h"
        ),
    ]

    found = {}

    for name, pattern in patterns:
        m = re.search(pattern, text)

        if m:
            base_name = (
                "latency" if name.startswith("latency")
                else "reliability" if name.startswith("reliability")
                else name
            )

            if base_name not in found:
                found[base_name] = m.group(1)

    if "latency" in found:
        fragments.append(
            f"latency requirement {found['latency']} ms"
        )

    if "throughput" in found:
        fragments.append(
            f"throughput requirement {found['throughput']} Mbps"
        )

    if "reliability" in found:
        fragments.append(
            f"reliability requirement {found['reliability']} percent"
        )

    if "density" in found:
        fragments.append(
            f"device density requirement {found['density']} devices per km2"
        )

    if "mobility" in found:
        fragments.append(
            f"mobility requirement {found['mobility']} km/h"
        )

    return fragments


def make_kpi_only_text(row):

    fragments = extract_observable_kpi_fragments(str(row["text"]))

    if not fragments:
        raise ValueError(
            f"No observable KPI extracted for id={row['id']}: "
            f"{row['text']}"
        )

    return "; ".join(fragments) + "."


# ============================================================
# Construct common L1/L3/L4 master data
# ============================================================

master = df.copy()

master["text_L1_original"] = master["text"]

master["text_L3_generic_kpi"] = master.apply(
    make_generic_kpi_text,
    axis=1
)

master["text_L4_kpi_only"] = master.apply(
    make_kpi_only_text,
    axis=1
)

master.to_csv(
    R1 / "data/r1_master_conditions.csv",
    index=False
)

print("R1 master conditions created:", master.shape)

display(
    master[
        [
            "id",
            "slice",
            "uc_kind",
            "ambiguous",
            "text_L1_original",
            "text_L3_generic_kpi",
            "text_L4_kpi_only",
        ]
    ].head(10)
)
```

```
# ============================================================
# R1 DATASET SANITY AUDIT
# ============================================================

audit = []

for _, row in master.iterrows():

    original = str(row["text_L1_original"])
    generic  = str(row["text_L3_generic_kpi"])
    kpi_only = str(row["text_L4_kpi_only"])

    uc = str(row["use_case"])

    audit.append({
        "id": row["id"],
        "slice": row["slice"],
        "uc_kind": row["uc_kind"],
        "ambiguous": row["ambiguous"],

        # L3 must remove original phrase
        "l3_original_phrase_removed":
            uc.lower() not in generic.lower(),

        # L3 must contain neutral phrase
        "l3_has_neutral_phrase":
            NEUTRAL_PHRASE.lower() in generic.lower(),

        # L4 must contain no original phrase
        "l4_original_phrase_removed":
            uc.lower() not in kpi_only.lower(),

        # L4 must contain at least one KPI
        "l4_has_kpi":
            any(
                x in kpi_only.lower()
                for x in [
                    "latency requirement",
                    "throughput requirement",
                    "reliability requirement",
                    "device density requirement",
                    "mobility requirement",
                ]
            ),

        "l1_length": len(original),
        "l3_length": len(generic),
        "l4_length": len(kpi_only),
    })

audit = pd.DataFrame(audit)

display(audit.head())

print("\n=== FAILURES ===")

for col in [
    "l3_original_phrase_removed",
    "l3_has_neutral_phrase",
    "l4_original_phrase_removed",
    "l4_has_kpi",
]:
    failures = (~audit[col]).sum()
    print(f"{col:32s}: {failures}")

audit.to_csv(
    R1 / "audit/r1_condition_sanity.csv",
    index=False
)

assert audit["l3_original_phrase_removed"].all()
assert audit["l3_has_neutral_phrase"].all()
assert audit["l4_original_phrase_removed"].all()
assert audit["l4_has_kpi"].all()

print("\nR1 CONDITION SANITY AUDIT: PASS")
```

```
from sklearn.model_selection import train_test_split
import pandas as pd
import json

SEEDS = [42, 43, 44, 45, 46]

split_summary = []

for seed in SEEDS:

    # ========================================================
    # L1/L3/L4 common row split
    # ========================================================

    train_idx, test_idx = train_test_split(
        master.index,
        test_size=0.20,
        random_state=seed,
        stratify=master["slice"]
    )

    common_split = pd.DataFrame({
        "id": master.loc[
            list(train_idx) + list(test_idx), "id"
        ].values,

        "partition":
            ["train"] * len(train_idx) +
            ["test"] * len(test_idx)
    })

    common_split.to_csv(
        R1 / f"splits/common_row_split_seed{seed}.csv",
        index=False
    )

    # ========================================================
    # L2 phrase-family split
    # ========================================================

    fam = pd.read_csv(
        R1 / f"splits/phrase_family_split_seed{seed}.csv"
    )

    train_families = set(
        fam.loc[
            fam["partition"] == "train",
            "use_case"
        ]
    )

    test_families = set(
        fam.loc[
            fam["partition"] == "test",
            "use_case"
        ]
    )

    # L2 deliberately excludes generic cases.
    l2_pool = master[
        master["uc_kind"].isin(["known", "unseen"])
    ].copy()

    l2_train = l2_pool[
        l2_pool["use_case"].isin(train_families)
    ].copy()

    l2_test = l2_pool[
        l2_pool["use_case"].isin(test_families)
    ].copy()

    # Critical leakage assertion
    actual_train_families = set(l2_train["use_case"])
    actual_test_families = set(l2_test["use_case"])

    overlap = actual_train_families & actual_test_families

    assert not overlap, (
        f"L2 leakage seed={seed}: {overlap}"
    )

    l2_split = pd.concat([
        l2_train[["id"]].assign(partition="train"),
        l2_test[["id"]].assign(partition="test")
    ])

    l2_split.to_csv(
        R1 / f"splits/L2_row_split_seed{seed}.csv",
        index=False
    )

    # ========================================================
    # Summary
    # ========================================================

    for condition, tr, te in [

        (
            "L1_original",
            master.loc[train_idx],
            master.loc[test_idx]
        ),

        (
            "L3_generic_kpi",
            master.loc[train_idx],
            master.loc[test_idx]
        ),

        (
            "L4_kpi_only",
            master.loc[train_idx],
            master.loc[test_idx]
        ),

        (
            "L2_phrase_family_heldout",
            l2_train,
            l2_test
        ),
    ]:

        split_summary.append({
            "seed": seed,
            "condition": condition,
            "train_n": len(tr),
            "test_n": len(te),

            "train_eMBB":
                int((tr["slice"] == "eMBB").sum()),
            "train_URLLC":
                int((tr["slice"] == "URLLC").sum()),
            "train_mMTC":
                int((tr["slice"] == "mMTC").sum()),

            "test_eMBB":
                int((te["slice"] == "eMBB").sum()),
            "test_URLLC":
                int((te["slice"] == "URLLC").sum()),
            "test_mMTC":
                int((te["slice"] == "mMTC").sum()),

            "test_ambiguous_0":
                int((te["ambiguous"] == 0).sum()),
            "test_ambiguous_1":
                int((te["ambiguous"] == 1).sum()),
        })


split_summary = pd.DataFrame(split_summary)

split_summary.to_csv(
    R1 / "audit/r1_split_summary.csv",
    index=False
)

display(split_summary)

print("\nR1 FINAL SPLITS: CREATED")
```

```
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_recall_fscore_support,
    confusion_matrix,
)

import exp1_posthoc as e1
import xagentic_common as xc


SLICES = ["eMBB", "URLLC", "mMTC"]
SEEDS = [42, 43, 44, 45, 46]


def encode_labels(series):
    mapping = {
        "eMBB": 0,
        "URLLC": 1,
        "mMTC": 2,
    }

    y = series.map(mapping)

    if y.isna().any():
        raise ValueError("Unknown slice label encountered.")

    return y.astype(int).to_numpy()


def median_impute_train_only(Xtr, Xte):
    """
    Match frozen E1:
    median is learned from TRAIN only.
    """
    med = Xtr.median(numeric_only=True)

    return (
        Xtr.fillna(med),
        Xte.fillna(med),
        med
    )
```

```
def select_model_train_only(Xtr, ytr, seed):

    # Important:
    # exp1_posthoc.build_models() uses module-level SEED.
    # Set it explicitly for each R1 run.
    e1.SEED = seed

    models = e1.build_models()

    skf = StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=seed
    )

    rows = []

    for name, model in models.items():

        cv = cross_val_score(
            model,
            Xtr,
            ytr,
            cv=skf,
            scoring="f1_macro",
            n_jobs=1
        )

        rows.append({
            "model": name,
            "cv_mean": float(np.mean(cv)),
            "cv_std": float(np.std(cv)),
        })

    table = (
        pd.DataFrame(rows)
        .sort_values(
            ["cv_mean", "model"],
            ascending=[False, True]
        )
        .reset_index(drop=True)
    )

    best_name = table.iloc[0]["model"]

    best_model = clone(models[best_name])

    return best_name, best_model, table
```

```
def load_r1_condition(seed, condition):

    if condition in {
        "L1_original",
        "L3_generic_kpi",
        "L4_kpi_only",
    }:

        sp = pd.read_csv(
            R1 / f"splits/common_row_split_seed{seed}.csv"
        )

    elif condition == "L2_phrase_family_heldout":

        sp = pd.read_csv(
            R1 / f"splits/L2_row_split_seed{seed}.csv"
        )

    else:
        raise ValueError(
            f"Unknown condition: {condition}"
        )

    train_ids = set(
        sp.loc[
            sp["partition"] == "train",
            "id"
        ]
    )

    test_ids = set(
        sp.loc[
            sp["partition"] == "test",
            "id"
        ]
    )

    assert train_ids.isdisjoint(test_ids)

    train_df = (
        master[
            master["id"].isin(train_ids)
        ]
        .copy()
        .sort_values("id")
        .reset_index(drop=True)
    )

    test_df = (
        master[
            master["id"].isin(test_ids)
        ]
        .copy()
        .sort_values("id")
        .reset_index(drop=True)
    )

    text_column = {
        "L1_original":
            "text_L1_original",

        "L2_phrase_family_heldout":
            "text_L1_original",

        "L3_generic_kpi":
            "text_L3_generic_kpi",

        "L4_kpi_only":
            "text_L4_kpi_only",
    }[condition]

    train_df["text"] = train_df[text_column]
    test_df["text"] = test_df[text_column]

    # ------------------------------------------
    # Special L2 leakage assertion
    # ------------------------------------------

    if condition == "L2_phrase_family_heldout":

        train_families = set(
            train_df["use_case"]
        )

        test_families = set(
            test_df["use_case"]
        )

        overlap = (
            train_families &
            test_families
        )

        assert len(overlap) == 0, (
            f"L2 FAMILY LEAKAGE: {overlap}"
        )

    return train_df, test_df
```

```
def run_one_r1(seed, condition):

    train_df, test_df = load_r1_condition(
        seed,
        condition
    )

    # ------------------------------------------
    # Frozen feature builder
    # ------------------------------------------

    Xtr_raw = xc.build_features(train_df)
    Xte_raw = xc.build_features(test_df)

    assert list(Xtr_raw.columns) == list(Xte_raw.columns)
    assert Xtr_raw.shape[1] == 14

    ytr = encode_labels(train_df["slice"])
    yte = encode_labels(test_df["slice"])

    # ------------------------------------------
    # TRAIN-only imputation
    # ------------------------------------------

    Xtr, Xte, medians = (
        median_impute_train_only(
            Xtr_raw,
            Xte_raw
        )
    )

    assert not Xtr.isna().any().any()
    assert not Xte.isna().any().any()

    # ------------------------------------------
    # Training-only CV model selection
    # ------------------------------------------

    best_name, model, cv_table = (
        select_model_train_only(
            Xtr,
            ytr,
            seed
        )
    )

    # ------------------------------------------
    # Final fit
    # ------------------------------------------

    model.fit(Xtr, ytr)

    pred = model.predict(Xte)

    # ------------------------------------------
    # Main metrics
    # ------------------------------------------

    acc = accuracy_score(yte, pred)

    macro_f1 = f1_score(
        yte,
        pred,
        average="macro"
    )

    pr, rc, f1c, sup = (
        precision_recall_fscore_support(
            yte,
            pred,
            labels=[0, 1, 2],
            zero_division=0
        )
    )

    cm = confusion_matrix(
        yte,
        pred,
        labels=[0, 1, 2]
    )

    # ------------------------------------------
    # Difficulty / ambiguity strata
    # ------------------------------------------

    test_df = test_df.copy()

    test_df["y_true"] = yte
    test_df["y_pred"] = pred
    test_df["correct"] = (
        test_df["y_true"] ==
        test_df["y_pred"]
    )

    subgroup_rows = []

    for ambiguity in [0, 1]:

        mask = (
            test_df["ambiguous"] ==
            ambiguity
        ).to_numpy()

        if mask.sum() == 0:
            continue

        subgroup_rows.append({
            "ambiguous": ambiguity,
            "n": int(mask.sum()),

            "accuracy":
                float(
                    accuracy_score(
                        yte[mask],
                        pred[mask]
                    )
                ),

            "macro_f1":
                float(
                    f1_score(
                        yte[mask],
                        pred[mask],
                        average="macro",
                        zero_division=0
                    )
                ),
        })

    # ------------------------------------------
    # Summary
    # ------------------------------------------

    summary = {
        "seed": seed,
        "condition": condition,

        "train_n": len(train_df),
        "test_n": len(test_df),

        "selected_model": best_name,

        "selected_cv_mean":
            float(
                cv_table.iloc[0]["cv_mean"]
            ),

        "selected_cv_std":
            float(
                cv_table.iloc[0]["cv_std"]
            ),

        "accuracy": float(acc),
        "macro_f1": float(macro_f1),

        "test_nonambiguous_n":
            int(
                (test_df["ambiguous"] == 0)
                .sum()
            ),

        "test_ambiguous_n":
            int(
                (test_df["ambiguous"] == 1)
                .sum()
            ),
    }

    per_class = pd.DataFrame({
        "slice": SLICES,
        "precision": pr,
        "recall": rc,
        "f1": f1c,
        "support": sup,
    })

    confusion = pd.DataFrame(
        cm,
        index=SLICES,
        columns=SLICES
    )

    subgroup = pd.DataFrame(
        subgroup_rows
    )

    return {
        "summary": summary,
        "cv": cv_table,
        "per_class": per_class,
        "confusion": confusion,
        "subgroup": subgroup,
        "predictions": test_df,
        "medians": medians,
    }
```

```
# ================================================================
# STEP 22 — FINAL R1 L1 / L3 / L4 DATASET CONSTRUCTION
# ================================================================

import re
import pandas as pd
import numpy as np
from pathlib import Path

SEEDS = [42, 43, 44, 45, 46]

FINAL_COND = R1 / "data/final_conditions"

for condition in [
    "L1_original",
    "L3_generic_kpi",
    "L4_kpi_only",
]:
    (FINAL_COND / condition).mkdir(
        parents=True,
        exist_ok=True
    )


# ------------------------------------------------
# L3:
# Replace the exact use-case phrase with a
# class-neutral phrase.
#
# KPI wording and template structure remain intact.
# ------------------------------------------------

def make_l3_text(row):

    text = str(row["text"])
    use_case = str(row["use_case"])

    if use_case not in text:
        raise ValueError(
            f"Use-case phrase not found.\n"
            f"id={row['id']}\n"
            f"use_case={use_case}\n"
            f"text={text}"
        )

    return text.replace(
        use_case,
        "network service",
        1
    )


# ------------------------------------------------
# L4:
# Keep ONLY KPI fragments explicitly observable
# in the original intent.
#
# IMPORTANT:
# We do NOT reconstruct unstated latent KPI values.
# ------------------------------------------------

KPI_PATTERNS = [
    # latency
    r"latency\s*(?:<=|≤|=|of|is|around|about)?\s*"
    r"\d+(?:\.\d+)?\s*ms",

    # throughput
    r"throughput\s*(?:>=|≥|=|of|is|around|about)?\s*"
    r"\d+(?:\.\d+)?\s*(?:Mbps|Gbps)",

    # reliability
    r"reliability\s*(?:>=|≥|=|of|is|around|about)?\s*"
    r"\d+(?:\.\d+)?\s*%",

    # density
    r"density\s*(?:>=|≥|=|of|is|around|about)?\s*"
    r"\d+(?:\.\d+)?\s*(?:devices/)?km(?:\^?2|²)",

    # mobility
    r"mobility\s*(?:<=|≤|=|of|is|around|about)?\s*"
    r"\d+(?:\.\d+)?\s*km/h",
]


def make_l4_text(text):

    text = str(text)

    fragments = []

    for pattern in KPI_PATTERNS:

        matches = re.findall(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        fragments.extend(matches)

    if not fragments:
        raise ValueError(
            "No observable KPI fragments found:\n"
            + text
        )

    return "; ".join(fragments)


# ------------------------------------------------
# Generate all seeds
# ------------------------------------------------

construction_rows = []

for seed in SEEDS:

    src = (
        FINAL_DATA /
        f"r1_corpus_seed{seed}.csv"
    )

    df = pd.read_csv(src)

    assert len(df) == 3000

    # ---------------------------
    # L1
    # ---------------------------

    l1 = df.copy()

    l1["r1_condition"] = "L1_original"
    l1["r1_text"] = l1["text"]

    # ---------------------------
    # L3
    # ---------------------------

    l3 = df.copy()

    l3["r1_condition"] = "L3_generic_kpi"

    l3["r1_text"] = l3.apply(
        make_l3_text,
        axis=1
    )

    # ---------------------------
    # L4
    # ---------------------------

    l4 = df.copy()

    l4["r1_condition"] = "L4_kpi_only"

    l4["r1_text"] = l4["text"].apply(
        make_l4_text
    )

    # ---------------------------
    # Save
    # ---------------------------

    l1.to_csv(
        FINAL_COND /
        "L1_original" /
        f"seed_{seed}.csv",
        index=False
    )

    l3.to_csv(
        FINAL_COND /
        "L3_generic_kpi" /
        f"seed_{seed}.csv",
        index=False
    )

    l4.to_csv(
        FINAL_COND /
        "L4_kpi_only" /
        f"seed_{seed}.csv",
        index=False
    )

    construction_rows.append({
        "seed": seed,
        "L1_n": len(l1),
        "L3_n": len(l3),
        "L4_n": len(l4),
        "L1_unique_ids": l1["id"].nunique(),
        "L3_unique_ids": l3["id"].nunique(),
        "L4_unique_ids": l4["id"].nunique(),
    })


construction = pd.DataFrame(
    construction_rows
)

display(construction)

assert (construction["L1_n"] == 3000).all()
assert (construction["L3_n"] == 3000).all()
assert (construction["L4_n"] == 3000).all()

print()
print("=" * 90)
print("FINAL L1/L3/L4 DATASET CONSTRUCTION: PASS")
print("=" * 90)
```

```
# ================================================================
# STEP 22A — CORRECT L4 USING FROZEN PARSER
# ================================================================

import numpy as np

def make_l4_text(text):
    """
    Construct KPI-only text using ONLY KPI values observable by
    the frozen xagentic_common.parse_slots() parser.

    No latent ground-truth KPI values are used.
    """

    text = str(text)

    parsed = xc.parse_slots(text)

    fragments = []

    # IMPORTANT:
    # Use wording understood by the same frozen parser.

    if np.isfinite(float(parsed.latency_ms)):
        fragments.append(
            f"latency {float(parsed.latency_ms):g} ms"
        )

    if np.isfinite(float(parsed.throughput_mbps)):
        fragments.append(
            f"throughput {float(parsed.throughput_mbps):g} Mbps"
        )

    if np.isfinite(float(parsed.reliability_pct)):
        fragments.append(
            f"reliability {float(parsed.reliability_pct):g}%"
        )

    if np.isfinite(float(parsed.density_per_km2)):
        fragments.append(
            f"density {float(parsed.density_per_km2):g} devices/km2"
        )

    if np.isfinite(float(parsed.mobility_kmh)):
        fragments.append(
            f"mobility {float(parsed.mobility_kmh):g} km/h"
        )

    if not fragments:
        raise ValueError(
            "Frozen parser found no observable KPI values:\n"
            + text
        )

    return "; ".join(fragments)


# ------------------------------------------------
# Test specifically against the failed example
# ------------------------------------------------

failed_example = (
    "Set up municipal network service. "
    "It should sustain 126 Mbps for roughly "
    "7,624 devices per km2."
)

print("ORIGINAL:")
print(failed_example)

print("\nFROZEN PARSER:")
print(xc.parse_slots(failed_example))

print("\nL4:")
print(make_l4_text(failed_example))

print("\nL4 RE-PARSED:")
print(
    xc.parse_slots(
        make_l4_text(failed_example)
    )
)
```

```
# ================================================================
# STEP 24 — FINAL L1 / L3 / L4 GENERATION + FULL HARD AUDIT
# ================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json

SEEDS = [42, 43, 44, 45, 46]

FINAL_COND = R1 / "data/final_conditions"

for condition in [
    "L1_original",
    "L3_generic_kpi",
    "L4_kpi_only",
]:
    (FINAL_COND / condition).mkdir(
        parents=True,
        exist_ok=True
    )

NON_LEX = [
    "latency_ms",
    "throughput_mbps",
    "reliability_pct",
    "density_per_km2",
    "mobility_kmh",
    "log_latency",
    "log_throughput",
    "log_density",
    "miss_latency",
    "miss_throughput",
    "miss_density",
]

LEX = [
    "lex_embb",
    "lex_urllc",
    "lex_mmtc",
]


# ------------------------------------------------
# L3 helper
# ------------------------------------------------

def make_l3_text(row):

    text = str(row["text"])
    use_case = str(row["use_case"])

    if use_case not in text:
        raise ValueError(
            f"Use-case phrase not found: "
            f"id={row['id']} | "
            f"use_case={use_case}"
        )

    return text.replace(
        use_case,
        "network service",
        1
    )


# ------------------------------------------------
# L4 helper — frozen parser based
# ------------------------------------------------

def make_l4_text(text):

    text = str(text)

    parsed = xc.parse_slots(text)

    fragments = []

    if np.isfinite(float(parsed.latency_ms)):
        fragments.append(
            f"latency {float(parsed.latency_ms):g} ms"
        )

    if np.isfinite(float(parsed.throughput_mbps)):
        fragments.append(
            f"throughput {float(parsed.throughput_mbps):g} Mbps"
        )

    if np.isfinite(float(parsed.reliability_pct)):
        fragments.append(
            f"reliability {float(parsed.reliability_pct):g}%"
        )

    if np.isfinite(float(parsed.density_per_km2)):
        fragments.append(
            f"density {float(parsed.density_per_km2):g} devices/km2"
        )

    if np.isfinite(float(parsed.mobility_kmh)):
        fragments.append(
            f"mobility {float(parsed.mobility_kmh):g} km/h"
        )

    if not fragments:
        raise ValueError(
            "No KPI observable by frozen parser:\n"
            + text
        )

    return "; ".join(fragments)


# ------------------------------------------------
# Utility: build features from r1_text
# ------------------------------------------------

def features_from_r1(df):

    tmp = df.copy()

    tmp["text"] = tmp["r1_text"]

    return xc.build_features(tmp)


# ------------------------------------------------
# Generate + audit
# ------------------------------------------------

audit_rows = []

for seed in SEEDS:

    print("\n" + "=" * 90)
    print(f"SEED {seed}")
    print("=" * 90)

    src = (
        FINAL_DATA /
        f"r1_corpus_seed{seed}.csv"
    )

    base = pd.read_csv(src)

    assert len(base) == 3000
    assert base["id"].nunique() == 3000

    # ============================================================
    # L1
    # ============================================================

    l1 = base.copy()

    l1["r1_condition"] = "L1_original"
    l1["r1_text"] = l1["text"]

    # ============================================================
    # L3
    # ============================================================

    l3 = base.copy()

    l3["r1_condition"] = "L3_generic_kpi"

    l3["r1_text"] = l3.apply(
        make_l3_text,
        axis=1
    )

    # ============================================================
    # L4
    # ============================================================

    l4 = base.copy()

    l4["r1_condition"] = "L4_kpi_only"

    l4["r1_text"] = l4["text"].apply(
        make_l4_text
    )

    # ============================================================
    # BASIC ID / LABEL PRESERVATION
    # ============================================================

    assert l1["id"].equals(l3["id"])
    assert l1["id"].equals(l4["id"])

    assert l1["slice"].equals(l3["slice"])
    assert l1["slice"].equals(l4["slice"])

    assert l1["ambiguous"].equals(
        l3["ambiguous"]
    )

    assert l1["ambiguous"].equals(
        l4["ambiguous"]
    )

    # ============================================================
    # FEATURE EXTRACTION
    # ============================================================

    X1 = features_from_r1(l1)
    X3 = features_from_r1(l3)
    X4 = features_from_r1(l4)

    # ============================================================
    # NON-LEXICAL PRESERVATION
    # ============================================================

    l1_l3_nonlex = np.allclose(
        X1[NON_LEX].to_numpy(),
        X3[NON_LEX].to_numpy(),
        equal_nan=True,
        rtol=0,
        atol=0
    )

    l1_l4_nonlex = np.allclose(
        X1[NON_LEX].to_numpy(),
        X4[NON_LEX].to_numpy(),
        equal_nan=True,
        rtol=0,
        atol=0
    )

    # ============================================================
    # LEXICAL ABLATION
    # ============================================================

    l3_lex_zero = np.allclose(
        X3[LEX].to_numpy(),
        0.0,
        rtol=0,
        atol=0
    )

    l4_lex_zero = np.allclose(
        X4[LEX].to_numpy(),
        0.0,
        rtol=0,
        atol=0
    )

    # ============================================================
    # L3 vs L4 REPRESENTATIONAL EQUIVALENCE
    # ============================================================

    l3_l4_full_equal = np.allclose(
        X3.to_numpy(),
        X4.to_numpy(),
        equal_nan=True,
        rtol=0,
        atol=0
    )

    # ============================================================
    # L3 phrase-removal audit
    # ============================================================

    phrase_failures = 0

    for _, row in l3.iterrows():

        original_phrase = str(
            row["use_case"]
        ).lower()

        transformed = str(
            row["r1_text"]
        ).lower()

        if original_phrase in transformed:
            phrase_failures += 1

    # ============================================================
    # Record
    # ============================================================

    row = {
        "seed": seed,

        "n": len(base),

        "L1_equals_L3_nonlex":
            bool(l1_l3_nonlex),

        "L1_equals_L4_nonlex":
            bool(l1_l4_nonlex),

        "L3_lex_zero":
            bool(l3_lex_zero),

        "L4_lex_zero":
            bool(l4_lex_zero),

        "L3_equals_L4_full_features":
            bool(l3_l4_full_equal),

        "L3_phrase_failures":
            int(phrase_failures),

        "L1_lexical_rows":
            int(
                (X1[LEX].sum(axis=1) > 0)
                .sum()
            ),

        "L3_lexical_rows":
            int(
                (X3[LEX].sum(axis=1) > 0)
                .sum()
            ),

        "L4_lexical_rows":
            int(
                (X4[LEX].sum(axis=1) > 0)
                .sum()
            ),
    }

    audit_rows.append(row)

    print(row)

    # ============================================================
    # HARD SEED-LEVEL GATES
    # ============================================================

    assert l1_l3_nonlex
    assert l1_l4_nonlex

    assert l3_lex_zero
    assert l4_lex_zero

    assert phrase_failures == 0

    # ============================================================
    # SAVE ONLY AFTER PASS
    # ============================================================

    l1.to_csv(
        FINAL_COND /
        "L1_original" /
        f"seed_{seed}.csv",
        index=False
    )

    l3.to_csv(
        FINAL_COND /
        "L3_generic_kpi" /
        f"seed_{seed}.csv",
        index=False
    )

    l4.to_csv(
        FINAL_COND /
        "L4_kpi_only" /
        f"seed_{seed}.csv",
        index=False
    )


# ================================================================
# FINAL AUDIT TABLE
# ================================================================

feature_audit = pd.DataFrame(
    audit_rows
)

audit_path = (
    R1 /
    "audit/final_L1_L3_L4_feature_audit.csv"
)

feature_audit.to_csv(
    audit_path,
    index=False
)

print("\n" + "=" * 100)
print("FINAL L1 / L3 / L4 FEATURE AUDIT")
print("=" * 100)

display(feature_audit)

print("\nSaved:")
print(audit_path)

print("\n" + "=" * 100)
print(
    "FINAL L1/L3/L4 CONSTRUCTION "
    "AND FEATURE AUDIT: PASS"
)
print("=" * 100)
```

```
# ================================================================
# STEP 25 — FINAL L2 STRICT PHRASE-FAMILY-HELD-OUT SPLITS
# ================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json
import hashlib

SEEDS = [42, 43, 44, 45, 46]

L2_DIR = FINAL_COND / "L2_phrase_family_heldout"
L2_DIR.mkdir(parents=True, exist_ok=True)

L2_SPLITS = R1 / "splits/L2_phrase_family_heldout"
L2_SPLITS.mkdir(parents=True, exist_ok=True)

L2_MANIFESTS = R1 / "manifests/L2_phrase_family_heldout"
L2_MANIFESTS.mkdir(parents=True, exist_ok=True)


def sha256_file(path):
    h = hashlib.sha256()

    with open(path, "rb") as f:
        for chunk in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            h.update(chunk)

    return h.hexdigest()


audit_rows = []

for seed in SEEDS:

    print("\n" + "=" * 100)
    print(f"L2 — SEED {seed}")
    print("=" * 100)

    src = (
        FINAL_DATA /
        f"r1_corpus_seed{seed}.csv"
    )

    df = pd.read_csv(src)

    assert len(df) == 3000

    # ------------------------------------------------------------
    # L2 MAIN TEST:
    # use only class-specific known/unseen phrase families.
    #
    # Generic phrases are deliberately excluded because they are
    # shared across labels and therefore cannot provide a strict
    # class-specific family-held-out partition.
    # ------------------------------------------------------------

    eligible = df[
        df["uc_kind"].isin(
            ["known", "unseen"]
        )
    ].copy()

    assert len(eligible) > 0

    rng = np.random.RandomState(seed)

    train_families = []
    test_families = []

    family_manifest = []

    # ------------------------------------------------------------
    # Split separately by class AND phrase type.
    # ------------------------------------------------------------

    for slice_name in xc.SLICES:

        for kind in ["known", "unseen"]:

            families = sorted(
                eligible.loc[
                    (eligible["slice"] == slice_name) &
                    (eligible["uc_kind"] == kind),
                    "use_case"
                ].unique().tolist()
            )

            print(
                f"{slice_name:6s} | "
                f"{kind:6s} | "
                f"{len(families)} families"
            )

            # Frozen corpus design:
            # known  = 5 families
            # unseen = 4 families
            if kind == "known":
                assert len(families) == 5
                n_test = 2
            else:
                assert len(families) == 4
                n_test = 1

            shuffled = list(families)
            rng.shuffle(shuffled)

            test_here = sorted(
                shuffled[:n_test]
            )

            train_here = sorted(
                shuffled[n_test:]
            )

            assert set(train_here).isdisjoint(
                set(test_here)
            )

            train_families.extend(
                [
                    (slice_name, kind, f)
                    for f in train_here
                ]
            )

            test_families.extend(
                [
                    (slice_name, kind, f)
                    for f in test_here
                ]
            )

            for family in train_here:
                family_manifest.append({
                    "seed": seed,
                    "slice": slice_name,
                    "uc_kind": kind,
                    "use_case": family,
                    "partition": "train",
                })

            for family in test_here:
                family_manifest.append({
                    "seed": seed,
                    "slice": slice_name,
                    "uc_kind": kind,
                    "use_case": family,
                    "partition": "test",
                })

    # ------------------------------------------------------------
    # Convert family lists to lookup sets.
    # Include slice + kind + use_case so the definition is explicit.
    # ------------------------------------------------------------

    train_keys = set(train_families)
    test_keys = set(test_families)

    assert train_keys.isdisjoint(
        test_keys
    )

    assert len(train_keys) == 18
    assert len(test_keys) == 9

    # ------------------------------------------------------------
    # Assign rows
    # ------------------------------------------------------------

    def row_key(row):
        return (
            row["slice"],
            row["uc_kind"],
            row["use_case"],
        )

    eligible["_family_key"] = eligible.apply(
        row_key,
        axis=1
    )

    train = eligible[
        eligible["_family_key"].isin(
            train_keys
        )
    ].copy()

    test = eligible[
        eligible["_family_key"].isin(
            test_keys
        )
    ].copy()

    # ------------------------------------------------------------
    # HARD COVERAGE CHECK
    # Every eligible row must belong to exactly one side.
    # ------------------------------------------------------------

    assert (
        len(train) + len(test)
        == len(eligible)
    )

    assert set(train["id"]).isdisjoint(
        set(test["id"])
    )

    # ------------------------------------------------------------
    # STRONGEST LEAKAGE CHECK:
    # NO use_case phrase may occur in both train and test.
    # ------------------------------------------------------------

    train_phrase_set = set(
        train["use_case"].unique()
    )

    test_phrase_set = set(
        test["use_case"].unique()
    )

    phrase_overlap = (
        train_phrase_set &
        test_phrase_set
    )

    assert len(phrase_overlap) == 0, (
        f"Phrase leakage detected: "
        f"{phrase_overlap}"
    )

    # ------------------------------------------------------------
    # Additional class-specific overlap check
    # ------------------------------------------------------------

    train_family_set = set(
        zip(
            train["slice"],
            train["uc_kind"],
            train["use_case"]
        )
    )

    test_family_set = set(
        zip(
            test["slice"],
            test["uc_kind"],
            test["use_case"]
        )
    )

    family_overlap = (
        train_family_set &
        test_family_set
    )

    assert len(family_overlap) == 0

    # ------------------------------------------------------------
    # Add R1 metadata
    # ------------------------------------------------------------

    train["r1_condition"] = (
        "L2_phrase_family_heldout"
    )

    test["r1_condition"] = (
        "L2_phrase_family_heldout"
    )

    train["r1_partition"] = "train"
    test["r1_partition"] = "test"

    train["r1_text"] = train["text"]
    test["r1_text"] = test["text"]

    train.drop(
        columns=["_family_key"],
        inplace=True
    )

    test.drop(
        columns=["_family_key"],
        inplace=True
    )

    # ------------------------------------------------------------
    # Counts
    # ------------------------------------------------------------

    train_counts = (
        train["slice"]
        .value_counts()
        .reindex(xc.SLICES)
        .fillna(0)
        .astype(int)
    )

    test_counts = (
        test["slice"]
        .value_counts()
        .reindex(xc.SLICES)
        .fillna(0)
        .astype(int)
    )

    test_amb_counts = (
        test["ambiguous"]
        .value_counts()
        .sort_index()
    )

    # ------------------------------------------------------------
    # Lexical-hit audit
    #
    # IMPORTANT:
    # L2 is NOT supposed to have zero lexical hits.
    # We preserve the original frozen feature extractor.
    # ------------------------------------------------------------

    Xtr_raw = xc.build_features(train)
    Xte_raw = xc.build_features(test)

    lex_cols = [
        "lex_embb",
        "lex_urllc",
        "lex_mmtc",
    ]

    train_lexical_rows = int(
        (
            Xtr_raw[lex_cols]
            .sum(axis=1) > 0
        ).sum()
    )

    test_lexical_rows = int(
        (
            Xte_raw[lex_cols]
            .sum(axis=1) > 0
        ).sum()
    )

    # ------------------------------------------------------------
    # Save split
    # ------------------------------------------------------------

    train_path = (
        L2_DIR /
        f"seed_{seed}_train.csv"
    )

    test_path = (
        L2_DIR /
        f"seed_{seed}_test.csv"
    )

    train.to_csv(
        train_path,
        index=False
    )

    test.to_csv(
        test_path,
        index=False
    )

    # ------------------------------------------------------------
    # Save family manifest
    # ------------------------------------------------------------

    manifest_df = pd.DataFrame(
        family_manifest
    ).sort_values(
        [
            "slice",
            "uc_kind",
            "partition",
            "use_case",
        ]
    )

    manifest_path = (
        L2_MANIFESTS /
        f"seed_{seed}_families.csv"
    )

    manifest_df.to_csv(
        manifest_path,
        index=False
    )

    # ------------------------------------------------------------
    # Save IDs separately for reproducibility
    # ------------------------------------------------------------

    pd.DataFrame({
        "id": train["id"].tolist()
    }).to_csv(
        L2_SPLITS /
        f"seed_{seed}_train_ids.csv",
        index=False
    )

    pd.DataFrame({
        "id": test["id"].tolist()
    }).to_csv(
        L2_SPLITS /
        f"seed_{seed}_test_ids.csv",
        index=False
    )

    # ------------------------------------------------------------
    # Audit row
    # ------------------------------------------------------------

    audit_row = {
        "seed": seed,

        "eligible_n":
            int(len(eligible)),

        "train_n":
            int(len(train)),

        "test_n":
            int(len(test)),

        "train_families":
            int(len(train_keys)),

        "test_families":
            int(len(test_keys)),

        "phrase_overlap":
            int(len(phrase_overlap)),

        "family_overlap":
            int(len(family_overlap)),

        "train_eMBB":
            int(train_counts["eMBB"]),

        "train_URLLC":
            int(train_counts["URLLC"]),

        "train_mMTC":
            int(train_counts["mMTC"]),

        "test_eMBB":
            int(test_counts["eMBB"]),

        "test_URLLC":
            int(test_counts["URLLC"]),

        "test_mMTC":
            int(test_counts["mMTC"]),

        "test_nonambiguous":
            int(test_amb_counts.get(0, 0)),

        "test_ambiguous":
            int(test_amb_counts.get(1, 0)),

        "train_lexical_rows":
            train_lexical_rows,

        "test_lexical_rows":
            test_lexical_rows,

        "train_sha256":
            sha256_file(train_path),

        "test_sha256":
            sha256_file(test_path),
    }

    audit_rows.append(
        audit_row
    )

    print("\nAUDIT:")
    print(audit_row)


# ================================================================
# FINAL L2 AUDIT
# ================================================================

l2_audit = pd.DataFrame(
    audit_rows
)

l2_audit_path = (
    R1 /
    "audit/final_L2_phrase_family_audit.csv"
)

l2_audit.to_csv(
    l2_audit_path,
    index=False
)

print("\n" + "=" * 110)
print("FINAL L2 PHRASE-FAMILY-HELD-OUT AUDIT")
print("=" * 110)

display(l2_audit)

# ------------------------------------------------
# FINAL HARD GATES
# ------------------------------------------------

assert (
    l2_audit["train_families"] == 18
).all()

assert (
    l2_audit["test_families"] == 9
).all()

assert (
    l2_audit["phrase_overlap"] == 0
).all()

assert (
    l2_audit["family_overlap"] == 0
).all()

assert (
    l2_audit[
        [
            "train_eMBB",
            "train_URLLC",
            "train_mMTC",
        ]
    ] > 0
).all().all()

assert (
    l2_audit[
        [
            "test_eMBB",
            "test_URLLC",
            "test_mMTC",
        ]
    ] > 0
).all().all()

print("\nSaved:")
print(l2_audit_path)

print("\n" + "=" * 110)
print(
    "FINAL L2 PHRASE-FAMILY-HELD-OUT "
    "CONSTRUCTION: PASS"
)
print("=" * 110)
```

```
# ================================================================
# STEP 26 — R1 SEED-42 FOUR-CONDITION TRAINING SMOKE TEST
# ================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json
import time

from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_recall_fscore_support,
    confusion_matrix,
)

SEED = 42

R1_RESULTS = R1 / "results" / f"seed_{SEED}"
R1_RESULTS.mkdir(parents=True, exist_ok=True)

# Make frozen model constructors deterministic for this seed.
xc.SEED = SEED
e1.SEED = SEED

xc.split.__globals__["SEED"] = SEED


# ================================================================
# HELPER — TRAIN / SELECT / EVALUATE
# ================================================================

def run_r1_condition(
    condition,
    train_df,
    test_df,
):

    print("\n" + "=" * 100)
    print(condition)
    print("=" * 100)

    # ------------------------------------------------------------
    # Build features from R1 text
    # ------------------------------------------------------------

    tr = train_df.copy()
    te = test_df.copy()

    tr["text"] = tr["r1_text"]
    te["text"] = te["r1_text"]

    Xtr_raw = xc.build_features(tr)
    Xte_raw = xc.build_features(te)

    label_map = {
        name: i
        for i, name in enumerate(xc.SLICES)
    }

    ytr = (
        train_df["slice"]
        .map(label_map)
        .to_numpy()
    )

    yte = (
        test_df["slice"]
        .map(label_map)
        .to_numpy()
    )

    # ------------------------------------------------------------
    # TRAINING-ONLY MEDIAN IMPUTATION
    # ------------------------------------------------------------

    med = Xtr_raw.median(
        numeric_only=True
    )

    Xtr = Xtr_raw.fillna(med)
    Xte = Xte_raw.fillna(med)

    # ------------------------------------------------------------
    # EXACT TRAINING-ONLY CV MODEL SELECTION
    # ------------------------------------------------------------

    skf = StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=SEED
    )

    cv_rows = []

    models = e1.build_models()

    for name, model in models.items():

        scores = cross_val_score(
            model,
            Xtr,
            ytr,
            cv=skf,
            scoring="f1_macro",
            n_jobs=1,
        )

        cv_rows.append({
            "model": name,
            "cv_mean": float(
                scores.mean()
            ),
            "cv_std": float(
                scores.std()
            ),
        })

    cv_df = (
        pd.DataFrame(cv_rows)
        .sort_values(
            "cv_mean",
            ascending=False
        )
        .reset_index(drop=True)
    )

    best_name = cv_df.iloc[0]["model"]

    print("\nTRAINING-ONLY CV:")
    display(cv_df)

    print(
        "\nSelected:",
        best_name
    )

    # ------------------------------------------------------------
    # FINAL FIT
    # ------------------------------------------------------------

    best_model = models[best_name]

    t0 = time.perf_counter()

    best_model.fit(
        Xtr,
        ytr
    )

    pred = best_model.predict(
        Xte
    )

    infer_seconds = (
        time.perf_counter() - t0
    )

    # ------------------------------------------------------------
    # CORE METRICS
    # ------------------------------------------------------------

    acc = accuracy_score(
        yte,
        pred
    )

    macro = f1_score(
        yte,
        pred,
        average="macro"
    )

    pr, rc, f1c, sup = (
        precision_recall_fscore_support(
            yte,
            pred,
            labels=[
                0, 1, 2
            ],
            zero_division=0
        )
    )

    cm = confusion_matrix(
        yte,
        pred,
        labels=[
            0, 1, 2
        ]
    )

    # ------------------------------------------------------------
    # AMBIGUITY STRATIFICATION
    # ------------------------------------------------------------

    ambiguity_rows = []

    amb_values = (
        test_df["ambiguous"]
        .to_numpy()
    )

    for amb in [0, 1]:

        mask = (
            amb_values == amb
        )

        if mask.sum() == 0:
            continue

        ambiguity_rows.append({
            "ambiguous": amb,
            "n": int(mask.sum()),
            "accuracy": float(
                accuracy_score(
                    yte[mask],
                    pred[mask]
                )
            ),
            "macro_f1": float(
                f1_score(
                    yte[mask],
                    pred[mask],
                    average="macro"
                )
            ),
        })

    ambiguity_df = pd.DataFrame(
        ambiguity_rows
    )

    # ------------------------------------------------------------
    # PREDICTION EVIDENCE
    # ------------------------------------------------------------

    predictions = pd.DataFrame({
        "id":
            test_df["id"].to_numpy(),

        "condition":
            condition,

        "true_id":
            yte,

        "pred_id":
            pred,

        "true_slice":
            [
                xc.SLICES[i]
                for i in yte
            ],

        "pred_slice":
            [
                xc.SLICES[i]
                for i in pred
            ],

        "ambiguous":
            test_df[
                "ambiguous"
            ].to_numpy(),

        "uc_kind":
            test_df[
                "uc_kind"
            ].to_numpy(),

        "use_case":
            test_df[
                "use_case"
            ].to_numpy(),

        "correct":
            (
                yte == pred
            ).astype(int),
    })

    # ------------------------------------------------------------
    # PER CLASS
    # ------------------------------------------------------------

    per_class = pd.DataFrame({
        "slice": xc.SLICES,
        "precision": pr,
        "recall": rc,
        "f1": f1c,
        "support": sup,
    })

    confusion = pd.DataFrame(
        cm,
        index=[
            f"true_{x}"
            for x in xc.SLICES
        ],
        columns=[
            f"pred_{x}"
            for x in xc.SLICES
        ],
    )

    # ------------------------------------------------------------
    # SAVE
    # ------------------------------------------------------------

    prefix = (
        condition
        .replace(" ", "_")
        .replace("/", "_")
    )

    cv_df.to_csv(
        R1_RESULTS /
        f"{prefix}_cv.csv",
        index=False
    )

    per_class.to_csv(
        R1_RESULTS /
        f"{prefix}_per_class.csv",
        index=False
    )

    ambiguity_df.to_csv(
        R1_RESULTS /
        f"{prefix}_ambiguity.csv",
        index=False
    )

    predictions.to_csv(
        R1_RESULTS /
        f"{prefix}_predictions.csv",
        index=False
    )

    confusion.to_csv(
        R1_RESULTS /
        f"{prefix}_confusion.csv"
    )

    summary = {
        "seed": SEED,
        "condition": condition,

        "train_n":
            int(len(train_df)),

        "test_n":
            int(len(test_df)),

        "selected_model":
            best_name,

        "selected_cv_mean":
            float(
                cv_df.iloc[0][
                    "cv_mean"
                ]
            ),

        "selected_cv_std":
            float(
                cv_df.iloc[0][
                    "cv_std"
                ]
            ),

        "accuracy":
            float(acc),

        "macro_f1":
            float(macro),

        "test_nonambiguous_n":
            int(
                (
                    test_df[
                        "ambiguous"
                    ] == 0
                ).sum()
            ),

        "test_ambiguous_n":
            int(
                (
                    test_df[
                        "ambiguous"
                    ] == 1
                ).sum()
            ),

        "fit_plus_predict_seconds":
            float(
                infer_seconds
            ),
    }

    with open(
        R1_RESULTS /
        f"{prefix}_summary.json",
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            summary,
            f,
            indent=2
        )

    print("\nSUMMARY:")
    print(json.dumps(
        summary,
        indent=2
    ))

    print("\nPER CLASS:")
    display(per_class)

    print("\nAMBIGUITY:")
    display(ambiguity_df)

    print("\nCONFUSION:")
    display(confusion)

    return summary


# ================================================================
# LOAD SEED-42 L1/L3/L4
# ================================================================

conditions = {}

for condition in [
    "L1_original",
    "L3_generic_kpi",
    "L4_kpi_only",
]:

    df = pd.read_csv(
        FINAL_COND /
        condition /
        f"seed_{SEED}.csv"
    )

    # EXACT frozen E1 70/30 split
    #
    # We use xc.split only to recover the exact row partition.
    # Feature matrices returned here are not reused.
    tmp = df.copy()
    tmp["text"] = tmp["r1_text"]

    Xtmp = xc.build_features(tmp)

    label_map = {
        name: i
        for i, name in enumerate(xc.SLICES)
    }

    ytmp = (
        df["slice"]
        .map(label_map)
        .to_numpy()
    )

    _, _, _, _, dtr, dte = xc.split(
        Xtmp,
        ytmp,
        df
    )

    conditions[condition] = (
        dtr.copy(),
        dte.copy()
    )


# ================================================================
# LOAD L2 PRECOMPUTED FAMILY SPLIT
# ================================================================

l2_train = pd.read_csv(
    L2_DIR /
    f"seed_{SEED}_train.csv"
)

l2_test = pd.read_csv(
    L2_DIR /
    f"seed_{SEED}_test.csv"
)

conditions[
    "L2_phrase_family_heldout"
] = (
    l2_train,
    l2_test
)


# ================================================================
# RUN FOUR CONDITIONS
# ================================================================

seed42_summaries = []

for condition in [
    "L1_original",
    "L2_phrase_family_heldout",
    "L3_generic_kpi",
    "L4_kpi_only",
]:

    tr, te = conditions[
        condition
    ]

    result = run_r1_condition(
        condition,
        tr,
        te
    )

    seed42_summaries.append(
        result
    )


# ================================================================
# COMPARISON
# ================================================================

seed42_comparison = pd.DataFrame(
    seed42_summaries
)

seed42_comparison.to_csv(
    R1_RESULTS /
    "seed42_condition_comparison.csv",
    index=False
)

print("\n" + "=" * 110)
print("R1 SEED-42 FOUR-CONDITION COMPARISON")
print("=" * 110)

display(
    seed42_comparison[
        [
            "condition",
            "train_n",
            "test_n",
            "selected_model",
            "selected_cv_mean",
            "accuracy",
            "macro_f1",
            "test_nonambiguous_n",
            "test_ambiguous_n",
        ]
    ]
)

print("\nSaved:")
print(
    R1_RESULTS /
    "seed42_condition_comparison.csv"
)

print("\n" + "=" * 110)
print("R1 SEED-42 FOUR-CONDITION SMOKE TEST: COMPLETE")
print("=" * 110)
```

```
# ================================================================
# STEP 28 — R1 FINAL FIVE-SEED AGGREGATION
# ================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json

SEEDS = [42, 43, 44, 45, 46]

AGG_DIR = R1 / "results" / "aggregate"
AGG_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 1. LOAD ALL FIVE SEED COMPARISON FILES
# ------------------------------------------------------------

frames = []

for seed in SEEDS:

    path = (
        R1 /
        "results" /
        f"seed_{seed}" /
        f"seed{seed}_condition_comparison.csv"
    )

    assert path.exists(), f"Missing: {path}"

    df = pd.read_csv(path)

    assert len(df) == 4, (
        f"Seed {seed}: expected 4 conditions, "
        f"found {len(df)}"
    )

    df["seed"] = seed

    frames.append(df)


all_results = pd.concat(
    frames,
    ignore_index=True
)

# ------------------------------------------------------------
# 2. BASIC HARD GATES
# ------------------------------------------------------------

expected_conditions = {
    "L1_original",
    "L2_phrase_family_heldout",
    "L3_generic_kpi",
    "L4_kpi_only",
}

assert set(all_results["condition"]) == expected_conditions

for seed in SEEDS:

    sdf = all_results[
        all_results["seed"] == seed
    ]

    assert len(sdf) == 4

    assert set(
        sdf["condition"]
    ) == expected_conditions


# ------------------------------------------------------------
# 3. L3/L4 EQUIVALENCE — ALL FIVE SEEDS
# ------------------------------------------------------------

equivalence_rows = []

for seed in SEEDS:

    sdf = all_results[
        all_results["seed"] == seed
    ]

    l3 = sdf[
        sdf["condition"] == "L3_generic_kpi"
    ].iloc[0]

    l4 = sdf[
        sdf["condition"] == "L4_kpi_only"
    ].iloc[0]

    same_model = (
        l3["selected_model"]
        == l4["selected_model"]
    )

    same_cv = np.isclose(
        l3["selected_cv_mean"],
        l4["selected_cv_mean"],
        atol=1e-12,
        rtol=0
    )

    same_acc = np.isclose(
        l3["accuracy"],
        l4["accuracy"],
        atol=1e-12,
        rtol=0
    )

    same_f1 = np.isclose(
        l3["macro_f1"],
        l4["macro_f1"],
        atol=1e-12,
        rtol=0
    )

    passed = (
        same_model
        and same_cv
        and same_acc
        and same_f1
    )

    equivalence_rows.append({
        "seed": seed,
        "same_model": same_model,
        "same_cv": same_cv,
        "same_accuracy": same_acc,
        "same_macro_f1": same_f1,
        "pass": passed,
    })


equivalence = pd.DataFrame(
    equivalence_rows
)

assert equivalence["pass"].all(), (
    "STOP: L3/L4 equivalence failed."
)


# ------------------------------------------------------------
# 4. FIVE-SEED AGGREGATION
# ------------------------------------------------------------

aggregate = (
    all_results
    .groupby(
        "condition",
        sort=False
    )
    .agg(
        n_seeds=("seed", "count"),

        accuracy_mean=("accuracy", "mean"),
        accuracy_std=("accuracy", "std"),

        macro_f1_mean=("macro_f1", "mean"),
        macro_f1_std=("macro_f1", "std"),

        cv_mean=("selected_cv_mean", "mean"),
        cv_across_seed_std=("selected_cv_mean", "std"),

        train_n_mean=("train_n", "mean"),
        test_n_mean=("test_n", "mean"),
    )
    .reset_index()
)

assert (
    aggregate["n_seeds"] == 5
).all()


# ------------------------------------------------------------
# 5. MODEL SELECTION COUNTS
# ------------------------------------------------------------

model_counts = (
    all_results
    .groupby(
        ["condition", "selected_model"]
    )
    .size()
    .reset_index(name="count")
    .sort_values(
        ["condition", "count"],
        ascending=[True, False]
    )
)


# ------------------------------------------------------------
# 6. PAIRED L1 -> L3/L4 DEGRADATION
# ------------------------------------------------------------

wide = all_results.pivot(
    index="seed",
    columns="condition",
    values=[
        "accuracy",
        "macro_f1"
    ]
)

paired_rows = []

for seed in SEEDS:

    l1_acc = wide.loc[
        seed,
        ("accuracy", "L1_original")
    ]

    l3_acc = wide.loc[
        seed,
        ("accuracy", "L3_generic_kpi")
    ]

    l1_f1 = wide.loc[
        seed,
        ("macro_f1", "L1_original")
    ]

    l3_f1 = wide.loc[
        seed,
        ("macro_f1", "L3_generic_kpi")
    ]

    paired_rows.append({
        "seed": seed,

        "L1_accuracy": l1_acc,
        "L3_L4_accuracy": l3_acc,
        "accuracy_drop": l1_acc - l3_acc,

        "L1_macro_f1": l1_f1,
        "L3_L4_macro_f1": l3_f1,
        "macro_f1_drop": l1_f1 - l3_f1,
    })


paired = pd.DataFrame(
    paired_rows
)


paired_summary = pd.DataFrame([{
    "accuracy_drop_mean":
        paired["accuracy_drop"].mean(),

    "accuracy_drop_std":
        paired["accuracy_drop"].std(),

    "macro_f1_drop_mean":
        paired["macro_f1_drop"].mean(),

    "macro_f1_drop_std":
        paired["macro_f1_drop"].std(),
}])


# ------------------------------------------------------------
# 7. L2 DESCRIPTIVE COMPARISON
#
# IMPORTANT:
# This is descriptive only.
# L2 uses a different test population from L1.
# ------------------------------------------------------------

l2_descriptive = []

for seed in SEEDS:

    sdf = all_results[
        all_results["seed"] == seed
    ]

    l1 = sdf[
        sdf["condition"] == "L1_original"
    ].iloc[0]

    l2 = sdf[
        sdf["condition"]
        == "L2_phrase_family_heldout"
    ].iloc[0]

    l2_descriptive.append({
        "seed": seed,

        "L1_accuracy":
            l1["accuracy"],

        "L2_accuracy":
            l2["accuracy"],

        "L2_minus_L1_accuracy":
            l2["accuracy"] - l1["accuracy"],

        "L1_macro_f1":
            l1["macro_f1"],

        "L2_macro_f1":
            l2["macro_f1"],

        "L2_minus_L1_macro_f1":
            l2["macro_f1"] - l1["macro_f1"],

        "L1_test_n":
            l1["test_n"],

        "L2_test_n":
            l2["test_n"],
    })


l2_descriptive = pd.DataFrame(
    l2_descriptive
)


# ------------------------------------------------------------
# 8. SAVE FINAL R1 TABLES
# ------------------------------------------------------------

all_results.to_csv(
    AGG_DIR /
    "r1_all_20_runs.csv",
    index=False
)

aggregate.to_csv(
    AGG_DIR /
    "r1_five_seed_summary.csv",
    index=False
)

model_counts.to_csv(
    AGG_DIR /
    "r1_model_selection_counts.csv",
    index=False
)

equivalence.to_csv(
    AGG_DIR /
    "r1_L3_L4_equivalence.csv",
    index=False
)

paired.to_csv(
    AGG_DIR /
    "r1_L1_vs_L3L4_paired.csv",
    index=False
)

paired_summary.to_csv(
    AGG_DIR /
    "r1_L1_vs_L3L4_summary.csv",
    index=False
)

l2_descriptive.to_csv(
    AGG_DIR /
    "r1_L2_descriptive_comparison.csv",
    index=False
)


# ------------------------------------------------------------
# 9. DISPLAY
# ------------------------------------------------------------

print("\n" + "=" * 120)
print("R1 FINAL FIVE-SEED SUMMARY")
print("=" * 120)

display(
    aggregate[
        [
            "condition",
            "n_seeds",
            "accuracy_mean",
            "accuracy_std",
            "macro_f1_mean",
            "macro_f1_std",
        ]
    ]
)


print("\n" + "=" * 120)
print("MODEL SELECTION COUNTS")
print("=" * 120)

display(model_counts)


print("\n" + "=" * 120)
print("L1 -> L3/L4 PAIRED DEGRADATION")
print("=" * 120)

display(paired)

print("\nMEAN DEGRADATION:")
display(paired_summary)


print("\n" + "=" * 120)
print("L2 DESCRIPTIVE COMPARISON — DO NOT INTERPRET AS PAIRED IMPROVEMENT")
print("=" * 120)

display(l2_descriptive)


print("\n" + "=" * 120)
print("L3/L4 FIVE-SEED EQUIVALENCE")
print("=" * 120)

display(equivalence)


print("\nSaved aggregate evidence to:")
print(AGG_DIR)

print("\n" + "=" * 120)
print("R1 FIVE-SEED AGGREGATION: PASS")
print("=" * 120)
```

```
from pathlib import Path

HANDOFF = R1 / "R1_HANDOFF_README.md"

handoff_text = r"""
# R1 — Phrase Generalization and Lexical Ablation

## Status

**PASS — CLOSED AND FROZEN**

R1 must not be rerun, modified, or overwritten unless a deliberate
new experiment version is created.

---

## Purpose

R1 addresses the reviewer concern that classification performance
may be caused primarily by memorization of synthetic class-specific
phrases or lexical cues.

The experiment evaluates phrase-family generalization and lexical
ablation while preserving the frozen implementation used for the
revision experiments.

---

## Seeds

Five independent corpus seeds were evaluated:

- 42
- 43
- 44
- 45
- 46

Total evaluated condition runs:

**5 seeds × 4 conditions = 20 runs**

---

## Experimental Conditions

### L1 — Original

Original synthetic intent text evaluated using the frozen feature
pipeline and training-only model selection.

### L2 — Phrase-family-held-out

Strict class-specific phrase/use-case family holdout.

Train and test phrase families are disjoint.

Generic use cases are excluded from the main L2 evaluation.

This condition directly tests whether the classifier collapses when
evaluated on phrase families that were absent from training.

### L3 — Generic + KPI

Class-specific use-case phrases are replaced with the neutral phrase:

`network service`

Observable KPI information is preserved.

### L4 — KPI-only

Only KPI information observable to the frozen parser is retained.

No hidden latent KPI values are introduced.

---

## Final Five-Seed Results

| Condition | Accuracy mean ± SD | Macro-F1 mean ± SD |
|---|---:|---:|
| L1 Original | 0.9311 ± 0.0065 | 0.9311 ± 0.0065 |
| L2 Phrase-family-held-out | 0.9549 ± 0.0081 | 0.9548 ± 0.0081 |
| L3 Generic + KPI | 0.8920 ± 0.0080 | 0.8920 ± 0.0078 |
| L4 KPI-only | 0.8920 ± 0.0080 | 0.8920 ± 0.0078 |

---

## Lexical Ablation Result

Mean L1 -> L3/L4 accuracy reduction:

**0.0391 ± 0.0090**

Mean L1 -> L3/L4 macro-F1 reduction:

**0.0392 ± 0.0089**

This corresponds to approximately:

**3.92 percentage points macro-F1**

---

## Main Scientific Finding

Performance does not collapse when class-specific phrase families
are held out from training.

Removing lexical/use-case information causes a consistent performance
reduction across all five seeds.

Therefore:

**Lexical intent information contributes to classification, but the
classifier is not solely dependent on memorization of class-specific
phrases.**

---

## Important Interpretation Restrictions

### L2

Do NOT claim:

> L2 improves performance over L1.

L2 and L1 use different evaluation populations because generic
examples are excluded from L2.

L2 should instead be interpreted as evidence that held-out phrase
families do not cause performance collapse.

### L3 and L4

L3 and L4 are different text interventions but produce identical
frozen model feature representations.

Their identical results are therefore an architectural /
representation finding.

They must NOT be presented as two independent model-level pieces
of evidence.

### Ambiguous intents

Synthetic ground-truth slice labels are assigned before KPI blending.

Therefore, KPI-only disagreement for ambiguous samples should not
automatically be interpreted as conventional classification error.

---

## Model Selection

Model selection was performed using training data only.

The held-out test set was not used for model selection.

Across seeds, model selection was not artificially fixed.

Gradient Boosting was selected most frequently, while Decision Tree
or Random Forest was selected for some seed/condition combinations.

---

## Reproducibility Note

During R1 preparation, a discrepancy was identified between an older
archived E1 result and reproduction using the currently packaged
source.

Direct execution of the current packaged source reproduced the new
R1 baseline exactly.

Therefore:

**The current frozen source snapshot is the authoritative
implementation for R1.**

The historical archived metric was preserved as historical evidence
and was NOT forced into the R1 results.

---

## Evidence Structure

Important directories include:

- `data/`
- `splits/`
- `results/`
- `audit/`
- `manifests/`
- `frozen_reference/`
- `final_evidence/`

Final aggregate evidence:

`results/aggregate/`

Final frozen evidence:

`final_evidence/`

---

## Key Final Files

### Aggregate

- `r1_all_20_runs.csv`
- `r1_five_seed_summary.csv`
- `r1_model_selection_counts.csv`
- `r1_L1_vs_L3L4_paired.csv`
- `r1_L1_vs_L3L4_summary.csv`
- `r1_L2_descriptive_comparison.csv`
- `r1_L3_L4_equivalence.csv`

### Final evidence

- `R1_publication_table.csv`
- `R1_final_findings.txt`
- `R1_closure.json`
- `R1_SHA256_MANIFEST.csv`

---

## Final R1 Status

**R1 STATUS: PASS — CLOSED AND FROZEN**

Do not modify R1 evidence while conducting R2 or later experiments.

Create separate experiment directories for subsequent reviewer
experiments.

---

## Next Planned Experiment

R2 — Strict Feasibility / Validity Audit

Primary objective:

Evaluate whether the current feasibility/validity mechanism is too
permissive and quantify false-feasible decisions under stricter
network-slicing constraints.

R2 should be started as a separate experiment and should not modify
R1.
"""

HANDOFF.write_text(handoff_text.strip() + "\n", encoding="utf-8")

print("Created:")
print(HANDOFF)
print("\nSize:", HANDOFF.stat().st_size, "bytes")
```

```
import pandas as pd
import hashlib

def sha256_file(path):
    h = hashlib.sha256()

    with open(path, "rb") as f:
        for block in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            h.update(block)

    return h.hexdigest()


rows = []

files = sorted(
    p for p in R1.rglob("*")
    if p.is_file()
)

print("Files to inventory:", len(files))

for i, p in enumerate(files, 1):

    rows.append({
        "relative_path": str(p.relative_to(R1)),
        "size_bytes": p.stat().st_size,
        "sha256": sha256_file(p),
    })

    if i % 50 == 0:
        print("Processed:", i)


inventory = pd.DataFrame(rows)

inventory_path = (
    R1 /
    "R1_COMPLETE_ARTIFACT_INVENTORY.csv"
)

inventory.to_csv(
    inventory_path,
    index=False
)

print("\nInventory created:")
print(inventory_path)

print("\nTotal files:", len(inventory))
print(
    "Total size:",
    round(
        inventory["size_bytes"].sum()
        / 1024 / 1024,
        2
    ),
    "MB"
)

display(inventory.head(20))
```

```
import zipfile
import hashlib
from pathlib import Path

ZIP = Path(
    "/content/"
    "XAgentic_R1_Phrase_Generalization_FINAL_2026-10-07.zip"
)

assert ZIP.exists()

with zipfile.ZipFile(ZIP, "r") as z:

    bad = z.testzip()

    names = z.namelist()

    print("ZIP entries:", len(names))

    assert bad is None, (
        f"Corrupt ZIP entry: {bad}"
    )

    required = [
        "R1_phrase_generalization/R1_HANDOFF_README.md",
        "R1_phrase_generalization/R1_EXPERIMENT_COMPLETE.json",

        "R1_phrase_generalization/"
        "final_evidence/R1_closure.json",

        "R1_phrase_generalization/"
        "final_evidence/R1_final_findings.txt",

        "R1_phrase_generalization/"
        "final_evidence/R1_publication_table.csv",

        "R1_phrase_generalization/"
        "final_evidence/R1_SHA256_MANIFEST.csv",

        "R1_phrase_generalization/"
        "results/aggregate/r1_all_20_runs.csv",

        "R1_phrase_generalization/"
        "results/aggregate/r1_five_seed_summary.csv",
    ]

    for item in required:
        assert item in names, (
            f"Missing from ZIP: {item}"
        )

print("\nZIP integrity: PASS")
print("Required final evidence: PASS")


def sha256(path):
    h = hashlib.sha256()

    with open(path, "rb") as f:
        for block in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            h.update(block)

    return h.hexdigest()


digest = sha256(ZIP)

print("\nZIP SHA-256:")
print(digest)

print("\nZIP:")
print(ZIP)

print("\n" + "=" * 100)
print("R1 ARCHIVE VERIFICATION: PASS")
print("=" * 100)
```
