```
# ==============================================================================================================
# R5-02 - RECOVER AND AUDIT FROZEN R4 EVIDENCE
# ==============================================================================================================

from pathlib import Path
import pandas as pd
import json
import hashlib
import zipfile
import shutil

print("=" * 110)
print("R5-02 - RECOVER AND AUDIT FROZEN R4 EVIDENCE")
print("=" * 110)

# --------------------------------------------------------------------------------------------------------------
# 1. Workspace
# --------------------------------------------------------------------------------------------------------------

R5_ROOT = Path("/content/R5_work")
R5_INPUT = R5_ROOT / "input"
R5_RECOVERED = R5_ROOT / "R4_recovered"
R5_AUDIT = R5_ROOT / "audit"

assert R5_ROOT.exists(), "STOP: R5 workspace missing."
assert R5_INPUT.exists(), "STOP: R5 input directory missing."

zip_files = sorted(R5_INPUT.glob("*.zip"))

assert len(zip_files) == 1, (
    f"STOP: expected exactly one R4 ZIP in {R5_INPUT}, found {len(zip_files)}"
)

r4_zip = zip_files[0]

print(f"R4 input ZIP: {r4_zip.name}")

# --------------------------------------------------------------------------------------------------------------
# 2. SHA256 utility
# --------------------------------------------------------------------------------------------------------------

def sha256_file(path, chunk_size=1024 * 1024):
    h = hashlib.sha256()

    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            h.update(chunk)

    return h.hexdigest()

# --------------------------------------------------------------------------------------------------------------
# 3. Reconfirm input ZIP identity
# --------------------------------------------------------------------------------------------------------------

input_sha = sha256_file(r4_zip)

prov_file = R5_AUDIT / "R5_R4_INPUT_PROVENANCE.json"

assert prov_file.exists(), "STOP: R5-01 provenance missing."

with open(prov_file, "r", encoding="utf-8") as f:
    prov = json.load(f)

assert input_sha == prov["sha256"], (
    "STOP: R4 ZIP SHA256 differs from the R5-01 recorded value."
)

print("PASS - R4 input ZIP unchanged since R5-01")
print("SHA256:", input_sha)

# --------------------------------------------------------------------------------------------------------------
# 4. Clean recovery directory only
# --------------------------------------------------------------------------------------------------------------

if R5_RECOVERED.exists():
    shutil.rmtree(R5_RECOVERED)

R5_RECOVERED.mkdir(parents=True, exist_ok=True)

# --------------------------------------------------------------------------------------------------------------
# 5. Safe extraction
# --------------------------------------------------------------------------------------------------------------

with zipfile.ZipFile(r4_zip, "r") as zf:

    root_resolved = R5_RECOVERED.resolve()

    for member in zf.infolist():

        target = (R5_RECOVERED / member.filename).resolve()

        if root_resolved not in target.parents and target != root_resolved:
            raise RuntimeError(
                f"STOP: unsafe ZIP member path detected: {member.filename}"
            )

    zf.extractall(R5_RECOVERED)

print("PASS - R4 package safely extracted")

# --------------------------------------------------------------------------------------------------------------
# 6. Locate recovered package root
# --------------------------------------------------------------------------------------------------------------

all_files = [p for p in R5_RECOVERED.rglob("*") if p.is_file()]

print(f"Recovered files: {len(all_files)}")

required_names = [
    "R4_FINAL_ARTIFACT_INVENTORY.csv",
    "R4_FINAL_FINDINGS_AND_INTERPRETATION.md",
    "R4_FINAL_SHA256_MANIFEST.csv",
    "R4_ORIGINAL_COMPLETED_RUN_HASHES.json",
    "R4_RECOVERED_completed_run_summary.json",
    "R4_RECOVERED_error_capture_by_uc_kind.csv",
    "R4_RECOVERED_final_system_comparison.csv",
    "R4_RECOVERED_frozen_policy_heldout_decisions.csv",
    "R4_RECOVERY_PROVENANCE.json",
    "R4_recovered_R3_manifest_audit.csv",
    "R4_runtime_recovery_summary.csv",
]

located = {}

print("\nREQUIRED R4 ARTIFACT CHECK")
print("-" * 110)

for name in required_names:

    matches = [
        p for p in all_files
        if p.name == name
    ]

    assert len(matches) == 1, (
        f"STOP: expected exactly one {name}, found {len(matches)}"
    )

    located[name] = matches[0]

    print(f"PASS {name}")

# --------------------------------------------------------------------------------------------------------------
# 7. Inspect packaged manifest
# --------------------------------------------------------------------------------------------------------------

manifest_path = located["R4_FINAL_SHA256_MANIFEST.csv"]
manifest = pd.read_csv(manifest_path)

print("\nR4 SHA256 MANIFEST")
print("-" * 110)

print("Rows   :", len(manifest))
print("Columns:", list(manifest.columns))

print("\nManifest preview:")
print(manifest.head(20).to_string(index=False))

# --------------------------------------------------------------------------------------------------------------
# 8. Automatically identify manifest filename/hash columns
# --------------------------------------------------------------------------------------------------------------

lower_map = {
    str(c).lower(): c
    for c in manifest.columns
}

filename_candidates = [
    "filename",
    "file",
    "name",
    "path",
    "relative_path",
    "artifact",
]

hash_candidates = [
    "sha256",
    "sha_256",
    "hash",
    "sha256_hash",
]

filename_col = None
hash_col = None

for c in filename_candidates:
    if c in lower_map:
        filename_col = lower_map[c]
        break

for c in hash_candidates:
    if c in lower_map:
        hash_col = lower_map[c]
        break

print("\nMANIFEST COLUMN DETECTION")
print("-" * 110)
print("Filename/path column :", filename_col)
print("SHA256 column        :", hash_col)

# --------------------------------------------------------------------------------------------------------------
# 9. Verify hashes where possible
# --------------------------------------------------------------------------------------------------------------

verified_count = 0
mismatch_count = 0
unresolved_count = 0

hash_audit_rows = []

if filename_col is not None and hash_col is not None:

    for _, row in manifest.iterrows():

        manifest_name = str(row[filename_col])
        expected_hash = str(row[hash_col]).strip().lower()

        candidate_matches = [
            p for p in all_files
            if p.name == Path(manifest_name).name
        ]

        if len(candidate_matches) == 1:

            actual_hash = sha256_file(candidate_matches[0]).lower()
            status = (
                "PASS"
                if actual_hash == expected_hash
                else "MISMATCH"
            )

            if status == "PASS":
                verified_count += 1
            else:
                mismatch_count += 1

            hash_audit_rows.append({
                "manifest_entry": manifest_name,
                "resolved_file": str(candidate_matches[0]),
                "expected_sha256": expected_hash,
                "actual_sha256": actual_hash,
                "status": status,
            })

        else:

            unresolved_count += 1

            hash_audit_rows.append({
                "manifest_entry": manifest_name,
                "resolved_file": "",
                "expected_sha256": expected_hash,
                "actual_sha256": "",
                "status": "UNRESOLVED",
            })

else:

    print(
        "NOTE - manifest columns could not be automatically interpreted; "
        "schema will be inspected manually."
    )

hash_audit = pd.DataFrame(hash_audit_rows)

hash_audit_path = R5_AUDIT / "R5_R4_MANIFEST_VERIFICATION.csv"
hash_audit.to_csv(hash_audit_path, index=False)

print("\nHASH VERIFICATION")
print("-" * 110)
print("Verified   :", verified_count)
print("Mismatch   :", mismatch_count)
print("Unresolved :", unresolved_count)

if mismatch_count:
    print("\nWARNING - manifest hash mismatches detected:")
    print(
        hash_audit[
            hash_audit["status"] == "MISMATCH"
        ].to_string(index=False)
    )

# --------------------------------------------------------------------------------------------------------------
# 10. Inspect held-out decision table
# --------------------------------------------------------------------------------------------------------------

heldout_path = located[
    "R4_RECOVERED_frozen_policy_heldout_decisions.csv"
]

heldout = pd.read_csv(heldout_path)

print("\n" + "=" * 110)
print("R4 FROZEN HELD-OUT DECISION TABLE")
print("=" * 110)

print("Rows   :", len(heldout))
print("Columns:", len(heldout.columns))

print("\nCOLUMN NAMES")
print("-" * 110)

for i, col in enumerate(heldout.columns, start=1):
    print(f"{i:02d}. {col}")

print("\nDATA TYPES")
print("-" * 110)
print(heldout.dtypes.to_string())

print("\nFIRST 10 ROWS")
print("-" * 110)

with pd.option_context(
    "display.max_columns", None,
    "display.width", 250,
    "display.max_colwidth", 80,
):
    print(heldout.head(10).to_string(index=False))

# --------------------------------------------------------------------------------------------------------------
# 11. Inspect final system comparison
# --------------------------------------------------------------------------------------------------------------

comparison_path = located[
    "R4_RECOVERED_final_system_comparison.csv"
]

comparison = pd.read_csv(comparison_path)

print("\n" + "=" * 110)
print("R4 FINAL SYSTEM COMPARISON")
print("=" * 110)

print("Rows   :", len(comparison))
print("Columns:", list(comparison.columns))

with pd.option_context(
    "display.max_columns", None,
    "display.width", 250,
):
    print(comparison.to_string(index=False))

# --------------------------------------------------------------------------------------------------------------
# 12. Inspect R4 error capture
# --------------------------------------------------------------------------------------------------------------

error_path = located[
    "R4_RECOVERED_error_capture_by_uc_kind.csv"
]

error_capture = pd.read_csv(error_path)

print("\n" + "=" * 110)
print("R4 ERROR CAPTURE BY UC KIND")
print("=" * 110)

print("Rows   :", len(error_capture))
print("Columns:", list(error_capture.columns))

with pd.option_context(
    "display.max_columns", None,
    "display.width", 250,
):
    print(error_capture.to_string(index=False))

# --------------------------------------------------------------------------------------------------------------
# 13. Inspect completed run summary JSON
# --------------------------------------------------------------------------------------------------------------

summary_path = located[
    "R4_RECOVERED_completed_run_summary.json"
]

with open(summary_path, "r", encoding="utf-8") as f:
    completed_summary = json.load(f)

print("\n" + "=" * 110)
print("R4 COMPLETED RUN SUMMARY")
print("=" * 110)

print(
    json.dumps(
        completed_summary,
        indent=2
    )
)

# --------------------------------------------------------------------------------------------------------------
# 14. Save R5 recovery map
# --------------------------------------------------------------------------------------------------------------

recovery_map = {
    name: {
        "path": str(path),
        "sha256": sha256_file(path),
        "size_bytes": path.stat().st_size,
    }
    for name, path in located.items()
}

recovery_map_path = R5_AUDIT / "R5_R4_RECOVERY_MAP.json"

with open(recovery_map_path, "w", encoding="utf-8") as f:
    json.dump(recovery_map, f, indent=2)

# --------------------------------------------------------------------------------------------------------------
# 15. Final R5-02 status
# --------------------------------------------------------------------------------------------------------------

print("\n" + "=" * 110)
print("R5-02 SUMMARY")
print("=" * 110)

print(f"Input ZIP SHA256       : {input_sha}")
print(f"Recovered files        : {len(all_files)}")
print(f"Required artifacts     : {len(located)}/{len(required_names)}")
print(f"Manifest verified      : {verified_count}")
print(f"Manifest mismatches    : {mismatch_count}")
print(f"Manifest unresolved    : {unresolved_count}")
print(f"Held-out decision rows : {len(heldout)}")
print(f"Held-out columns       : {len(heldout.columns)}")

print("\nRecovery map:")
print(recovery_map_path)

print("\nManifest audit:")
print(hash_audit_path)

print("\n" + "=" * 110)

if mismatch_count == 0:
    print("PASS - R5-02 R4 frozen evidence recovered and audited")
else:
    print("REVIEW REQUIRED - R4 evidence recovered but manifest mismatch detected")

print("=" * 110)
```

```
# ==============================================================================================================
# R5-03 - DISCOVER AND CHARACTERIZE R4 RESIDUAL HIGH-CONFIDENCE ERRORS
# ==============================================================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json
import hashlib

print("=" * 110)
print("R5-03 - DISCOVER AND CHARACTERIZE R4 RESIDUAL HIGH-CONFIDENCE ERRORS")
print("=" * 110)

# --------------------------------------------------------------------------------------------------------------
# 1. Paths
# --------------------------------------------------------------------------------------------------------------

R5_ROOT = Path("/content/R5_work")
R5_RECOVERED = R5_ROOT / "R4_recovered"
R5_RESULTS = R5_ROOT / "R5_results"
R5_AUDIT = R5_ROOT / "audit"

assert R5_ROOT.exists(), "STOP: R5 workspace missing."
assert R5_RECOVERED.exists(), "STOP: recovered R4 evidence missing."
assert R5_RESULTS.exists(), "STOP: R5 results directory missing."

# --------------------------------------------------------------------------------------------------------------
# 2. Locate frozen held-out decisions
# --------------------------------------------------------------------------------------------------------------

matches = list(
    R5_RECOVERED.rglob(
        "R4_RECOVERED_frozen_policy_heldout_decisions.csv"
    )
)

assert len(matches) == 1, (
    "STOP: expected exactly one frozen R4 held-out decision file; "
    f"found {len(matches)}"
)

heldout_path = matches[0]

print("Frozen held-out evidence:")
print(heldout_path)

# --------------------------------------------------------------------------------------------------------------
# 3. SHA256 utility
# --------------------------------------------------------------------------------------------------------------

def sha256_file(path, chunk_size=1024 * 1024):
    h = hashlib.sha256()

    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            h.update(chunk)

    return h.hexdigest()

heldout_sha = sha256_file(heldout_path)

print("SHA256:", heldout_sha)

# --------------------------------------------------------------------------------------------------------------
# 4. Load frozen evidence
# --------------------------------------------------------------------------------------------------------------

df = pd.read_csv(heldout_path)

assert len(df) == 360, (
    f"STOP: expected 360 held-out cases, found {len(df)}"
)

required_cols = [
    "position",
    "id",
    "truth",
    "baseline_pred",
    "baseline_correct",
    "strict_v2_feasible_set",
    "candidate_count",
    "p_eMBB",
    "p_URLLC",
    "p_mMTC",
    "masked_pred",
    "masked_correct",
    "uc_kind",
    "final_action",
    "final_pred",
    "r4_top1_confidence",
    "r4_top2_confidence",
    "r4_margin",
    "r4_r3_autonomous",
    "r4_r3_wrong",
    "r4_uncertainty_escalated",
    "r4_final_escalated",
    "r4_final_autonomous",
    "r4_final_correct",
    "r4_final_wrong",
]

missing = [c for c in required_cols if c not in df.columns]

assert not missing, (
    f"STOP: required columns missing: {missing}"
)

print("\nPASS - frozen held-out schema verified")

# --------------------------------------------------------------------------------------------------------------
# 5. Reconstruct R4 headline counts
# --------------------------------------------------------------------------------------------------------------

total_n = len(df)

r4_escalated_n = int(df["r4_final_escalated"].sum())
r4_autonomous_n = int(df["r4_final_autonomous"].sum())

r4_correct_n = int(
    (
        df["r4_final_autonomous"] &
        df["r4_final_correct"]
    ).sum()
)

r4_wrong_n = int(
    (
        df["r4_final_autonomous"] &
        df["r4_final_wrong"]
    ).sum()
)

print("\nR4 HEADLINE RECONSTRUCTION")
print("-" * 110)

print(f"Total held-out cases       : {total_n}")
print(f"R4 escalated               : {r4_escalated_n}")
print(f"R4 autonomous              : {r4_autonomous_n}")
print(f"R4 autonomous correct      : {r4_correct_n}")
print(f"R4 autonomous wrong        : {r4_wrong_n}")

assert total_n == 360
assert r4_escalated_n == 54
assert r4_autonomous_n == 306
assert r4_correct_n == 301
assert r4_wrong_n == 5

print("\nPASS - R4 headline metrics reproduced exactly")

# --------------------------------------------------------------------------------------------------------------
# 6. Isolate residual autonomous errors
# --------------------------------------------------------------------------------------------------------------

residual = df[
    df["r4_final_autonomous"] &
    df["r4_final_wrong"]
].copy()

assert len(residual) == 5, (
    f"STOP: expected 5 R4 residual autonomous errors, found {len(residual)}"
)

# --------------------------------------------------------------------------------------------------------------
# 7. Add diagnostic fields
# --------------------------------------------------------------------------------------------------------------

prob_cols = {
    "eMBB": "p_eMBB",
    "URLLC": "p_URLLC",
    "mMTC": "p_mMTC",
}

def truth_probability(row):
    col = prob_cols.get(row["truth"])
    if col is None:
        return np.nan
    return row[col]

residual["truth_probability"] = residual.apply(
    truth_probability,
    axis=1
)

residual["confidence_gap_vs_truth"] = (
    residual["r4_top1_confidence"] -
    residual["truth_probability"]
)

residual["high_conf_80"] = (
    residual["r4_top1_confidence"] >= 0.80
)

residual["high_conf_90"] = (
    residual["r4_top1_confidence"] >= 0.90
)

residual["high_conf_95"] = (
    residual["r4_top1_confidence"] >= 0.95
)

residual["high_margin_70"] = (
    residual["r4_margin"] >= 0.70
)

residual["truth_in_feasible_set"] = residual.apply(
    lambda r:
        str(r["truth"]) in
        str(r["strict_v2_feasible_set"]).split("|"),
    axis=1
)

residual["pred_in_feasible_set"] = residual.apply(
    lambda r:
        str(r["final_pred"]) in
        str(r["strict_v2_feasible_set"]).split("|"),
    axis=1
)

# --------------------------------------------------------------------------------------------------------------
# 8. Sort by confidence
# --------------------------------------------------------------------------------------------------------------

residual = residual.sort_values(
    ["r4_top1_confidence", "r4_margin"],
    ascending=False
).reset_index(drop=True)

# --------------------------------------------------------------------------------------------------------------
# 9. Display residual errors
# --------------------------------------------------------------------------------------------------------------

display_cols = [
    "position",
    "id",
    "uc_kind",
    "truth",
    "final_pred",
    "baseline_pred",
    "masked_pred",
    "strict_v2_feasible_set",
    "candidate_count",
    "p_eMBB",
    "p_URLLC",
    "p_mMTC",
    "truth_probability",
    "r4_top1_confidence",
    "r4_top2_confidence",
    "r4_margin",
    "confidence_gap_vs_truth",
    "truth_in_feasible_set",
    "pred_in_feasible_set",
    "high_conf_80",
    "high_conf_90",
    "high_conf_95",
    "high_margin_70",
]

print("\n" + "=" * 110)
print("R4 RESIDUAL AUTONOMOUS ERRORS")
print("=" * 110)

with pd.option_context(
    "display.max_columns", None,
    "display.width", 300,
    "display.max_colwidth", 100,
):
    print(
        residual[display_cols].to_string(
            index=False
        )
    )

# --------------------------------------------------------------------------------------------------------------
# 10. Confidence threshold analysis
# --------------------------------------------------------------------------------------------------------------

thresholds = [
    0.50,
    0.60,
    0.70,
    0.80,
    0.90,
    0.95,
    0.99,
]

threshold_rows = []

for threshold in thresholds:

    n = int(
        (
            residual["r4_top1_confidence"] >= threshold
        ).sum()
    )

    threshold_rows.append({
        "confidence_threshold": threshold,
        "residual_errors_at_or_above": n,
        "fraction_of_r4_residual_errors": n / len(residual),
    })

threshold_df = pd.DataFrame(threshold_rows)

print("\nHIGH-CONFIDENCE RESIDUAL ERROR COUNTS")
print("-" * 110)
print(threshold_df.to_string(index=False))

# --------------------------------------------------------------------------------------------------------------
# 11. Margin analysis
# --------------------------------------------------------------------------------------------------------------

margin_thresholds = [
    0.30,
    0.50,
    0.70,
    0.80,
    0.90,
]

margin_rows = []

for threshold in margin_thresholds:

    n = int(
        (
            residual["r4_margin"] >= threshold
        ).sum()
    )

    margin_rows.append({
        "margin_threshold": threshold,
        "residual_errors_at_or_above": n,
        "fraction_of_r4_residual_errors": n / len(residual),
    })

margin_df = pd.DataFrame(margin_rows)

print("\nHIGH-MARGIN RESIDUAL ERROR COUNTS")
print("-" * 110)
print(margin_df.to_string(index=False))

# --------------------------------------------------------------------------------------------------------------
# 12. Residual errors by UC kind
# --------------------------------------------------------------------------------------------------------------

uc_summary = (
    residual
    .groupby("uc_kind", dropna=False)
    .agg(
        residual_errors=("id", "count"),
        mean_top1_confidence=("r4_top1_confidence", "mean"),
        min_top1_confidence=("r4_top1_confidence", "min"),
        max_top1_confidence=("r4_top1_confidence", "max"),
        mean_margin=("r4_margin", "mean"),
        max_margin=("r4_margin", "max"),
    )
    .reset_index()
)

print("\nRESIDUAL ERRORS BY UC KIND")
print("-" * 110)
print(uc_summary.to_string(index=False))

# --------------------------------------------------------------------------------------------------------------
# 13. Truth -> prediction confusion among residual errors
# --------------------------------------------------------------------------------------------------------------

confusion = (
    residual
    .groupby(
        ["truth", "final_pred"],
        dropna=False
    )
    .size()
    .reset_index(name="count")
    .sort_values(
        "count",
        ascending=False
    )
)

print("\nRESIDUAL ERROR CONFUSIONS")
print("-" * 110)
print(confusion.to_string(index=False))

# --------------------------------------------------------------------------------------------------------------
# 14. Feasibility structure
# --------------------------------------------------------------------------------------------------------------

feasibility_summary = pd.DataFrame([
    {
        "metric": "Residual errors",
        "count": len(residual),
    },
    {
        "metric": "Truth remains in feasible set",
        "count": int(
            residual["truth_in_feasible_set"].sum()
        ),
    },
    {
        "metric": "Wrong prediction remains in feasible set",
        "count": int(
            residual["pred_in_feasible_set"].sum()
        ),
    },
    {
        "metric": "Candidate count = 1",
        "count": int(
            (residual["candidate_count"] == 1).sum()
        ),
    },
    {
        "metric": "Candidate count > 1",
        "count": int(
            (residual["candidate_count"] > 1).sum()
        ),
    },
])

print("\nFEASIBILITY STRUCTURE OF RESIDUAL ERRORS")
print("-" * 110)
print(feasibility_summary.to_string(index=False))

# --------------------------------------------------------------------------------------------------------------
# 15. Compare residual errors against correct autonomous cases
# --------------------------------------------------------------------------------------------------------------

correct_auto = df[
    df["r4_final_autonomous"] &
    df["r4_final_correct"]
].copy()

comparison_rows = []

for group_name, group_df in [
    ("R4 correct autonomous", correct_auto),
    ("R4 residual errors", residual),
]:

    comparison_rows.append({
        "group": group_name,
        "n": len(group_df),
        "mean_top1_confidence":
            group_df["r4_top1_confidence"].mean(),
        "median_top1_confidence":
            group_df["r4_top1_confidence"].median(),
        "min_top1_confidence":
            group_df["r4_top1_confidence"].min(),
        "max_top1_confidence":
            group_df["r4_top1_confidence"].max(),
        "mean_margin":
            group_df["r4_margin"].mean(),
        "median_margin":
            group_df["r4_margin"].median(),
        "min_margin":
            group_df["r4_margin"].min(),
        "max_margin":
            group_df["r4_margin"].max(),
    })

group_comparison = pd.DataFrame(comparison_rows)

print("\nCORRECT VS WRONG AUTONOMOUS CONFIDENCE PROFILE")
print("-" * 110)

print(group_comparison.to_string(index=False))

# --------------------------------------------------------------------------------------------------------------
# 16. Save evidence
# --------------------------------------------------------------------------------------------------------------

residual_path = (
    R5_RESULTS /
    "R5_r4_residual_autonomous_errors.csv"
)

threshold_path = (
    R5_RESULTS /
    "R5_residual_error_confidence_thresholds.csv"
)

margin_path = (
    R5_RESULTS /
    "R5_residual_error_margin_thresholds.csv"
)

uc_path = (
    R5_RESULTS /
    "R5_residual_errors_by_uc_kind.csv"
)

confusion_path = (
    R5_RESULTS /
    "R5_residual_error_confusions.csv"
)

comparison_path = (
    R5_RESULTS /
    "R5_correct_vs_wrong_autonomous_profile.csv"
)

residual.to_csv(
    residual_path,
    index=False
)

threshold_df.to_csv(
    threshold_path,
    index=False
)

margin_df.to_csv(
    margin_path,
    index=False
)

uc_summary.to_csv(
    uc_path,
    index=False
)

confusion.to_csv(
    confusion_path,
    index=False
)

group_comparison.to_csv(
    comparison_path,
    index=False
)

# --------------------------------------------------------------------------------------------------------------
# 17. R5-03 summary JSON
# --------------------------------------------------------------------------------------------------------------

summary = {
    "stage": "R5-03",
    "source_sha256": heldout_sha,

    "total_heldout": int(total_n),

    "r4": {
        "escalated": int(r4_escalated_n),
        "autonomous": int(r4_autonomous_n),
        "autonomous_correct": int(r4_correct_n),
        "autonomous_wrong": int(r4_wrong_n),
    },

    "residual_error_analysis": {
        "count": int(len(residual)),

        "confidence_ge_080": int(
            residual["high_conf_80"].sum()
        ),

        "confidence_ge_090": int(
            residual["high_conf_90"].sum()
        ),

        "confidence_ge_095": int(
            residual["high_conf_95"].sum()
        ),

        "margin_ge_070": int(
            residual["high_margin_70"].sum()
        ),

        "truth_in_feasible_set": int(
            residual["truth_in_feasible_set"].sum()
        ),

        "pred_in_feasible_set": int(
            residual["pred_in_feasible_set"].sum()
        ),
    },
}

summary_path = (
    R5_RESULTS /
    "R5_03_summary.json"
)

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        summary,
        f,
        indent=2
    )

# --------------------------------------------------------------------------------------------------------------
# 18. Final status
# --------------------------------------------------------------------------------------------------------------

print("\nARTIFACTS WRITTEN")
print("-" * 110)

for p in [
    residual_path,
    threshold_path,
    margin_path,
    uc_path,
    confusion_path,
    comparison_path,
    summary_path,
]:
    print(
        f"PASS {p.name:<50} "
        f"{p.stat().st_size:>10,} bytes"
    )

print("\n" + "=" * 110)
print("R5-03 SUMMARY")
print("=" * 110)

print(f"R4 autonomous cases          : {r4_autonomous_n}")
print(f"R4 correct autonomous        : {r4_correct_n}")
print(f"R4 residual autonomous errors: {len(residual)}")

print(
    "Residual errors conf >=0.80 :",
    int(residual["high_conf_80"].sum())
)

print(
    "Residual errors conf >=0.90 :",
    int(residual["high_conf_90"].sum())
)

print(
    "Residual errors conf >=0.95 :",
    int(residual["high_conf_95"].sum())
)

print(
    "Residual errors margin >=.70:",
    int(residual["high_margin_70"].sum())
)

print("\nPASS - R5-03 residual high-confidence errors discovered and characterized")
print("=" * 110)
```

```
# ==============================================================================================================
# R5-07 - DEVELOPMENT-ONLY SEMANTIC VERIFICATION SIGNAL EXPERIMENT
# ==============================================================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json
import hashlib

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.metrics import roc_auc_score

print("=" * 110)
print("R5-07 - DEVELOPMENT-ONLY SEMANTIC VERIFICATION SIGNAL EXPERIMENT")
print("=" * 110)

# --------------------------------------------------------------------------------------------------------------
# 1. Paths
# --------------------------------------------------------------------------------------------------------------

R5_ROOT = Path("/content/R5_work")
R3_RECOVERED = R5_ROOT / "R3_recovered"
R5_RESULTS = R5_ROOT / "R5_results"
R5_AUDIT = R5_ROOT / "audit"

assert R3_RECOVERED.exists(), "STOP: R3 recovery missing."
assert R5_RESULTS.exists(), "STOP: R5 results directory missing."

# --------------------------------------------------------------------------------------------------------------
# 2. Locate OOF development/calibration evidence
# --------------------------------------------------------------------------------------------------------------

matches = list(
    R3_RECOVERED.rglob(
        "R3_oof_training_calibration_rows.csv"
    )
)

assert len(matches) == 1, (
    "STOP: expected exactly one OOF calibration file."
)

source_path = matches[0]

# --------------------------------------------------------------------------------------------------------------
# 3. SHA256 utility
# --------------------------------------------------------------------------------------------------------------

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()

source_sha = sha256_file(source_path)

print("Development source:")
print(source_path)
print("SHA256:", source_sha)

# --------------------------------------------------------------------------------------------------------------
# 4. Load development data
# --------------------------------------------------------------------------------------------------------------

dev = pd.read_csv(source_path)

assert len(dev) == 840

required = [
    "id",
    "text",
    "truth",
    "uc_kind",
    "strict_v2_feasible_set",
    "candidate_count",
    "oof_pred",
    "oof_p_eMBB",
    "oof_p_URLLC",
    "oof_p_mMTC",
    "oof_correct",
    "oof_feasible_probability_mass",
    "oof_strict_valid",
]

missing = [
    c for c in required
    if c not in dev.columns
]

assert not missing, (
    f"STOP: missing columns: {missing}"
)

print("\nPASS - 840-row development schema verified")

# --------------------------------------------------------------------------------------------------------------
# 5. Guard against accidental held-out contamination
# --------------------------------------------------------------------------------------------------------------

test_matches = list(
    R3_RECOVERED.rglob(
        "R3_frozen1200_test_split.csv"
    )
)

assert len(test_matches) == 1

test_split = pd.read_csv(
    test_matches[0]
)

dev_ids = set(
    pd.to_numeric(
        dev["id"],
        errors="coerce"
    ).dropna().astype(int)
)

test_ids = set(
    pd.to_numeric(
        test_split["id"],
        errors="coerce"
    ).dropna().astype(int)
)

overlap = dev_ids.intersection(
    test_ids
)

print("\nDATA SEPARATION CHECK")
print("-" * 110)

print("Development IDs :", len(dev_ids))
print("Held-out IDs    :", len(test_ids))
print("ID overlap      :", len(overlap))

assert len(overlap) == 0, (
    "STOP: development/held-out ID overlap detected."
)

print("PASS - development and held-out IDs are disjoint")

# --------------------------------------------------------------------------------------------------------------
# 6. Normalize text
# --------------------------------------------------------------------------------------------------------------

texts = (
    dev["text"]
    .fillna("")
    .astype(str)
    .tolist()
)

labels = (
    dev["truth"]
    .astype(str)
    .to_numpy()
)

classes = [
    "eMBB",
    "URLLC",
    "mMTC",
]

assert set(labels).issubset(
    set(classes)
)

# --------------------------------------------------------------------------------------------------------------
# 7. Fit TF-IDF vocabulary using DEVELOPMENT TEXT ONLY
# --------------------------------------------------------------------------------------------------------------

vectorizer = TfidfVectorizer(
    lowercase=True,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.98,
    sublinear_tf=True,
    norm="l2",
)

X = vectorizer.fit_transform(
    texts
)

print("\nTF-IDF DEVELOPMENT MODEL")
print("-" * 110)

print("Rows       :", X.shape[0])
print("Features   :", X.shape[1])
print("N-grams    : 1-2")
print("Source     : development only")

# --------------------------------------------------------------------------------------------------------------
# 8. Create class sums for leave-one-out prototypes
# --------------------------------------------------------------------------------------------------------------

class_indices = {}

class_sums = {}

class_counts = {}

for cls in classes:

    idx = np.where(
        labels == cls
    )[0]

    class_indices[cls] = idx
    class_counts[cls] = len(idx)

    class_sums[cls] = X[idx].sum(
        axis=0
    )

print("\nCLASS COUNTS")
print("-" * 110)

for cls in classes:
    print(
        f"{cls:<8}: "
        f"{class_counts[cls]}"
    )

# --------------------------------------------------------------------------------------------------------------
# 9. Leave-one-out semantic prototype scoring
# --------------------------------------------------------------------------------------------------------------

semantic_rows = []

for i in range(len(dev)):

    row_vector = X[i]

    similarities = {}

    for cls in classes:

        class_sum = class_sums[cls]

        count = class_counts[cls]

        # Leave the current observation out of its true-class prototype.
        if labels[i] == cls:

            prototype_sum = (
                class_sum -
                row_vector
            )

            prototype_count = (
                count - 1
            )

        else:

            prototype_sum = class_sum
            prototype_count = count

        assert prototype_count > 0

        prototype = (
            prototype_sum /
            prototype_count
        )

        sim = cosine_similarity(
            row_vector,
            prototype
        )[0, 0]

        similarities[cls] = float(sim)

    ranked = sorted(
        similarities.items(),
        key=lambda x: x[1],
        reverse=True
    )

    semantic_pred = ranked[0][0]
    semantic_top1 = ranked[0][1]
    semantic_top2 = ranked[1][1]

    semantic_margin = (
        semantic_top1 -
        semantic_top2
    )

    oof_pred = str(
        dev.iloc[i]["oof_pred"]
    )

    pred_similarity = similarities.get(
        oof_pred,
        np.nan
    )

    competing = [
        v
        for cls, v in similarities.items()
        if cls != oof_pred
    ]

    best_other_similarity = max(
        competing
    )

    semantic_support_margin = (
        pred_similarity -
        best_other_similarity
    )

    semantic_disagreement = (
        semantic_pred != oof_pred
    )

    semantic_rows.append({
        "id":
            int(dev.iloc[i]["id"]),

        "truth":
            labels[i],

        "uc_kind":
            dev.iloc[i]["uc_kind"],

        "oof_pred":
            oof_pred,

        "oof_correct":
            bool(dev.iloc[i]["oof_correct"]),

        "sim_eMBB":
            similarities["eMBB"],

        "sim_URLLC":
            similarities["URLLC"],

        "sim_mMTC":
            similarities["mMTC"],

        "semantic_pred":
            semantic_pred,

        "semantic_top1_similarity":
            semantic_top1,

        "semantic_top2_similarity":
            semantic_top2,

        "semantic_margin":
            semantic_margin,

        "pred_semantic_similarity":
            pred_similarity,

        "best_other_semantic_similarity":
            best_other_similarity,

        "semantic_support_margin":
            semantic_support_margin,

        "semantic_disagreement":
            semantic_disagreement,
    })

semantic = pd.DataFrame(
    semantic_rows
)

# --------------------------------------------------------------------------------------------------------------
# 10. Attach frozen OOF / feasibility information
# --------------------------------------------------------------------------------------------------------------

keep_cols = [
    "id",
    "strict_v2_feasible_set",
    "candidate_count",
    "oof_p_eMBB",
    "oof_p_URLLC",
    "oof_p_mMTC",
    "oof_feasible_probability_mass",
    "oof_strict_valid",
]

semantic = semantic.merge(
    dev[keep_cols],
    on="id",
    how="left",
    validate="one_to_one"
)

# --------------------------------------------------------------------------------------------------------------
# 11. Calculate OOF confidence and margin
# --------------------------------------------------------------------------------------------------------------

def probability_profile(row):

    values = {
        "eMBB":
            float(row["oof_p_eMBB"]),

        "URLLC":
            float(row["oof_p_URLLC"]),

        "mMTC":
            float(row["oof_p_mMTC"]),
    }

    ordered = sorted(
        values.values(),
        reverse=True
    )

    return pd.Series({
        "oof_top1_confidence":
            ordered[0],

        "oof_top2_confidence":
            ordered[1],

        "oof_margin":
            ordered[0] - ordered[1],
    })

profiles = semantic.apply(
    probability_profile,
    axis=1
)

semantic = pd.concat(
    [
        semantic,
        profiles
    ],
    axis=1
)

# --------------------------------------------------------------------------------------------------------------
# 12. Determine prediction feasibility
# --------------------------------------------------------------------------------------------------------------

semantic["pred_in_feasible_set"] = (
    semantic.apply(
        lambda r:
            str(r["oof_pred"])
            in
            str(
                r["strict_v2_feasible_set"]
            ).split("|"),
        axis=1
    )
)

# --------------------------------------------------------------------------------------------------------------
# 13. Development error indicator
# --------------------------------------------------------------------------------------------------------------

semantic["error"] = (
    ~semantic["oof_correct"].astype(bool)
)

error_n = int(
    semantic["error"].sum()
)

correct_n = int(
    (~semantic["error"]).sum()
)

print("\nDEVELOPMENT OOF PERFORMANCE")
print("-" * 110)

print("Total       :", len(semantic))
print("Correct     :", correct_n)
print("Errors      :", error_n)
print(
    "Accuracy    :",
    round(
        correct_n / len(semantic),
        6
    )
)

# --------------------------------------------------------------------------------------------------------------
# 14. Semantic prediction accuracy
# --------------------------------------------------------------------------------------------------------------

semantic["semantic_correct"] = (
    semantic["semantic_pred"] ==
    semantic["truth"]
)

semantic_accuracy = float(
    semantic["semantic_correct"].mean()
)

print("\nSEMANTIC PROTOTYPE PERFORMANCE")
print("-" * 110)

print(
    "Semantic prototype accuracy:",
    round(
        semantic_accuracy,
        6
    )
)

# --------------------------------------------------------------------------------------------------------------
# 15. Semantic disagreement enrichment
# --------------------------------------------------------------------------------------------------------------

disagreement_summary = (
    semantic
    .groupby(
        "semantic_disagreement",
        dropna=False
    )
    .agg(
        n=("id", "count"),
        errors=("error", "sum"),
        mean_oof_confidence=(
            "oof_top1_confidence",
            "mean"
        ),
        mean_semantic_support_margin=(
            "semantic_support_margin",
            "mean"
        ),
    )
    .reset_index()
)

disagreement_summary[
    "error_rate"
] = (
    disagreement_summary["errors"] /
    disagreement_summary["n"]
)

print("\nSEMANTIC DISAGREEMENT ERROR ENRICHMENT")
print("-" * 110)

print(
    disagreement_summary.to_string(
        index=False
    )
)

# --------------------------------------------------------------------------------------------------------------
# 16. Error-detection AUC for semantic contradiction
# --------------------------------------------------------------------------------------------------------------

# Lower semantic support margin = more suspicious.
# Negate so higher score means more likely error.

auc = np.nan

if (
    semantic["error"].nunique() == 2
):

    auc = roc_auc_score(
        semantic["error"].astype(int),
        -semantic[
            "semantic_support_margin"
        ]
    )

print("\nSEMANTIC ERROR-DETECTION SIGNAL")
print("-" * 110)

print(
    "AUC (-semantic support margin):",
    round(float(auc), 6)
    if not np.isnan(auc)
    else "N/A"
)

# --------------------------------------------------------------------------------------------------------------
# 17. Quantile analysis
# --------------------------------------------------------------------------------------------------------------

quantiles = [
    0.01,
    0.025,
    0.05,
    0.10,
    0.15,
    0.20,
    0.25,
]

quantile_rows = []

support_values = semantic[
    "semantic_support_margin"
]

for q in quantiles:

    threshold = float(
        support_values.quantile(q)
    )

    flagged = (
        semantic[
            "semantic_support_margin"
        ] <= threshold
    )

    flagged_n = int(
        flagged.sum()
    )

    flagged_errors = int(
        (
            flagged &
            semantic["error"]
        ).sum()
    )

    total_errors = int(
        semantic["error"].sum()
    )

    precision = (
        flagged_errors / flagged_n
        if flagged_n
        else np.nan
    )

    recall = (
        flagged_errors / total_errors
        if total_errors
        else np.nan
    )

    correct_flagged = (
        flagged_n -
        flagged_errors
    )

    quantile_rows.append({
        "quantile":
            q,

        "threshold":
            threshold,

        "flagged_n":
            flagged_n,

        "flagged_errors":
            flagged_errors,

        "flagged_correct":
            correct_flagged,

        "verification_precision":
            precision,

        "error_detection_recall":
            recall,

        "additional_escalation_rate":
            flagged_n / len(semantic),
    })

quantile_df = pd.DataFrame(
    quantile_rows
)

print("\nDEVELOPMENT SEMANTIC THRESHOLD SWEEP")
print("-" * 110)

print(
    quantile_df.to_string(
        index=False
    )
)

# --------------------------------------------------------------------------------------------------------------
# 18. High-confidence development errors
# --------------------------------------------------------------------------------------------------------------

high_conf_errors = semantic[
    semantic["error"] &
    (
        semantic[
            "oof_top1_confidence"
        ] >= 0.80
    )
].copy()

print("\nHIGH-CONFIDENCE DEVELOPMENT ERRORS")
print("-" * 110)

print(
    "OOF errors confidence >= 0.80:",
    len(high_conf_errors)
)

if len(high_conf_errors):

    high_conf_disagreement = int(
        high_conf_errors[
            "semantic_disagreement"
        ].sum()
    )

    print(
        "Semantic disagreements among them:",
        high_conf_disagreement
    )

    print(
        "Disagreement capture rate:",
        round(
            high_conf_disagreement /
            len(high_conf_errors),
            6
        )
    )

# --------------------------------------------------------------------------------------------------------------
# 19. UC-kind breakdown
# --------------------------------------------------------------------------------------------------------------

uc_summary = (
    semantic
    .groupby(
        "uc_kind",
        dropna=False
    )
    .agg(
        n=("id", "count"),
        errors=("error", "sum"),
        semantic_disagreements=(
            "semantic_disagreement",
            "sum"
        ),
    )
    .reset_index()
)

uc_summary[
    "error_rate"
] = (
    uc_summary["errors"] /
    uc_summary["n"]
)

uc_summary[
    "semantic_disagreement_rate"
] = (
    uc_summary[
        "semantic_disagreements"
    ] /
    uc_summary["n"]
)

print("\nDEVELOPMENT RESULTS BY UC KIND")
print("-" * 110)

print(
    uc_summary.to_string(
        index=False
    )
)

# --------------------------------------------------------------------------------------------------------------
# 20. Save artifacts
# --------------------------------------------------------------------------------------------------------------

rows_path = (
    R5_RESULTS /
    "R5_dev_semantic_verification_rows.csv"
)

sweep_path = (
    R5_RESULTS /
    "R5_dev_semantic_threshold_sweep.csv"
)

disagreement_path = (
    R5_RESULTS /
    "R5_dev_semantic_disagreement_summary.csv"
)

uc_path = (
    R5_RESULTS /
    "R5_dev_semantic_by_uc_kind.csv"
)

semantic.to_csv(
    rows_path,
    index=False
)

quantile_df.to_csv(
    sweep_path,
    index=False
)

disagreement_summary.to_csv(
    disagreement_path,
    index=False
)

uc_summary.to_csv(
    uc_path,
    index=False
)

# --------------------------------------------------------------------------------------------------------------
# 21. Summary
# --------------------------------------------------------------------------------------------------------------

summary = {
    "stage":
        "R5-07",

    "method":
        "development-only leave-one-out TF-IDF class semantic prototypes",

    "source_sha256":
        source_sha,

    "development_rows":
        int(len(semantic)),

    "heldout_rows_used_for_training":
        0,

    "development_heldout_id_overlap":
        int(len(overlap)),

    "oof_errors":
        error_n,

    "semantic_prototype_accuracy":
        semantic_accuracy,

    "semantic_error_detection_auc":
        (
            float(auc)
            if not np.isnan(auc)
            else None
        ),

    "high_confidence_oof_errors_ge_080":
        int(len(high_conf_errors)),

    "high_confidence_errors_semantic_disagreement":
        int(
            high_conf_errors[
                "semantic_disagreement"
            ].sum()
        )
        if len(high_conf_errors)
        else 0,

    "policy_frozen":
        False,

    "heldout_evaluated":
        False,
}

summary_path = (
    R5_RESULTS /
    "R5_07_semantic_signal_summary.json"
)

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=2
    )

# --------------------------------------------------------------------------------------------------------------
# 22. Final
# --------------------------------------------------------------------------------------------------------------

print("\nARTIFACTS WRITTEN")
print("-" * 110)

for p in [
    rows_path,
    sweep_path,
    disagreement_path,
    uc_path,
    summary_path,
]:

    print(
        f"PASS {p.name:<52} "
        f"{p.stat().st_size:>10,} bytes"
    )

print("\n" + "=" * 110)
print("R5-07 SUMMARY")
print("=" * 110)

print("Development rows            :", len(semantic))
print("Held-out rows used          : 0")
print("OOF development errors      :", error_n)

print(
    "Semantic prototype accuracy :",
    round(
        semantic_accuracy,
        6
    )
)

print(
    "Error-detection AUC         :",
    round(float(auc), 6)
    if not np.isnan(auc)
    else "N/A"
)

print(
    "High-conf OOF errors >=.80  :",
    len(high_conf_errors)
)

print(
    "Caught by disagreement      :",
    int(
        high_conf_errors[
            "semantic_disagreement"
        ].sum()
    )
    if len(high_conf_errors)
    else 0
)

print("\nPOLICY STATUS: NOT FROZEN")
print("HELD-OUT STATUS: NOT EVALUATED")

print("\nPASS - R5-07 development-only semantic signal experiment completed")
print("=" * 110)
```

```
# ==============================================================================================================
# R5-07B - CORRECTED DEVELOPMENT-ONLY SEMANTIC VERIFICATION SIGNAL EXPERIMENT
# Fix: avoid np.matrix incompatibility in sklearn cosine_similarity
# ==============================================================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json
import hashlib

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import roc_auc_score

print("=" * 110)
print("R5-07B - CORRECTED DEVELOPMENT-ONLY SEMANTIC VERIFICATION SIGNAL EXPERIMENT")
print("=" * 110)

# --------------------------------------------------------------------------------------------------------------
# 1. Paths
# --------------------------------------------------------------------------------------------------------------

R5_ROOT = Path("/content/R5_work")
R3_RECOVERED = R5_ROOT / "R3_recovered"
R5_RESULTS = R5_ROOT / "R5_results"
R5_AUDIT = R5_ROOT / "audit"

assert R3_RECOVERED.exists(), "STOP: R3 recovery missing."
assert R5_RESULTS.exists(), "STOP: R5 results directory missing."

# --------------------------------------------------------------------------------------------------------------
# 2. Locate frozen development evidence
# --------------------------------------------------------------------------------------------------------------

matches = list(
    R3_RECOVERED.rglob(
        "R3_oof_training_calibration_rows.csv"
    )
)

assert len(matches) == 1

source_path = matches[0]

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()

source_sha = sha256_file(source_path)

print("Development source:")
print(source_path)
print("SHA256:", source_sha)

# --------------------------------------------------------------------------------------------------------------
# 3. Load development data
# --------------------------------------------------------------------------------------------------------------

dev = pd.read_csv(source_path)

assert len(dev) == 840

required = [
    "id",
    "text",
    "truth",
    "uc_kind",
    "strict_v2_feasible_set",
    "candidate_count",
    "oof_pred",
    "oof_p_eMBB",
    "oof_p_URLLC",
    "oof_p_mMTC",
    "oof_correct",
    "oof_feasible_probability_mass",
    "oof_strict_valid",
]

missing = [
    c for c in required
    if c not in dev.columns
]

assert not missing, (
    f"STOP: missing columns: {missing}"
)

print("\nPASS - 840-row development schema verified")

# --------------------------------------------------------------------------------------------------------------
# 4. Reconfirm strict development / held-out separation
# --------------------------------------------------------------------------------------------------------------

test_matches = list(
    R3_RECOVERED.rglob(
        "R3_frozen1200_test_split.csv"
    )
)

assert len(test_matches) == 1

test_split = pd.read_csv(
    test_matches[0]
)

dev_ids = set(
    pd.to_numeric(
        dev["id"],
        errors="coerce"
    ).dropna().astype(int)
)

test_ids = set(
    pd.to_numeric(
        test_split["id"],
        errors="coerce"
    ).dropna().astype(int)
)

overlap = dev_ids.intersection(
    test_ids
)

print("\nDATA SEPARATION CHECK")
print("-" * 110)

print("Development IDs :", len(dev_ids))
print("Held-out IDs    :", len(test_ids))
print("ID overlap      :", len(overlap))

assert len(overlap) == 0

print("PASS - development and held-out IDs are disjoint")

# --------------------------------------------------------------------------------------------------------------
# 5. TF-IDF using development text only
# --------------------------------------------------------------------------------------------------------------

texts = (
    dev["text"]
    .fillna("")
    .astype(str)
    .tolist()
)

labels = (
    dev["truth"]
    .astype(str)
    .to_numpy()
)

classes = [
    "eMBB",
    "URLLC",
    "mMTC",
]

assert set(labels).issubset(
    set(classes)
)

vectorizer = TfidfVectorizer(
    lowercase=True,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.98,
    sublinear_tf=True,
    norm="l2",
)

X = vectorizer.fit_transform(
    texts
)

print("\nTF-IDF DEVELOPMENT MODEL")
print("-" * 110)

print("Rows       :", X.shape[0])
print("Features   :", X.shape[1])
print("N-grams    : 1-2")
print("Source     : development only")

# --------------------------------------------------------------------------------------------------------------
# 6. Build dense class sums safely
# --------------------------------------------------------------------------------------------------------------

class_sums = {}
class_counts = {}

for cls in classes:

    idx = np.where(
        labels == cls
    )[0]

    class_counts[cls] = len(idx)

    # IMPORTANT FIX:
    # Convert sparse sum result to ndarray immediately.
    class_sums[cls] = np.asarray(
        X[idx].sum(axis=0)
    ).ravel()

print("\nCLASS COUNTS")
print("-" * 110)

for cls in classes:

    print(
        f"{cls:<8}: "
        f"{class_counts[cls]}"
    )

assert all(
    class_counts[c] == 280
    for c in classes
)

print("PASS - balanced development classes verified")

# --------------------------------------------------------------------------------------------------------------
# 7. Helper: cosine between sparse row and dense prototype
# --------------------------------------------------------------------------------------------------------------

def sparse_dense_cosine(
    sparse_row,
    dense_vector
):

    row_norm = np.sqrt(
        sparse_row.multiply(
            sparse_row
        ).sum()
    )

    dense_norm = np.linalg.norm(
        dense_vector
    )

    if row_norm == 0 or dense_norm == 0:
        return 0.0

    dot = sparse_row.dot(
        dense_vector
    )

    dot = np.asarray(
        dot
    ).ravel()[0]

    return float(
        dot /
        (
            row_norm *
            dense_norm
        )
    )

# --------------------------------------------------------------------------------------------------------------
# 8. Leave-one-out semantic prototype scoring
# --------------------------------------------------------------------------------------------------------------

semantic_rows = []

for i in range(len(dev)):

    row_vector = X[i]

    row_dense = row_vector.toarray().ravel()

    similarities = {}

    for cls in classes:

        prototype_sum = (
            class_sums[cls].copy()
        )

        prototype_count = (
            class_counts[cls]
        )

        # Leave current sample out of its TRUE class prototype.
        if labels[i] == cls:

            prototype_sum -= row_dense

            prototype_count -= 1

        assert prototype_count > 0

        prototype = (
            prototype_sum /
            prototype_count
        )

        similarities[cls] = (
            sparse_dense_cosine(
                row_vector,
                prototype
            )
        )

    ranked = sorted(
        similarities.items(),
        key=lambda x: x[1],
        reverse=True
    )

    semantic_pred = ranked[0][0]

    semantic_top1 = float(
        ranked[0][1]
    )

    semantic_top2 = float(
        ranked[1][1]
    )

    semantic_margin = (
        semantic_top1 -
        semantic_top2
    )

    oof_pred = str(
        dev.iloc[i]["oof_pred"]
    )

    pred_similarity = float(
        similarities[oof_pred]
    )

    best_other_similarity = max(
        similarities[c]
        for c in classes
        if c != oof_pred
    )

    semantic_support_margin = (
        pred_similarity -
        best_other_similarity
    )

    semantic_disagreement = (
        semantic_pred != oof_pred
    )

    semantic_rows.append({

        "id":
            int(dev.iloc[i]["id"]),

        "truth":
            labels[i],

        "uc_kind":
            dev.iloc[i]["uc_kind"],

        "oof_pred":
            oof_pred,

        "oof_correct":
            bool(dev.iloc[i]["oof_correct"]),

        "sim_eMBB":
            similarities["eMBB"],

        "sim_URLLC":
            similarities["URLLC"],

        "sim_mMTC":
            similarities["mMTC"],

        "semantic_pred":
            semantic_pred,

        "semantic_top1_similarity":
            semantic_top1,

        "semantic_top2_similarity":
            semantic_top2,

        "semantic_margin":
            semantic_margin,

        "pred_semantic_similarity":
            pred_similarity,

        "best_other_semantic_similarity":
            best_other_similarity,

        "semantic_support_margin":
            semantic_support_margin,

        "semantic_disagreement":
            semantic_disagreement,
    })

semantic = pd.DataFrame(
    semantic_rows
)

assert len(semantic) == 840

print("\nPASS - leave-one-out semantic prototype scores generated")

# --------------------------------------------------------------------------------------------------------------
# 9. Attach frozen OOF / feasibility information
# --------------------------------------------------------------------------------------------------------------

keep_cols = [
    "id",
    "strict_v2_feasible_set",
    "candidate_count",
    "oof_p_eMBB",
    "oof_p_URLLC",
    "oof_p_mMTC",
    "oof_feasible_probability_mass",
    "oof_strict_valid",
]

semantic = semantic.merge(
    dev[keep_cols],
    on="id",
    how="left",
    validate="one_to_one"
)

# --------------------------------------------------------------------------------------------------------------
# 10. OOF confidence and margin
# --------------------------------------------------------------------------------------------------------------

def probability_profile(row):

    values = [
        float(row["oof_p_eMBB"]),
        float(row["oof_p_URLLC"]),
        float(row["oof_p_mMTC"]),
    ]

    ordered = sorted(
        values,
        reverse=True
    )

    return pd.Series({

        "oof_top1_confidence":
            ordered[0],

        "oof_top2_confidence":
            ordered[1],

        "oof_margin":
            ordered[0] - ordered[1],
    })

profiles = semantic.apply(
    probability_profile,
    axis=1
)

semantic = pd.concat(
    [
        semantic,
        profiles
    ],
    axis=1
)

# --------------------------------------------------------------------------------------------------------------
# 11. Feasibility and error indicators
# --------------------------------------------------------------------------------------------------------------

semantic[
    "pred_in_feasible_set"
] = semantic.apply(

    lambda r:
        str(r["oof_pred"])
        in
        str(
            r["strict_v2_feasible_set"]
        ).split("|"),

    axis=1
)

semantic["error"] = (
    ~semantic[
        "oof_correct"
    ].astype(bool)
)

semantic[
    "semantic_correct"
] = (
    semantic[
        "semantic_pred"
    ] ==
    semantic["truth"]
)

error_n = int(
    semantic["error"].sum()
)

correct_n = int(
    (~semantic["error"]).sum()
)

# --------------------------------------------------------------------------------------------------------------
# 12. Development OOF performance
# --------------------------------------------------------------------------------------------------------------

print("\nDEVELOPMENT OOF PERFORMANCE")
print("-" * 110)

print("Total       :", len(semantic))
print("Correct     :", correct_n)
print("Errors      :", error_n)

print(
    "Accuracy    :",
    round(
        correct_n /
        len(semantic),
        6
    )
)

# --------------------------------------------------------------------------------------------------------------
# 13. Semantic prototype accuracy
# --------------------------------------------------------------------------------------------------------------

semantic_accuracy = float(
    semantic[
        "semantic_correct"
    ].mean()
)

print("\nSEMANTIC PROTOTYPE PERFORMANCE")
print("-" * 110)

print(
    "Semantic prototype accuracy:",
    round(
        semantic_accuracy,
        6
    )
)

# --------------------------------------------------------------------------------------------------------------
# 14. Semantic disagreement enrichment
# --------------------------------------------------------------------------------------------------------------

disagreement_summary = (
    semantic
    .groupby(
        "semantic_disagreement",
        dropna=False
    )
    .agg(
        n=("id", "count"),
        errors=("error", "sum"),

        mean_oof_confidence=(
            "oof_top1_confidence",
            "mean"
        ),

        mean_semantic_support_margin=(
            "semantic_support_margin",
            "mean"
        ),
    )
    .reset_index()
)

disagreement_summary[
    "error_rate"
] = (
    disagreement_summary["errors"] /
    disagreement_summary["n"]
)

print("\nSEMANTIC DISAGREEMENT ERROR ENRICHMENT")
print("-" * 110)

print(
    disagreement_summary.to_string(
        index=False
    )
)

# --------------------------------------------------------------------------------------------------------------
# 15. Error-detection AUC
# --------------------------------------------------------------------------------------------------------------

auc = np.nan

if semantic["error"].nunique() == 2:

    auc = roc_auc_score(
        semantic[
            "error"
        ].astype(int),

        -semantic[
            "semantic_support_margin"
        ]
    )

print("\nSEMANTIC ERROR-DETECTION SIGNAL")
print("-" * 110)

print(
    "AUC (-semantic support margin):",
    round(
        float(auc),
        6
    )
    if not np.isnan(auc)
    else "N/A"
)

# --------------------------------------------------------------------------------------------------------------
# 16. Development threshold sweep
# --------------------------------------------------------------------------------------------------------------

quantiles = [
    0.01,
    0.025,
    0.05,
    0.10,
    0.15,
    0.20,
    0.25,
]

quantile_rows = []

support_values = semantic[
    "semantic_support_margin"
]

total_errors = int(
    semantic["error"].sum()
)

for q in quantiles:

    threshold = float(
        support_values.quantile(q)
    )

    flagged = (
        semantic[
            "semantic_support_margin"
        ] <= threshold
    )

    flagged_n = int(
        flagged.sum()
    )

    flagged_errors = int(
        (
            flagged &
            semantic["error"]
        ).sum()
    )

    flagged_correct = (
        flagged_n -
        flagged_errors
    )

    precision = (
        flagged_errors /
        flagged_n
        if flagged_n
        else np.nan
    )

    recall = (
        flagged_errors /
        total_errors
        if total_errors
        else np.nan
    )

    quantile_rows.append({

        "quantile":
            q,

        "threshold":
            threshold,

        "flagged_n":
            flagged_n,

        "flagged_errors":
            flagged_errors,

        "flagged_correct":
            flagged_correct,

        "verification_precision":
            precision,

        "error_detection_recall":
            recall,

        "additional_escalation_rate":
            flagged_n /
            len(semantic),
    })

quantile_df = pd.DataFrame(
    quantile_rows
)

print("\nDEVELOPMENT SEMANTIC THRESHOLD SWEEP")
print("-" * 110)

print(
    quantile_df.to_string(
        index=False
    )
)

# --------------------------------------------------------------------------------------------------------------
# 17. High-confidence development errors
# --------------------------------------------------------------------------------------------------------------

high_conf_errors = semantic[
    semantic["error"] &
    (
        semantic[
            "oof_top1_confidence"
        ] >= 0.80
    )
].copy()

print("\nHIGH-CONFIDENCE DEVELOPMENT ERRORS")
print("-" * 110)

print(
    "OOF errors confidence >= 0.80:",
    len(high_conf_errors)
)

if len(high_conf_errors):

    caught = int(
        high_conf_errors[
            "semantic_disagreement"
        ].sum()
    )

    print(
        "Semantic disagreements among them:",
        caught
    )

    print(
        "Disagreement capture rate:",
        round(
            caught /
            len(high_conf_errors),
            6
        )
    )

# --------------------------------------------------------------------------------------------------------------
# 18. More stringent high-confidence groups
# --------------------------------------------------------------------------------------------------------------

hc_rows = []

for conf_threshold in [
    0.80,
    0.90,
    0.95
]:

    subset = semantic[
        semantic["error"] &
        (
            semantic[
                "oof_top1_confidence"
            ] >= conf_threshold
        )
    ]

    caught = int(
        subset[
            "semantic_disagreement"
        ].sum()
    )

    hc_rows.append({

        "confidence_threshold":
            conf_threshold,

        "high_conf_errors":
            len(subset),

        "semantic_disagreement_caught":
            caught,

        "capture_rate":
            (
                caught /
                len(subset)
                if len(subset)
                else np.nan
            ),
    })

hc_summary = pd.DataFrame(
    hc_rows
)

print("\nHIGH-CONFIDENCE ERROR CAPTURE")
print("-" * 110)

print(
    hc_summary.to_string(
        index=False
    )
)

# --------------------------------------------------------------------------------------------------------------
# 19. UC-kind breakdown
# --------------------------------------------------------------------------------------------------------------

uc_summary = (
    semantic
    .groupby(
        "uc_kind",
        dropna=False
    )
    .agg(
        n=("id", "count"),
        errors=("error", "sum"),

        semantic_disagreements=(
            "semantic_disagreement",
            "sum"
        ),
    )
    .reset_index()
)

uc_summary[
    "error_rate"
] = (
    uc_summary["errors"] /
    uc_summary["n"]
)

uc_summary[
    "semantic_disagreement_rate"
] = (
    uc_summary[
        "semantic_disagreements"
    ] /
    uc_summary["n"]
)

print("\nDEVELOPMENT RESULTS BY UC KIND")
print("-" * 110)

print(
    uc_summary.to_string(
        index=False
    )
)

# --------------------------------------------------------------------------------------------------------------
# 20. Save corrected artifacts
# --------------------------------------------------------------------------------------------------------------

rows_path = (
    R5_RESULTS /
    "R5_dev_semantic_verification_rows.csv"
)

sweep_path = (
    R5_RESULTS /
    "R5_dev_semantic_threshold_sweep.csv"
)

disagreement_path = (
    R5_RESULTS /
    "R5_dev_semantic_disagreement_summary.csv"
)

uc_path = (
    R5_RESULTS /
    "R5_dev_semantic_by_uc_kind.csv"
)

hc_path = (
    R5_RESULTS /
    "R5_dev_high_confidence_error_capture.csv"
)

semantic.to_csv(
    rows_path,
    index=False
)

quantile_df.to_csv(
    sweep_path,
    index=False
)

disagreement_summary.to_csv(
    disagreement_path,
    index=False
)

uc_summary.to_csv(
    uc_path,
    index=False
)

hc_summary.to_csv(
    hc_path,
    index=False
)

# --------------------------------------------------------------------------------------------------------------
# 21. Save summary
# --------------------------------------------------------------------------------------------------------------

summary = {

    "stage":
        "R5-07B",

    "status":
        "CORRECTED_RERUN_AFTER_NUMPY_MATRIX_COMPATIBILITY_ERROR",

    "method":
        "development-only leave-one-out TF-IDF class semantic prototypes",

    "source_sha256":
        source_sha,

    "development_rows":
        int(len(semantic)),

    "heldout_rows_used_for_training":
        0,

    "development_heldout_id_overlap":
        int(len(overlap)),

    "tfidf_features":
        int(X.shape[1]),

    "oof_errors":
        int(error_n),

    "semantic_prototype_accuracy":
        semantic_accuracy,

    "semantic_error_detection_auc":
        (
            float(auc)
            if not np.isnan(auc)
            else None
        ),

    "high_confidence_oof_errors_ge_080":
        int(len(high_conf_errors)),

    "high_confidence_errors_semantic_disagreement":
        (
            int(
                high_conf_errors[
                    "semantic_disagreement"
                ].sum()
            )
            if len(high_conf_errors)
            else 0
        ),

    "policy_frozen":
        False,

    "heldout_evaluated":
        False,
}

summary_path = (
    R5_RESULTS /
    "R5_07_semantic_signal_summary.json"
)

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=2
    )

# --------------------------------------------------------------------------------------------------------------
# 22. Final
# --------------------------------------------------------------------------------------------------------------

print("\nARTIFACTS WRITTEN")
print("-" * 110)

for p in [
    rows_path,
    sweep_path,
    disagreement_path,
    uc_path,
    hc_path,
    summary_path,
]:

    print(
        f"PASS {p.name:<55} "
        f"{p.stat().st_size:>10,} bytes"
    )

print("\n" + "=" * 110)
print("R5-07B SUMMARY")
print("=" * 110)

print(
    "Development rows            :",
    len(semantic)
)

print(
    "Held-out rows used          :",
    0
)

print(
    "OOF development errors      :",
    error_n
)

print(
    "Semantic prototype accuracy :",
    round(
        semantic_accuracy,
        6
    )
)

print(
    "Error-detection AUC         :",
    round(
        float(auc),
        6
    )
    if not np.isnan(auc)
    else "N/A"
)

print(
    "High-conf OOF errors >=.80  :",
    len(high_conf_errors)
)

print(
    "Caught by disagreement      :",
    (
        int(
            high_conf_errors[
                "semantic_disagreement"
            ].sum()
        )
        if len(high_conf_errors)
        else 0
    )
)

print("\nPOLICY STATUS: NOT FROZEN")
print("HELD-OUT STATUS: NOT EVALUATED")

print("\nPASS - R5-07B corrected semantic signal experiment completed")
print("=" * 110)
```

```
# ==============================================================================================================
# R5-08 - CALIBRATE SEMANTIC VERIFIER ON FROZEN R4-SURVIVING DEVELOPMENT CASES
# ==============================================================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json
import hashlib

print("=" * 110)
print("R5-08 - CALIBRATE SEMANTIC VERIFIER ON FROZEN R4-SURVIVING DEVELOPMENT CASES")
print("=" * 110)

R5_ROOT = Path("/content/R5_work")
R5_RESULTS = R5_ROOT / "R5_results"
R5_AUDIT = R5_ROOT / "audit"

ROWS_PATH = (
    R5_RESULTS /
    "R5_dev_semantic_verification_rows.csv"
)

assert ROWS_PATH.exists(), (
    "STOP: R5-07B semantic rows missing."
)

# --------------------------------------------------------------------------------------------------------------
# 1. Frozen upstream thresholds
# --------------------------------------------------------------------------------------------------------------

R3_TAU = 0.5486315987262682
R4_MARGIN_THRESHOLD = 0.7097637192002952

print("\nFROZEN UPSTREAM POLICY")
print("-" * 110)
print("R3 feasible-mass tau :", R3_TAU)
print("R4 margin threshold  :", R4_MARGIN_THRESHOLD)

# --------------------------------------------------------------------------------------------------------------
# 2. Load development semantic evidence
# --------------------------------------------------------------------------------------------------------------

dev = pd.read_csv(ROWS_PATH)

assert len(dev) == 840

required = [
    "id",
    "truth",
    "uc_kind",
    "oof_pred",
    "oof_correct",
    "oof_feasible_probability_mass",
    "oof_strict_valid",
    "oof_top1_confidence",
    "oof_margin",
    "semantic_pred",
    "semantic_support_margin",
    "semantic_disagreement",
]

missing = [
    c for c in required
    if c not in dev.columns
]

assert not missing, (
    f"STOP: missing R5-07B columns: {missing}"
)

print("\nPASS - R5-07B development evidence loaded")
print("Rows:", len(dev))

# --------------------------------------------------------------------------------------------------------------
# 3. Reconstruct frozen R3 decision
#
# R3 escalates when feasible probability mass is below frozen tau.
# Preserve any invalid strict decision conservatively as escalation.
# --------------------------------------------------------------------------------------------------------------

dev["r3_escalated"] = (
    (dev["oof_feasible_probability_mass"] < R3_TAU)
    |
    (~dev["oof_strict_valid"].astype(bool))
)

dev["r3_autonomous"] = (
    ~dev["r3_escalated"]
)

# --------------------------------------------------------------------------------------------------------------
# 4. Apply frozen R4 uncertainty policy
#
# R4 only considers R3-autonomous cases.
# Low probability margin -> additional uncertainty escalation.
# --------------------------------------------------------------------------------------------------------------

dev["r4_uncertainty_escalated"] = (
    dev["r3_autonomous"]
    &
    (
        dev["oof_margin"] <
        R4_MARGIN_THRESHOLD
    )
)

dev["r4_final_escalated"] = (
    dev["r3_escalated"]
    |
    dev["r4_uncertainty_escalated"]
)

dev["r4_final_autonomous"] = (
    ~dev["r4_final_escalated"]
)

dev["r4_final_wrong"] = (
    dev["r4_final_autonomous"]
    &
    (~dev["oof_correct"].astype(bool))
)

dev["r4_final_correct"] = (
    dev["r4_final_autonomous"]
    &
    dev["oof_correct"].astype(bool)
)

# --------------------------------------------------------------------------------------------------------------
# 5. Frozen R4 development reconstruction
# --------------------------------------------------------------------------------------------------------------

total = len(dev)

r3_auto = int(
    dev["r3_autonomous"].sum()
)

r3_esc = int(
    dev["r3_escalated"].sum()
)

r4_auto = int(
    dev["r4_final_autonomous"].sum()
)

r4_esc = int(
    dev["r4_final_escalated"].sum()
)

r4_errors = int(
    dev["r4_final_wrong"].sum()
)

r4_correct = int(
    dev["r4_final_correct"].sum()
)

print("\nFROZEN R4 DEVELOPMENT RECONSTRUCTION")
print("-" * 110)

print("Total observations          :", total)
print("R3 autonomous               :", r3_auto)
print("R3 escalated                :", r3_esc)
print("R4 autonomous               :", r4_auto)
print("R4 escalated                :", r4_esc)
print("R4 autonomous correct       :", r4_correct)
print("R4 autonomous errors        :", r4_errors)

if r4_auto:
    print(
        "R4 autonomous accuracy      :",
        round(
            r4_correct / r4_auto,
            6
        )
    )

    print(
        "R4 selective risk           :",
        round(
            r4_errors / r4_auto,
            6
        )
    )

# --------------------------------------------------------------------------------------------------------------
# 6. Isolate only cases reaching R5
# --------------------------------------------------------------------------------------------------------------

r5_input = dev[
    dev["r4_final_autonomous"]
].copy()

assert len(r5_input) == r4_auto

print("\nR5 DEVELOPMENT INPUT")
print("-" * 110)

print(
    "Cases reaching R5 verifier  :",
    len(r5_input)
)

print(
    "Residual R4 errors          :",
    int(
        (~r5_input["oof_correct"].astype(bool)).sum()
    )
)

print(
    "Residual R4 correct cases   :",
    int(
        r5_input["oof_correct"].astype(bool).sum()
    )
)

# --------------------------------------------------------------------------------------------------------------
# 7. Semantic signal among residual R4 errors
# --------------------------------------------------------------------------------------------------------------

r5_input["error"] = (
    ~r5_input["oof_correct"].astype(bool)
)

residual_errors = r5_input[
    r5_input["error"]
]

semantic_disagreement_errors = int(
    residual_errors[
        "semantic_disagreement"
    ].sum()
)

print("\nSEMANTIC SIGNAL AMONG R4 RESIDUAL ERRORS")
print("-" * 110)

print(
    "Residual R4 errors          :",
    len(residual_errors)
)

print(
    "Semantic disagreement       :",
    semantic_disagreement_errors
)

if len(residual_errors):

    print(
        "Disagreement capture rate  :",
        round(
            semantic_disagreement_errors /
            len(residual_errors),
            6
        )
    )

# --------------------------------------------------------------------------------------------------------------
# 8. Candidate semantic-support thresholds
#
# Thresholds are DEVELOPMENT-ONLY quantiles.
# More negative semantic_support_margin = stronger contradiction.
# --------------------------------------------------------------------------------------------------------------

quantiles = [
    0.01,
    0.025,
    0.05,
    0.075,
    0.10,
    0.125,
    0.15,
    0.175,
    0.20,
    0.25,
]

threshold_rows = []

support = r5_input[
    "semantic_support_margin"
]

total_r5_errors = int(
    r5_input["error"].sum()
)

total_r5_correct = int(
    (~r5_input["error"]).sum()
)

for q in quantiles:

    threshold = float(
        support.quantile(q)
    )

    verification_flag = (
        r5_input[
            "semantic_support_margin"
        ] <= threshold
    )

    flagged_n = int(
        verification_flag.sum()
    )

    captured_errors = int(
        (
            verification_flag
            &
            r5_input["error"]
        ).sum()
    )

    correct_escalated = int(
        (
            verification_flag
            &
            (~r5_input["error"])
        ).sum()
    )

    remaining_auto = (
        len(r5_input) -
        flagged_n
    )

    remaining_errors = (
        total_r5_errors -
        captured_errors
    )

    remaining_correct = (
        total_r5_correct -
        correct_escalated
    )

    precision = (
        captured_errors / flagged_n
        if flagged_n
        else np.nan
    )

    error_recall = (
        captured_errors / total_r5_errors
        if total_r5_errors
        else np.nan
    )

    false_positive_rate = (
        correct_escalated /
        total_r5_correct
        if total_r5_correct
        else np.nan
    )

    coverage = (
        remaining_auto / total
    )

    autonomous_accuracy = (
        remaining_correct /
        remaining_auto
        if remaining_auto
        else np.nan
    )

    selective_risk = (
        remaining_errors /
        remaining_auto
        if remaining_auto
        else np.nan
    )

    correct_preservation = (
        remaining_correct /
        total_r5_correct
        if total_r5_correct
        else np.nan
    )

    threshold_rows.append({

        "quantile":
            q,

        "semantic_support_threshold":
            threshold,

        "r5_input_n":
            len(r5_input),

        "r4_residual_errors":
            total_r5_errors,

        "verification_flagged":
            flagged_n,

        "errors_captured":
            captured_errors,

        "correct_escalated":
            correct_escalated,

        "verification_precision":
            precision,

        "error_detection_recall":
            error_recall,

        "verification_false_positive_rate":
            false_positive_rate,

        "remaining_autonomous":
            remaining_auto,

        "remaining_autonomous_errors":
            remaining_errors,

        "remaining_autonomous_correct":
            remaining_correct,

        "final_coverage_over_840":
            coverage,

        "final_autonomous_accuracy":
            autonomous_accuracy,

        "final_selective_risk":
            selective_risk,

        "correct_autonomous_preservation":
            correct_preservation,

        "additional_escalation_rate_over_r5_input":
            flagged_n /
            len(r5_input),
    })

sweep = pd.DataFrame(
    threshold_rows
)

print("\nR5 DEVELOPMENT THRESHOLD SWEEP")
print("-" * 110)

with pd.option_context(
    "display.max_columns", None,
    "display.width", 350
):
    print(
        sweep.to_string(
            index=False
        )
    )

# --------------------------------------------------------------------------------------------------------------
# 9. Pure semantic-disagreement operating point
# --------------------------------------------------------------------------------------------------------------

flag = (
    r5_input[
        "semantic_disagreement"
    ].astype(bool)
)

disagreement_flagged = int(
    flag.sum()
)

disagreement_captured = int(
    (
        flag &
        r5_input["error"]
    ).sum()
)

disagreement_correct = int(
    (
        flag &
        (~r5_input["error"])
    ).sum()
)

remaining_auto = (
    len(r5_input) -
    disagreement_flagged
)

remaining_errors = (
    total_r5_errors -
    disagreement_captured
)

remaining_correct = (
    total_r5_correct -
    disagreement_correct
)

disagreement_policy = {
    "policy":
        "semantic_disagreement",

    "verification_flagged":
        disagreement_flagged,

    "errors_captured":
        disagreement_captured,

    "correct_escalated":
        disagreement_correct,

    "verification_precision":
        (
            disagreement_captured /
            disagreement_flagged
            if disagreement_flagged
            else None
        ),

    "error_detection_recall":
        (
            disagreement_captured /
            total_r5_errors
            if total_r5_errors
            else None
        ),

    "final_coverage_over_840":
        remaining_auto / total,

    "final_autonomous_accuracy":
        (
            remaining_correct /
            remaining_auto
            if remaining_auto
            else None
        ),

    "final_selective_risk":
        (
            remaining_errors /
            remaining_auto
            if remaining_auto
            else None
        ),

    "correct_autonomous_preservation":
        (
            remaining_correct /
            total_r5_correct
            if total_r5_correct
            else None
        ),
}

print("\nPURE SEMANTIC-DISAGREEMENT POLICY")
print("-" * 110)

for k, v in disagreement_policy.items():
    print(f"{k:<35}: {v}")

# --------------------------------------------------------------------------------------------------------------
# 10. Rank candidate thresholds
#
# Development-only ranking:
#   1. maximize captured residual errors
#   2. minimize correct escalations
#   3. minimize added escalation
#
# This does NOT freeze the policy yet.
# --------------------------------------------------------------------------------------------------------------

ranked = sweep.sort_values(
    by=[
        "errors_captured",
        "correct_escalated",
        "verification_flagged",
    ],
    ascending=[
        False,
        True,
        True,
    ]
).reset_index(drop=True)

print("\nDEVELOPMENT-ONLY CANDIDATE RANKING")
print("-" * 110)

with pd.option_context(
    "display.max_columns", None,
    "display.width", 350
):
    print(
        ranked.head(10).to_string(
            index=False
        )
    )

# --------------------------------------------------------------------------------------------------------------
# 11. Save artifacts
# --------------------------------------------------------------------------------------------------------------

r5_input_path = (
    R5_RESULTS /
    "R5_dev_R4_surviving_cases.csv"
)

sweep_path = (
    R5_RESULTS /
    "R5_dev_R4_survivor_semantic_threshold_sweep.csv"
)

ranked_path = (
    R5_RESULTS /
    "R5_dev_R4_survivor_candidate_ranking.csv"
)

r5_input.to_csv(
    r5_input_path,
    index=False
)

sweep.to_csv(
    sweep_path,
    index=False
)

ranked.to_csv(
    ranked_path,
    index=False
)

# --------------------------------------------------------------------------------------------------------------
# 12. Summary
# --------------------------------------------------------------------------------------------------------------

summary = {

    "stage":
        "R5-08",

    "r3_tau":
        R3_TAU,

    "r4_margin_threshold":
        R4_MARGIN_THRESHOLD,

    "development_total":
        total,

    "r3_autonomous":
        r3_auto,

    "r3_escalated":
        r3_esc,

    "r4_autonomous":
        r4_auto,

    "r4_escalated":
        r4_esc,

    "r4_autonomous_correct":
        r4_correct,

    "r4_autonomous_errors":
        r4_errors,

    "semantic_disagreement_residual_errors":
        semantic_disagreement_errors,

    "semantic_disagreement_policy":
        disagreement_policy,

    "policy_frozen":
        False,

    "heldout_evaluated":
        False,
}

summary_path = (
    R5_RESULTS /
    "R5_08_R4_survivor_calibration_summary.json"
)

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=2
    )

print("\nARTIFACTS WRITTEN")
print("-" * 110)

for p in [
    r5_input_path,
    sweep_path,
    ranked_path,
    summary_path,
]:

    print(
        f"PASS {p.name:<60} "
        f"{p.stat().st_size:>10,} bytes"
    )

print("\n" + "=" * 110)
print("R5-08 SUMMARY")
print("=" * 110)

print("Development total          :", total)
print("R4 autonomous              :", r4_auto)
print("R4 residual errors         :", r4_errors)
print(
    "Semantic disagreements on errors:",
    semantic_disagreement_errors
)

print("\nPOLICY STATUS: NOT YET FROZEN")
print("HELD-OUT STATUS: NOT EVALUATED")

print("\nPASS - R5-08 R4-survivor semantic calibration completed")
print("=" * 110)
```

```
# ==============================================================================================================
# R5-09 - FREEZE SEMANTIC VERIFICATION POLICY
# ==============================================================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import hashlib
import json
import datetime

print("=" * 110)
print("R5-09 - FREEZE SEMANTIC VERIFICATION POLICY")
print("=" * 110)

R5_ROOT = Path("/content/R5_work")
R5_RESULTS = R5_ROOT / "R5_results"
R5_AUDIT = R5_ROOT / "audit"

SWEEP_PATH = (
    R5_RESULTS /
    "R5_dev_R4_survivor_semantic_threshold_sweep.csv"
)

SEMANTIC_ROWS_PATH = (
    R5_RESULTS /
    "R5_dev_semantic_verification_rows.csv"
)

R4_SURVIVOR_PATH = (
    R5_RESULTS /
    "R5_dev_R4_surviving_cases.csv"
)

assert SWEEP_PATH.exists()
assert SEMANTIC_ROWS_PATH.exists()
assert R4_SURVIVOR_PATH.exists()

# --------------------------------------------------------------------------------------------------------------
# 1. Utility
# --------------------------------------------------------------------------------------------------------------

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()

# --------------------------------------------------------------------------------------------------------------
# 2. Load development-only calibration sweep
# --------------------------------------------------------------------------------------------------------------

sweep = pd.read_csv(
    SWEEP_PATH
)

assert len(sweep) > 0

# --------------------------------------------------------------------------------------------------------------
# 3. Freeze selected operating point
#
# Selection rationale:
# 7.5% quantile provides the strongest conservative trade-off:
# - captures 8/13 residual R4 errors
# - only 53/704 additional escalations
# - preserves 93.49% of correct R4 autonomous cases
# - reduces development selective risk from 1.8466% to 0.7680%
#
# IMPORTANT:
# No frozen held-out labels/results are used here.
# --------------------------------------------------------------------------------------------------------------

TARGET_QUANTILE = 0.075

selected = sweep[
    np.isclose(
        sweep["quantile"],
        TARGET_QUANTILE
    )
]

assert len(selected) == 1, (
    "STOP: selected development operating point missing."
)

selected = selected.iloc[0]

FROZEN_THRESHOLD = float(
    selected[
        "semantic_support_threshold"
    ]
)

print("\nSELECTED DEVELOPMENT OPERATING POINT")
print("-" * 110)

print("Quantile                       :", TARGET_QUANTILE)
print("Semantic support threshold     :", FROZEN_THRESHOLD)
print("Verification flagged           :", int(selected["verification_flagged"]))
print("Residual errors captured       :", int(selected["errors_captured"]))
print("Correct cases escalated        :", int(selected["correct_escalated"]))

print(
    "Error detection recall         :",
    float(selected["error_detection_recall"])
)

print(
    "Verification precision         :",
    float(selected["verification_precision"])
)

print(
    "Correct autonomous preservation:",
    float(selected["correct_autonomous_preservation"])
)

print(
    "Final development coverage     :",
    float(selected["final_coverage_over_840"])
)

print(
    "Final autonomous accuracy      :",
    float(selected["final_autonomous_accuracy"])
)

print(
    "Final selective risk           :",
    float(selected["final_selective_risk"])
)

# --------------------------------------------------------------------------------------------------------------
# 4. Explicit methodological guard
# --------------------------------------------------------------------------------------------------------------

assert TARGET_QUANTILE == 0.075

assert int(
    selected["r4_residual_errors"]
) == 13

assert int(
    selected["errors_captured"]
) == 8

assert int(
    selected["verification_flagged"]
) == 53

assert int(
    selected["correct_escalated"]
) == 45

print("\nPASS - expected development-only operating point verified")

# --------------------------------------------------------------------------------------------------------------
# 5. Freeze complete R5 policy
# --------------------------------------------------------------------------------------------------------------

policy = {

    "policy_name":
        "R5_semantic_verification_v1",

    "status":
        "FROZEN_BEFORE_HELDOUT_EVALUATION",

    "stage":
        "R5-09",

    "upstream_policy": {

        "r3_feasible_mass_threshold":
            0.5486315987262682,

        "r4_margin_threshold":
            0.7097637192002952,
    },

    "semantic_verifier": {

        "representation":
            "TF-IDF",

        "ngram_range":
            [1, 2],

        "min_df":
            2,

        "max_df":
            0.98,

        "sublinear_tf":
            True,

        "normalization":
            "l2",

        "prototype_method":
            "class_mean_TFIDF",

        "development_scoring":
            "leave_one_out_class_prototypes",

        "deployment_scoring":
            "full_development_class_prototypes",

        "verification_score":
            "predicted_class_similarity_minus_best_other_class_similarity",

        "escalation_rule":
            "semantic_support_margin <= frozen_threshold",

        "frozen_semantic_support_threshold":
            FROZEN_THRESHOLD,

        "selected_development_quantile":
            TARGET_QUANTILE,
    },

    "development_selection_evidence": {

        "development_total":
            840,

        "r4_autonomous":
            704,

        "r4_residual_errors":
            13,

        "verification_flagged":
            int(
                selected[
                    "verification_flagged"
                ]
            ),

        "errors_captured":
            int(
                selected[
                    "errors_captured"
                ]
            ),

        "correct_escalated":
            int(
                selected[
                    "correct_escalated"
                ]
            ),

        "error_detection_recall":
            float(
                selected[
                    "error_detection_recall"
                ]
            ),

        "verification_precision":
            float(
                selected[
                    "verification_precision"
                ]
            ),

        "correct_autonomous_preservation":
            float(
                selected[
                    "correct_autonomous_preservation"
                ]
            ),

        "final_coverage_over_840":
            float(
                selected[
                    "final_coverage_over_840"
                ]
            ),

        "final_autonomous_accuracy":
            float(
                selected[
                    "final_autonomous_accuracy"
                ]
            ),

        "final_selective_risk":
            float(
                selected[
                    "final_selective_risk"
                ]
            ),
    },

    "selection_rationale":
        (
            "The 7.5% development quantile was selected as a conservative "
            "risk-coverage operating point. It captures 8 of 13 residual "
            "R4 development errors while escalating 53 of 704 R4-autonomous "
            "development cases and preserving approximately 93.49% of "
            "correct R4 autonomous decisions. More aggressive thresholds "
            "capture at most one additional development error while causing "
            "substantially more correct-case escalation."
        ),

    "data_governance": {

        "development_rows":
            840,

        "heldout_rows":
            360,

        "development_heldout_id_overlap":
            0,

        "heldout_labels_used_for_threshold_selection":
            False,

        "heldout_results_used_for_threshold_selection":
            False,

        "post_heldout_retuning_allowed":
            False,
    },
}

# --------------------------------------------------------------------------------------------------------------
# 6. Evidence hashes
# --------------------------------------------------------------------------------------------------------------

policy[
    "development_evidence_sha256"
] = {

    "semantic_rows":
        sha256_file(
            SEMANTIC_ROWS_PATH
        ),

    "r4_surviving_cases":
        sha256_file(
            R4_SURVIVOR_PATH
        ),

    "threshold_sweep":
        sha256_file(
            SWEEP_PATH
        ),
}

# --------------------------------------------------------------------------------------------------------------
# 7. Write canonical policy JSON
# --------------------------------------------------------------------------------------------------------------

POLICY_PATH = (
    R5_RESULTS /
    "R5_FROZEN_SEMANTIC_VERIFICATION_POLICY.json"
)

with open(
    POLICY_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        policy,
        f,
        indent=2,
        sort_keys=True
    )

policy_sha = sha256_file(
    POLICY_PATH
)

# --------------------------------------------------------------------------------------------------------------
# 8. Write freeze record
# --------------------------------------------------------------------------------------------------------------

freeze_record = {

    "stage":
        "R5-09",

    "freeze_status":
        "PASS",

    "policy_file":
        str(POLICY_PATH),

    "policy_sha256":
        policy_sha,

    "frozen_semantic_support_threshold":
        FROZEN_THRESHOLD,

    "selected_quantile":
        TARGET_QUANTILE,

    "heldout_evaluated_before_freeze":
        False,

    "heldout_used_for_selection":
        False,

    "retuning_after_heldout":
        "PROHIBITED",

    "note":
        (
            "R5 semantic verifier frozen using development evidence only. "
            "The next stage may evaluate this immutable policy exactly once "
            "on the frozen 360-row held-out set."
        ),
}

FREEZE_PATH = (
    R5_AUDIT /
    "R5_09_POLICY_FREEZE_RECORD.json"
)

with open(
    FREEZE_PATH,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        freeze_record,
        f,
        indent=2,
        sort_keys=True
    )

freeze_sha = sha256_file(
    FREEZE_PATH
)

# --------------------------------------------------------------------------------------------------------------
# 9. Re-read and verify immutability identity
# --------------------------------------------------------------------------------------------------------------

with open(
    POLICY_PATH,
    "r",
    encoding="utf-8"
) as f:

    check_policy = json.load(f)

assert (
    float(
        check_policy[
            "semantic_verifier"
        ][
            "frozen_semantic_support_threshold"
        ]
    )
    ==
    FROZEN_THRESHOLD
)

assert (
    check_policy[
        "data_governance"
    ][
        "heldout_labels_used_for_threshold_selection"
    ]
    is False
)

assert (
    check_policy[
        "data_governance"
    ][
        "post_heldout_retuning_allowed"
    ]
    is False
)

assert sha256_file(
    POLICY_PATH
) == policy_sha

print("\n" + "=" * 110)
print("R5 POLICY FREEZE")
print("=" * 110)

print("Policy status       : FROZEN")
print("Selected quantile   :", TARGET_QUANTILE)
print("Frozen threshold    :", FROZEN_THRESHOLD)
print("Policy SHA256       :", policy_sha)
print("Freeze record SHA256:", freeze_sha)

print("\nPolicy file:")
print(POLICY_PATH)

print("\nFreeze record:")
print(FREEZE_PATH)

print("\nHELD-OUT EVALUATED: NO")
print("POST-HELDOUT RETUNING: PROHIBITED")

print("\nPASS - R5-09 semantic verification policy frozen")
print("=" * 110)
```

```
# ==============================================================================================================
# R5-10 - FIRST AND ONLY FROZEN HELD-OUT EVALUATION
# ==============================================================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json
import hashlib

from sklearn.feature_extraction.text import TfidfVectorizer

print("=" * 110)
print("R5-10 - FIRST AND ONLY FROZEN HELD-OUT EVALUATION")
print("=" * 110)

R5_ROOT = Path("/content/R5_work")
R3_RECOVERED = R5_ROOT / "R3_recovered"
R4_RECOVERED = R5_ROOT / "R4_recovered"
R5_RESULTS = R5_ROOT / "R5_results"
R5_AUDIT = R5_ROOT / "audit"

POLICY_PATH = (
    R5_RESULTS /
    "R5_FROZEN_SEMANTIC_VERIFICATION_POLICY.json"
)

EXPECTED_POLICY_SHA = (
    "b8e5ef13e6895fcfb3656f7257c94727"
    "e3cea17d38f08563436d55b6cf0ba243"
)

# --------------------------------------------------------------------------------------------------------------
# 1. Utility
# --------------------------------------------------------------------------------------------------------------

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()

# --------------------------------------------------------------------------------------------------------------
# 2. Verify immutable frozen policy
# --------------------------------------------------------------------------------------------------------------

assert POLICY_PATH.exists(), (
    "STOP: frozen R5 policy missing."
)

actual_policy_sha = sha256_file(
    POLICY_PATH
)

assert actual_policy_sha == EXPECTED_POLICY_SHA, (
    "STOP: R5 policy SHA256 changed."
)

with open(
    POLICY_PATH,
    "r",
    encoding="utf-8"
) as f:

    policy = json.load(f)

FROZEN_THRESHOLD = float(
    policy[
        "semantic_verifier"
    ][
        "frozen_semantic_support_threshold"
    ]
)

R3_TAU = float(
    policy[
        "upstream_policy"
    ][
        "r3_feasible_mass_threshold"
    ]
)

R4_MARGIN_THRESHOLD = float(
    policy[
        "upstream_policy"
    ][
        "r4_margin_threshold"
    ]
)

print("\nFROZEN POLICY VERIFIED")
print("-" * 110)
print("Policy SHA256      :", actual_policy_sha)
print("R3 tau             :", R3_TAU)
print("R4 margin threshold:", R4_MARGIN_THRESHOLD)
print("R5 threshold       :", FROZEN_THRESHOLD)

# --------------------------------------------------------------------------------------------------------------
# 3. Load DEVELOPMENT data for deployment prototypes
#
# IMPORTANT:
# Vectorizer and semantic prototypes are fitted ONLY from the 840 development rows.
# --------------------------------------------------------------------------------------------------------------

dev_matches = list(
    R3_RECOVERED.rglob(
        "R3_oof_training_calibration_rows.csv"
    )
)

assert len(dev_matches) == 1

dev = pd.read_csv(
    dev_matches[0]
)

assert len(dev) == 840

required_dev = [
    "id",
    "text",
    "truth",
]

assert all(
    c in dev.columns
    for c in required_dev
)

# --------------------------------------------------------------------------------------------------------------
# 4. Load frozen R4 held-out decisions
# --------------------------------------------------------------------------------------------------------------

r4_matches = list(
    R4_RECOVERED.rglob(
        "R4_RECOVERED_frozen_policy_heldout_decisions.csv"
    )
)

assert len(r4_matches) == 1, (
    "STOP: frozen R4 held-out decision file missing or ambiguous."
)

r4_path = r4_matches[0]

r4 = pd.read_csv(
    r4_path
)

assert len(r4) == 360

required_r4 = [
    "position",
    "id",
    "truth",
    "final_pred",
    "uc_kind",
    "r4_final_escalated",
    "r4_final_autonomous",
    "r4_final_correct",
    "r4_final_wrong",
]

missing = [
    c for c in required_r4
    if c not in r4.columns
]

assert not missing, (
    f"STOP: missing R4 columns: {missing}"
)

# --------------------------------------------------------------------------------------------------------------
# 5. Load held-out semantic text from frozen R3 source
# --------------------------------------------------------------------------------------------------------------

align_matches = list(
    R3_RECOVERED.rglob(
        "R3_heldout360_R2V2_alignment.csv"
    )
)

assert len(align_matches) == 1

alignment = pd.read_csv(
    align_matches[0]
)

assert len(alignment) == 360

required_alignment = [
    "id",
    "text",
]

assert all(
    c in alignment.columns
    for c in required_alignment
)

# --------------------------------------------------------------------------------------------------------------
# 6. Reconfirm development / held-out separation
# --------------------------------------------------------------------------------------------------------------

dev_ids = set(
    pd.to_numeric(
        dev["id"],
        errors="coerce"
    ).dropna().astype(int)
)

heldout_ids = set(
    pd.to_numeric(
        alignment["id"],
        errors="coerce"
    ).dropna().astype(int)
)

overlap = (
    dev_ids &
    heldout_ids
)

assert len(overlap) == 0, (
    "STOP: development / held-out contamination."
)

print("\nDATA GOVERNANCE")
print("-" * 110)
print("Development rows :", len(dev))
print("Held-out rows    :", len(alignment))
print("ID overlap       :", len(overlap))
print("PASS - development / held-out separation preserved")

# --------------------------------------------------------------------------------------------------------------
# 7. Merge semantic text onto frozen R4 decisions
# --------------------------------------------------------------------------------------------------------------

heldout = r4.merge(
    alignment[
        [
            "id",
            "text",
        ]
    ],
    on="id",
    how="left",
    validate="one_to_one"
)

assert len(heldout) == 360

assert heldout["text"].notna().all()

# --------------------------------------------------------------------------------------------------------------
# 8. Reconstruct development-only TF-IDF deployment model
# --------------------------------------------------------------------------------------------------------------

classes = [
    "eMBB",
    "URLLC",
    "mMTC",
]

dev_text = (
    dev["text"]
    .fillna("")
    .astype(str)
    .tolist()
)

dev_labels = (
    dev["truth"]
    .astype(str)
    .to_numpy()
)

vectorizer = TfidfVectorizer(
    lowercase=True,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.98,
    sublinear_tf=True,
    norm="l2",
)

X_dev = vectorizer.fit_transform(
    dev_text
)

print("\nDEPLOYMENT SEMANTIC MODEL")
print("-" * 110)
print("Development rows:", X_dev.shape[0])
print("Features        :", X_dev.shape[1])
print("Held-out fit    : NO")

# --------------------------------------------------------------------------------------------------------------
# 9. Full development class prototypes
# --------------------------------------------------------------------------------------------------------------

prototypes = {}

for cls in classes:

    idx = np.where(
        dev_labels == cls
    )[0]

    assert len(idx) == 280

    prototypes[cls] = np.asarray(
        X_dev[idx].mean(axis=0)
    ).ravel()

# --------------------------------------------------------------------------------------------------------------
# 10. Helper
# --------------------------------------------------------------------------------------------------------------

def sparse_dense_cosine(
    sparse_row,
    dense_vector
):

    row_norm = np.sqrt(
        sparse_row.multiply(
            sparse_row
        ).sum()
    )

    dense_norm = np.linalg.norm(
        dense_vector
    )

    if row_norm == 0 or dense_norm == 0:
        return 0.0

    dot = sparse_row.dot(
        dense_vector
    )

    dot = np.asarray(
        dot
    ).ravel()[0]

    return float(
        dot /
        (
            row_norm *
            dense_norm
        )
    )

# --------------------------------------------------------------------------------------------------------------
# 11. Transform held-out text
# --------------------------------------------------------------------------------------------------------------

X_test = vectorizer.transform(
    heldout["text"]
    .fillna("")
    .astype(str)
    .tolist()
)

# --------------------------------------------------------------------------------------------------------------
# 12. Compute held-out semantic scores
#
# IMPORTANT:
# At this point the policy is already frozen.
# No threshold selection occurs here.
# --------------------------------------------------------------------------------------------------------------

semantic_rows = []

for i in range(len(heldout)):

    row_vector = X_test[i]

    similarities = {
        cls:
            sparse_dense_cosine(
                row_vector,
                prototypes[cls]
            )
        for cls in classes
    }

    ranked = sorted(
        similarities.items(),
        key=lambda x: x[1],
        reverse=True
    )

    semantic_pred = ranked[0][0]

    predicted_class = str(
        heldout.iloc[i]["final_pred"]
    )

    assert predicted_class in classes

    predicted_similarity = float(
        similarities[
            predicted_class
        ]
    )

    best_other_similarity = max(
        similarities[c]
        for c in classes
        if c != predicted_class
    )

    semantic_support_margin = (
        predicted_similarity -
        best_other_similarity
    )

    verification_flag = (
        semantic_support_margin
        <=
        FROZEN_THRESHOLD
    )

    semantic_rows.append({

        "sim_eMBB":
            similarities["eMBB"],

        "sim_URLLC":
            similarities["URLLC"],

        "sim_mMTC":
            similarities["mMTC"],

        "semantic_pred":
            semantic_pred,

        "pred_semantic_similarity":
            predicted_similarity,

        "best_other_semantic_similarity":
            best_other_similarity,

        "semantic_support_margin":
            semantic_support_margin,

        "r5_verification_flag":
            verification_flag,
    })

semantic_df = pd.DataFrame(
    semantic_rows
)

heldout = pd.concat(
    [
        heldout.reset_index(drop=True),
        semantic_df
    ],
    axis=1
)

# --------------------------------------------------------------------------------------------------------------
# 13. R5 applies ONLY to R4-autonomous cases
# --------------------------------------------------------------------------------------------------------------

heldout[
    "r5_additional_escalation"
] = (
    heldout[
        "r4_final_autonomous"
    ].astype(bool)
    &
    heldout[
        "r5_verification_flag"
    ].astype(bool)
)

heldout[
    "r5_final_escalated"
] = (
    heldout[
        "r4_final_escalated"
    ].astype(bool)
    |
    heldout[
        "r5_additional_escalation"
    ].astype(bool)
)

heldout[
    "r5_final_autonomous"
] = (
    ~heldout[
        "r5_final_escalated"
    ]
)

heldout[
    "r5_final_correct"
] = (
    heldout[
        "r5_final_autonomous"
    ]
    &
    (
        heldout["final_pred"]
        ==
        heldout["truth"]
    )
)

heldout[
    "r5_final_wrong"
] = (
    heldout[
        "r5_final_autonomous"
    ]
    &
    (
        heldout["final_pred"]
        !=
        heldout["truth"]
    )
)

# --------------------------------------------------------------------------------------------------------------
# 14. Frozen R4 baseline metrics
# --------------------------------------------------------------------------------------------------------------

r4_auto = int(
    heldout[
        "r4_final_autonomous"
    ].sum()
)

r4_esc = int(
    heldout[
        "r4_final_escalated"
    ].sum()
)

r4_correct = int(
    heldout[
        "r4_final_correct"
    ].sum()
)

r4_errors = int(
    heldout[
        "r4_final_wrong"
    ].sum()
)

# --------------------------------------------------------------------------------------------------------------
# 15. Frozen R5 metrics
# --------------------------------------------------------------------------------------------------------------

additional = int(
    heldout[
        "r5_additional_escalation"
    ].sum()
)

captured_errors = int(
    (
        heldout[
            "r5_additional_escalation"
        ]
        &
        heldout[
            "r4_final_wrong"
        ]
    ).sum()
)

correct_escalated = int(
    (
        heldout[
            "r5_additional_escalation"
        ]
        &
        heldout[
            "r4_final_correct"
        ]
    ).sum()
)

r5_auto = int(
    heldout[
        "r5_final_autonomous"
    ].sum()
)

r5_esc = int(
    heldout[
        "r5_final_escalated"
    ].sum()
)

r5_correct = int(
    heldout[
        "r5_final_correct"
    ].sum()
)

r5_errors = int(
    heldout[
        "r5_final_wrong"
    ].sum()
)

# --------------------------------------------------------------------------------------------------------------
# 16. Headline metrics
# --------------------------------------------------------------------------------------------------------------

total = len(heldout)

error_capture_recall = (
    captured_errors /
    r4_errors
    if r4_errors
    else np.nan
)

verification_precision = (
    captured_errors /
    additional
    if additional
    else np.nan
)

verification_fpr = (
    correct_escalated /
    r4_correct
    if r4_correct
    else np.nan
)

correct_preservation = (
    (
        r4_correct -
        correct_escalated
    )
    /
    r4_correct
    if r4_correct
    else np.nan
)

r5_coverage = (
    r5_auto /
    total
)

r5_accuracy = (
    r5_correct /
    r5_auto
    if r5_auto
    else np.nan
)

r5_risk = (
    r5_errors /
    r5_auto
    if r5_auto
    else np.nan
)

wrong_autonomous_rate = (
    r5_errors /
    total
)

additional_escalation_rate = (
    additional /
    r4_auto
    if r4_auto
    else np.nan
)

# --------------------------------------------------------------------------------------------------------------
# 17. Print first-and-only result
# --------------------------------------------------------------------------------------------------------------

print("\n" + "=" * 110)
print("FIRST AND ONLY FROZEN HELD-OUT RESULT")
print("=" * 110)

print("\nR4 BASELINE")
print("-" * 110)
print("Total                     :", total)
print("Autonomous                :", r4_auto)
print("Escalated                 :", r4_esc)
print("Autonomous correct        :", r4_correct)
print("Autonomous errors         :", r4_errors)
print("Coverage                  :", round(r4_auto / total, 6))
print("Autonomous accuracy       :", round(r4_correct / r4_auto, 6))
print("Selective risk            :", round(r4_errors / r4_auto, 6))

print("\nR5 SEMANTIC VERIFICATION")
print("-" * 110)
print("Additional escalations    :", additional)
print("R4 errors captured        :", captured_errors)
print("Correct cases escalated   :", correct_escalated)

print(
    "Error detection recall    :",
    round(error_capture_recall, 6)
)

print(
    "Verification precision    :",
    round(verification_precision, 6)
    if not np.isnan(verification_precision)
    else "N/A"
)

print(
    "Verification false-pos rate:",
    round(verification_fpr, 6)
)

print(
    "Correct preservation      :",
    round(correct_preservation, 6)
)

print("\nFINAL R5 SYSTEM")
print("-" * 110)
print("Autonomous                :", r5_auto)
print("Escalated                 :", r5_esc)
print("Autonomous correct        :", r5_correct)
print("Autonomous errors         :", r5_errors)
print("Coverage                  :", round(r5_coverage, 6))
print("Autonomous accuracy       :", round(r5_accuracy, 6))
print("Selective risk            :", round(r5_risk, 6))
print("Wrong autonomous / total  :", round(wrong_autonomous_rate, 6))
print("Added escalation / R4 auto:", round(additional_escalation_rate, 6))

# --------------------------------------------------------------------------------------------------------------
# 18. Residual R4 error detail
# --------------------------------------------------------------------------------------------------------------

residual_detail = heldout[
    heldout[
        "r4_final_wrong"
    ].astype(bool)
][
    [
        "position",
        "id",
        "uc_kind",
        "truth",
        "final_pred",
        "semantic_pred",
        "semantic_support_margin",
        "r5_verification_flag",
        "r5_additional_escalation",
    ]
].copy()

print("\nR4 RESIDUAL ERROR VERIFICATION")
print("-" * 110)

with pd.option_context(
    "display.max_columns", None,
    "display.width", 250
):
    print(
        residual_detail.to_string(
            index=False
        )
    )

# --------------------------------------------------------------------------------------------------------------
# 19. Comparison table
# --------------------------------------------------------------------------------------------------------------

comparison = pd.DataFrame(
    [
        {
            "system":
                "Frozen R4",

            "total":
                total,

            "autonomous":
                r4_auto,

            "escalated":
                r4_esc,

            "autonomous_correct":
                r4_correct,

            "autonomous_errors":
                r4_errors,

            "coverage":
                r4_auto / total,

            "autonomous_accuracy":
                r4_correct / r4_auto,

            "selective_risk":
                r4_errors / r4_auto,

            "wrong_autonomous_over_total":
                r4_errors / total,
        },

        {
            "system":
                "Frozen R4 + R5",

            "total":
                total,

            "autonomous":
                r5_auto,

            "escalated":
                r5_esc,

            "autonomous_correct":
                r5_correct,

            "autonomous_errors":
                r5_errors,

            "coverage":
                r5_coverage,

            "autonomous_accuracy":
                r5_accuracy,

            "selective_risk":
                r5_risk,

            "wrong_autonomous_over_total":
                wrong_autonomous_rate,
        },
    ]
)

# --------------------------------------------------------------------------------------------------------------
# 20. Save evidence
# --------------------------------------------------------------------------------------------------------------

decisions_path = (
    R5_RESULTS /
    "R5_FROZEN_heldout_decisions.csv"
)

comparison_path = (
    R5_RESULTS /
    "R5_FROZEN_system_comparison.csv"
)

residual_path = (
    R5_RESULTS /
    "R5_FROZEN_R4_residual_error_verification.csv"
)

heldout.to_csv(
    decisions_path,
    index=False
)

comparison.to_csv(
    comparison_path,
    index=False
)

residual_detail.to_csv(
    residual_path,
    index=False
)

summary = {

    "stage":
        "R5-10",

    "evaluation":
        "FIRST_AND_ONLY_FROZEN_HELDOUT_EVALUATION",

    "policy_sha256":
        actual_policy_sha,

    "frozen_semantic_support_threshold":
        FROZEN_THRESHOLD,

    "total":
        total,

    "r4": {
        "autonomous":
            r4_auto,

        "escalated":
            r4_esc,

        "autonomous_correct":
            r4_correct,

        "autonomous_errors":
            r4_errors,

        "coverage":
            r4_auto / total,

        "autonomous_accuracy":
            r4_correct / r4_auto,

        "selective_risk":
            r4_errors / r4_auto,
    },

    "r5": {
        "additional_escalations":
            additional,

        "r4_errors_captured":
            captured_errors,

        "correct_cases_escalated":
            correct_escalated,

        "error_detection_recall":
            error_capture_recall,

        "verification_precision":
            verification_precision,

        "verification_false_positive_rate":
            verification_fpr,

        "correct_autonomous_preservation":
            correct_preservation,

        "autonomous":
            r5_auto,

        "escalated":
            r5_esc,

        "autonomous_correct":
            r5_correct,

        "autonomous_errors":
            r5_errors,

        "coverage":
            r5_coverage,

        "autonomous_accuracy":
            r5_accuracy,

        "selective_risk":
            r5_risk,

        "wrong_autonomous_over_total":
            wrong_autonomous_rate,

        "additional_escalation_rate_over_r4_autonomous":
            additional_escalation_rate,
    },

    "post_heldout_retuning":
        False,
}

summary_path = (
    R5_RESULTS /
    "R5_10_FROZEN_HELDOUT_SUMMARY.json"
)

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=2,
        sort_keys=True
    )

# --------------------------------------------------------------------------------------------------------------
# 21. Hash evidence immediately
# --------------------------------------------------------------------------------------------------------------

print("\nFROZEN EVIDENCE HASHES")
print("-" * 110)

for p in [
    decisions_path,
    comparison_path,
    residual_path,
    summary_path,
]:

    print(
        f"{p.name:<55} "
        f"{sha256_file(p)}"
    )

print("\nPOLICY SHA256:")
print(actual_policy_sha)

print("\nPOST-HELDOUT RETUNING: PROHIBITED")
print("THRESHOLD REMAINS:", FROZEN_THRESHOLD)

print("\nPASS - R5-10 first and only frozen held-out evaluation completed")
print("=" * 110)
```

```
# ==============================================================================================================
# R5-10B - CORRECTED FIRST AND ONLY FROZEN HELD-OUT EVALUATION
#
# Correction:
# Semantic verification is evaluated ONLY for R4-autonomous cases.
# R4-escalated cases never reach R5 and therefore do not require a valid final_pred class.
#
# NO policy change.
# NO threshold change.
# NO held-out tuning.
# ==============================================================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import json
import hashlib

from sklearn.feature_extraction.text import TfidfVectorizer

print("=" * 110)
print("R5-10B - CORRECTED FIRST AND ONLY FROZEN HELD-OUT EVALUATION")
print("=" * 110)

R5_ROOT = Path("/content/R5_work")
R3_RECOVERED = R5_ROOT / "R3_recovered"
R4_RECOVERED = R5_ROOT / "R4_recovered"
R5_RESULTS = R5_ROOT / "R5_results"
R5_AUDIT = R5_ROOT / "audit"

POLICY_PATH = (
    R5_RESULTS /
    "R5_FROZEN_SEMANTIC_VERIFICATION_POLICY.json"
)

EXPECTED_POLICY_SHA = (
    "b8e5ef13e6895fcfb3656f7257c94727"
    "e3cea17d38f08563436d55b6cf0ba243"
)

CLASSES = [
    "eMBB",
    "URLLC",
    "mMTC",
]

# --------------------------------------------------------------------------------------------------------------
# 1. Utility
# --------------------------------------------------------------------------------------------------------------

def sha256_file(path, chunk_size=1024 * 1024):

    h = hashlib.sha256()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            h.update(chunk)

    return h.hexdigest()

# --------------------------------------------------------------------------------------------------------------
# 2. Verify frozen policy BEFORE evaluation
# --------------------------------------------------------------------------------------------------------------

assert POLICY_PATH.exists(), (
    "STOP: frozen R5 policy missing."
)

actual_policy_sha = sha256_file(
    POLICY_PATH
)

assert actual_policy_sha == EXPECTED_POLICY_SHA, (
    "STOP: frozen R5 policy SHA changed."
)

with open(
    POLICY_PATH,
    "r",
    encoding="utf-8"
) as f:

    policy = json.load(f)

FROZEN_THRESHOLD = float(
    policy[
        "semantic_verifier"
    ][
        "frozen_semantic_support_threshold"
    ]
)

R3_TAU = float(
    policy[
        "upstream_policy"
    ][
        "r3_feasible_mass_threshold"
    ]
)

R4_MARGIN_THRESHOLD = float(
    policy[
        "upstream_policy"
    ][
        "r4_margin_threshold"
    ]
)

assert np.isclose(
    FROZEN_THRESHOLD,
    -0.0275677986595984
)

print("\nFROZEN POLICY VERIFIED")
print("-" * 110)

print("Policy SHA256       :", actual_policy_sha)
print("R3 tau              :", R3_TAU)
print("R4 margin threshold :", R4_MARGIN_THRESHOLD)
print("R5 threshold        :", FROZEN_THRESHOLD)

print("\nPASS - frozen R5 policy unchanged")

# --------------------------------------------------------------------------------------------------------------
# 3. Locate source files
# --------------------------------------------------------------------------------------------------------------

dev_matches = list(
    R3_RECOVERED.rglob(
        "R3_oof_training_calibration_rows.csv"
    )
)

align_matches = list(
    R3_RECOVERED.rglob(
        "R3_heldout360_R2V2_alignment.csv"
    )
)

r4_matches = list(
    R4_RECOVERED.rglob(
        "R4_RECOVERED_frozen_policy_heldout_decisions.csv"
    )
)

assert len(dev_matches) == 1
assert len(align_matches) == 1
assert len(r4_matches) == 1

DEV_PATH = dev_matches[0]
ALIGN_PATH = align_matches[0]
R4_PATH = r4_matches[0]

# --------------------------------------------------------------------------------------------------------------
# 4. Load evidence
# --------------------------------------------------------------------------------------------------------------

dev = pd.read_csv(
    DEV_PATH
)

alignment = pd.read_csv(
    ALIGN_PATH
)

r4 = pd.read_csv(
    R4_PATH
)

assert len(dev) == 840
assert len(alignment) == 360
assert len(r4) == 360

print("\nSOURCE EVIDENCE")
print("-" * 110)

print("Development rows :", len(dev))
print("Alignment rows   :", len(alignment))
print("R4 decision rows :", len(r4))

# --------------------------------------------------------------------------------------------------------------
# 5. Data separation guard
# --------------------------------------------------------------------------------------------------------------

dev_ids = set(
    pd.to_numeric(
        dev["id"],
        errors="coerce"
    ).dropna().astype(int)
)

heldout_ids = set(
    pd.to_numeric(
        alignment["id"],
        errors="coerce"
    ).dropna().astype(int)
)

overlap = dev_ids & heldout_ids

assert len(overlap) == 0, (
    "STOP: development/held-out contamination."
)

print("\nDATA GOVERNANCE")
print("-" * 110)

print("Development IDs :", len(dev_ids))
print("Held-out IDs    :", len(heldout_ids))
print("ID overlap      :", len(overlap))

print("PASS - development / held-out separation preserved")

# --------------------------------------------------------------------------------------------------------------
# 6. Merge held-out text onto frozen R4 decisions
# --------------------------------------------------------------------------------------------------------------

heldout = r4.merge(
    alignment[
        [
            "id",
            "text",
        ]
    ],
    on="id",
    how="left",
    validate="one_to_one"
)

assert len(heldout) == 360
assert heldout["text"].notna().all()

# --------------------------------------------------------------------------------------------------------------
# 7. Audit frozen R4 routing BEFORE semantic evaluation
# --------------------------------------------------------------------------------------------------------------

r4_auto_mask = (
    heldout[
        "r4_final_autonomous"
    ].astype(bool)
)

r4_esc_mask = (
    heldout[
        "r4_final_escalated"
    ].astype(bool)
)

assert int(r4_auto_mask.sum()) == 306, (
    "STOP: expected 306 frozen R4 autonomous cases."
)

assert int(r4_esc_mask.sum()) == 54, (
    "STOP: expected 54 frozen R4 escalated cases."
)

assert not (
    r4_auto_mask &
    r4_esc_mask
).any()

assert (
    r4_auto_mask |
    r4_esc_mask
).all()

print("\nFROZEN R4 ROUTING")
print("-" * 110)

print("R4 autonomous :", int(r4_auto_mask.sum()))
print("R4 escalated  :", int(r4_esc_mask.sum()))

# --------------------------------------------------------------------------------------------------------------
# 8. Audit prediction values
# --------------------------------------------------------------------------------------------------------------

print("\nFINAL_PRED AUDIT")
print("-" * 110)

all_pred_counts = (
    heldout["final_pred"]
    .fillna("<NA>")
    .astype(str)
    .value_counts(dropna=False)
)

print("\nAll 360 rows:")
print(all_pred_counts.to_string())

auto_pred_counts = (
    heldout.loc[
        r4_auto_mask,
        "final_pred"
    ]
    .fillna("<NA>")
    .astype(str)
    .value_counts(dropna=False)
)

print("\nR4-autonomous rows only:")
print(auto_pred_counts.to_string())

invalid_auto = heldout[
    r4_auto_mask
    &
    (
        ~heldout[
            "final_pred"
        ].astype(str).isin(CLASSES)
    )
]

assert len(invalid_auto) == 0, (
    "STOP: an R4-autonomous row lacks a valid slice prediction."
)

print("\nPASS - all 306 R4-autonomous predictions are valid slice classes")

# --------------------------------------------------------------------------------------------------------------
# 9. Build semantic model from DEVELOPMENT ONLY
# --------------------------------------------------------------------------------------------------------------

dev_texts = (
    dev["text"]
    .fillna("")
    .astype(str)
    .tolist()
)

dev_labels = (
    dev["truth"]
    .astype(str)
    .to_numpy()
)

assert set(dev_labels).issubset(
    set(CLASSES)
)

vectorizer = TfidfVectorizer(
    lowercase=True,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.98,
    sublinear_tf=True,
    norm="l2",
)

X_dev = vectorizer.fit_transform(
    dev_texts
)

assert X_dev.shape[0] == 840

print("\nDEPLOYMENT SEMANTIC MODEL")
print("-" * 110)

print("Development rows :", X_dev.shape[0])
print("TF-IDF features  :", X_dev.shape[1])
print("Held-out fit     : NO")

# --------------------------------------------------------------------------------------------------------------
# 10. Full development class prototypes
# --------------------------------------------------------------------------------------------------------------

prototypes = {}

for cls in CLASSES:

    idx = np.where(
        dev_labels == cls
    )[0]

    assert len(idx) == 280

    prototypes[cls] = np.asarray(
        X_dev[idx].mean(axis=0)
    ).ravel()

print("PASS - development-only class prototypes constructed")

# --------------------------------------------------------------------------------------------------------------
# 11. Cosine helper
# --------------------------------------------------------------------------------------------------------------

def sparse_dense_cosine(
    sparse_row,
    dense_vector
):

    row_norm = np.sqrt(
        sparse_row.multiply(
            sparse_row
        ).sum()
    )

    dense_norm = np.linalg.norm(
        dense_vector
    )

    if row_norm == 0 or dense_norm == 0:
        return 0.0

    dot = sparse_row.dot(
        dense_vector
    )

    dot = np.asarray(
        dot
    ).ravel()[0]

    return float(
        dot /
        (
            row_norm *
            dense_norm
        )
    )

# --------------------------------------------------------------------------------------------------------------
# 12. Initialize semantic columns
#
# R4-escalated cases are deliberately left unevaluated by R5.
# --------------------------------------------------------------------------------------------------------------

heldout["sim_eMBB"] = np.nan
heldout["sim_URLLC"] = np.nan
heldout["sim_mMTC"] = np.nan

heldout["semantic_pred"] = pd.NA

heldout[
    "pred_semantic_similarity"
] = np.nan

heldout[
    "best_other_semantic_similarity"
] = np.nan

heldout[
    "semantic_support_margin"
] = np.nan

heldout[
    "r5_verification_flag"
] = False

# --------------------------------------------------------------------------------------------------------------
# 13. Transform ONLY R4-autonomous held-out text
# --------------------------------------------------------------------------------------------------------------

auto_indices = heldout.index[
    r4_auto_mask
].tolist()

auto_texts = (
    heldout.loc[
        auto_indices,
        "text"
    ]
    .fillna("")
    .astype(str)
    .tolist()
)

X_auto = vectorizer.transform(
    auto_texts
)

assert X_auto.shape[0] == 306

print("\nR5 EVALUATION POPULATION")
print("-" * 110)

print("R4-autonomous cases reaching R5:", X_auto.shape[0])
print("R4-escalated cases bypass R5   :", 54)

# --------------------------------------------------------------------------------------------------------------
# 14. Frozen semantic verification
#
# IMPORTANT:
# Threshold was frozen before this evaluation.
# --------------------------------------------------------------------------------------------------------------

for local_i, dataframe_index in enumerate(
    auto_indices
):

    row_vector = X_auto[
        local_i
    ]

    similarities = {
        cls:
            sparse_dense_cosine(
                row_vector,
                prototypes[cls]
            )
        for cls in CLASSES
    }

    ranked = sorted(
        similarities.items(),
        key=lambda x: x[1],
        reverse=True
    )

    semantic_pred = ranked[0][0]

    predicted_class = str(
        heldout.at[
            dataframe_index,
            "final_pred"
        ]
    )

    assert predicted_class in CLASSES

    predicted_similarity = float(
        similarities[
            predicted_class
        ]
    )

    best_other_similarity = max(
        similarities[c]
        for c in CLASSES
        if c != predicted_class
    )

    semantic_support_margin = (
        predicted_similarity -
        best_other_similarity
    )

    verification_flag = (
        semantic_support_margin
        <=
        FROZEN_THRESHOLD
    )

    heldout.at[
        dataframe_index,
        "sim_eMBB"
    ] = similarities["eMBB"]

    heldout.at[
        dataframe_index,
        "sim_URLLC"
    ] = similarities["URLLC"]

    heldout.at[
        dataframe_index,
        "sim_mMTC"
    ] = similarities["mMTC"]

    heldout.at[
        dataframe_index,
        "semantic_pred"
    ] = semantic_pred

    heldout.at[
        dataframe_index,
        "pred_semantic_similarity"
    ] = predicted_similarity

    heldout.at[
        dataframe_index,
        "best_other_semantic_similarity"
    ] = best_other_similarity

    heldout.at[
        dataframe_index,
        "semantic_support_margin"
    ] = semantic_support_margin

    heldout.at[
        dataframe_index,
        "r5_verification_flag"
    ] = bool(
        verification_flag
    )

print("PASS - frozen R5 verifier applied to exactly 306 cases")

# --------------------------------------------------------------------------------------------------------------
# 15. Construct final R5 routing
# --------------------------------------------------------------------------------------------------------------

heldout[
    "r5_additional_escalation"
] = (
    heldout[
        "r4_final_autonomous"
    ].astype(bool)
    &
    heldout[
        "r5_verification_flag"
    ].astype(bool)
)

heldout[
    "r5_final_escalated"
] = (
    heldout[
        "r4_final_escalated"
    ].astype(bool)
    |
    heldout[
        "r5_additional_escalation"
    ].astype(bool)
)

heldout[
    "r5_final_autonomous"
] = (
    ~heldout[
        "r5_final_escalated"
    ]
)

heldout[
    "r5_final_correct"
] = (
    heldout[
        "r5_final_autonomous"
    ]
    &
    (
        heldout["final_pred"]
        ==
        heldout["truth"]
    )
)

heldout[
    "r5_final_wrong"
] = (
    heldout[
        "r5_final_autonomous"
    ]
    &
    (
        heldout["final_pred"]
        !=
        heldout["truth"]
    )
)

# --------------------------------------------------------------------------------------------------------------
# 16. Verify frozen R4 baseline
# --------------------------------------------------------------------------------------------------------------

total = len(heldout)

r4_auto = int(
    heldout[
        "r4_final_autonomous"
    ].sum()
)

r4_esc = int(
    heldout[
        "r4_final_escalated"
    ].sum()
)

r4_correct = int(
    heldout[
        "r4_final_correct"
    ].sum()
)

r4_errors = int(
    heldout[
        "r4_final_wrong"
    ].sum()
)

assert total == 360
assert r4_auto == 306
assert r4_esc == 54
assert r4_correct == 301
assert r4_errors == 5

print("\nPASS - frozen R4 held-out baseline exactly reproduced")

# --------------------------------------------------------------------------------------------------------------
# 17. R5 headline counts
# --------------------------------------------------------------------------------------------------------------

additional = int(
    heldout[
        "r5_additional_escalation"
    ].sum()
)

captured_errors = int(
    (
        heldout[
            "r5_additional_escalation"
        ]
        &
        heldout[
            "r4_final_wrong"
        ]
    ).sum()
)

correct_escalated = int(
    (
        heldout[
            "r5_additional_escalation"
        ]
        &
        heldout[
            "r4_final_correct"
        ]
    ).sum()
)

r5_auto = int(
    heldout[
        "r5_final_autonomous"
    ].sum()
)

r5_esc = int(
    heldout[
        "r5_final_escalated"
    ].sum()
)

r5_correct = int(
    heldout[
        "r5_final_correct"
    ].sum()
)

r5_errors = int(
    heldout[
        "r5_final_wrong"
    ].sum()
)

assert additional == (
    captured_errors +
    correct_escalated
)

assert r5_auto + r5_esc == total

assert r5_correct + r5_errors == r5_auto

# --------------------------------------------------------------------------------------------------------------
# 18. Metrics
# --------------------------------------------------------------------------------------------------------------

error_capture_recall = (
    captured_errors /
    r4_errors
)

verification_precision = (
    captured_errors /
    additional
    if additional
    else np.nan
)

verification_fpr = (
    correct_escalated /
    r4_correct
)

correct_preservation = (
    (
        r4_correct -
        correct_escalated
    )
    /
    r4_correct
)

r5_coverage = (
    r5_auto /
    total
)

r5_accuracy = (
    r5_correct /
    r5_auto
    if r5_auto
    else np.nan
)

r5_risk = (
    r5_errors /
    r5_auto
    if r5_auto
    else np.nan
)

wrong_autonomous_rate = (
    r5_errors /
    total
)

additional_escalation_rate = (
    additional /
    r4_auto
)

# --------------------------------------------------------------------------------------------------------------
# 19. FIRST VALID FROZEN HELD-OUT RESULT
# --------------------------------------------------------------------------------------------------------------

print("\n" + "=" * 110)
print("FIRST VALID FROZEN HELD-OUT RESULT")
print("=" * 110)

print("\nR4 BASELINE")
print("-" * 110)

print("Total                      :", total)
print("Autonomous                 :", r4_auto)
print("Escalated                  :", r4_esc)
print("Autonomous correct         :", r4_correct)
print("Autonomous errors          :", r4_errors)
print("Coverage                   :", round(r4_auto / total, 6))
print("Autonomous accuracy        :", round(r4_correct / r4_auto, 6))
print("Selective risk             :", round(r4_errors / r4_auto, 6))

print("\nR5 SEMANTIC VERIFICATION")
print("-" * 110)

print("Additional escalations     :", additional)
print("R4 errors captured         :", captured_errors)
print("Correct cases escalated    :", correct_escalated)

print(
    "Error detection recall     :",
    round(
        error_capture_recall,
        6
    )
)

print(
    "Verification precision     :",
    (
        round(
            verification_precision,
            6
        )
        if not np.isnan(
            verification_precision
        )
        else "N/A"
    )
)

print(
    "Verification false-pos rate:",
    round(
        verification_fpr,
        6
    )
)

print(
    "Correct preservation       :",
    round(
        correct_preservation,
        6
    )
)

print("\nFINAL R5 SYSTEM")
print("-" * 110)

print("Autonomous                 :", r5_auto)
print("Escalated                  :", r5_esc)
print("Autonomous correct         :", r5_correct)
print("Autonomous errors          :", r5_errors)
print("Coverage                   :", round(r5_coverage, 6))
print("Autonomous accuracy        :", round(r5_accuracy, 6))
print("Selective risk             :", round(r5_risk, 6))
print("Wrong autonomous / total   :", round(wrong_autonomous_rate, 6))
print("Added escalation / R4 auto :", round(additional_escalation_rate, 6))

# --------------------------------------------------------------------------------------------------------------
# 20. R4 residual error detail
# --------------------------------------------------------------------------------------------------------------

residual_detail = heldout[
    heldout[
        "r4_final_wrong"
    ].astype(bool)
][
    [
        "position",
        "id",
        "uc_kind",
        "truth",
        "final_pred",
        "semantic_pred",
        "semantic_support_margin",
        "r5_verification_flag",
        "r5_additional_escalation",
    ]
].copy()

print("\nR4 RESIDUAL ERROR VERIFICATION")
print("-" * 110)

with pd.option_context(
    "display.max_columns", None,
    "display.width", 250
):

    print(
        residual_detail.to_string(
            index=False
        )
    )

# --------------------------------------------------------------------------------------------------------------
# 21. By use-case kind
# --------------------------------------------------------------------------------------------------------------

auto_before = heldout[
    heldout[
        "r4_final_autonomous"
    ]
].copy()

uc_rows = []

for uc_kind, group in auto_before.groupby(
    "uc_kind",
    dropna=False
):

    errors = int(
        group[
            "r4_final_wrong"
        ].sum()
    )

    flagged = int(
        group[
            "r5_additional_escalation"
        ].sum()
    )

    captured = int(
        (
            group[
                "r5_additional_escalation"
            ]
            &
            group[
                "r4_final_wrong"
            ]
        ).sum()
    )

    correct_flagged = (
        flagged -
        captured
    )

    uc_rows.append({

        "uc_kind":
            uc_kind,

        "r4_autonomous":
            len(group),

        "r4_errors":
            errors,

        "r5_flagged":
            flagged,

        "errors_captured":
            captured,

        "correct_escalated":
            correct_flagged,

        "error_capture_rate":
            (
                captured / errors
                if errors
                else np.nan
            ),
    })

uc_summary = pd.DataFrame(
    uc_rows
)

print("\nR5 RESULTS BY UC KIND")
print("-" * 110)

print(
    uc_summary.to_string(
        index=False
    )
)

# --------------------------------------------------------------------------------------------------------------
# 22. System comparison
# --------------------------------------------------------------------------------------------------------------

comparison = pd.DataFrame(
    [
        {
            "system":
                "Frozen R4",

            "total":
                total,

            "autonomous":
                r4_auto,

            "escalated":
                r4_esc,

            "autonomous_correct":
                r4_correct,

            "autonomous_errors":
                r4_errors,

            "coverage":
                r4_auto / total,

            "autonomous_accuracy":
                r4_correct / r4_auto,

            "selective_risk":
                r4_errors / r4_auto,

            "wrong_autonomous_over_total":
                r4_errors / total,
        },

        {
            "system":
                "Frozen R4 + R5",

            "total":
                total,

            "autonomous":
                r5_auto,

            "escalated":
                r5_esc,

            "autonomous_correct":
                r5_correct,

            "autonomous_errors":
                r5_errors,

            "coverage":
                r5_coverage,

            "autonomous_accuracy":
                r5_accuracy,

            "selective_risk":
                r5_risk,

            "wrong_autonomous_over_total":
                wrong_autonomous_rate,
        },
    ]
)

print("\nSYSTEM COMPARISON")
print("-" * 110)

with pd.option_context(
    "display.max_columns", None,
    "display.width", 250
):

    print(
        comparison.to_string(
            index=False
        )
    )

# --------------------------------------------------------------------------------------------------------------
# 23. Save FIRST VALID frozen evidence
# --------------------------------------------------------------------------------------------------------------

decisions_path = (
    R5_RESULTS /
    "R5_FROZEN_heldout_decisions.csv"
)

comparison_path = (
    R5_RESULTS /
    "R5_FROZEN_system_comparison.csv"
)

residual_path = (
    R5_RESULTS /
    "R5_FROZEN_R4_residual_error_verification.csv"
)

uc_path = (
    R5_RESULTS /
    "R5_FROZEN_results_by_uc_kind.csv"
)

heldout.to_csv(
    decisions_path,
    index=False
)

comparison.to_csv(
    comparison_path,
    index=False
)

residual_detail.to_csv(
    residual_path,
    index=False
)

uc_summary.to_csv(
    uc_path,
    index=False
)

summary = {

    "stage":
        "R5-10B",

    "evaluation":
        "FIRST_VALID_FROZEN_HELDOUT_EVALUATION",

    "previous_R5_10_status":
        "ABORTED_BEFORE_RESULT_DUE_TO_NONCLASS_FINAL_PRED_ON_R4_ESCALATED_ROWS",

    "methodological_change":
        False,

    "implementation_correction":
        (
            "R5 semantic verification applied only to R4-autonomous "
            "cases, which are the defined R5 input population."
        ),

    "policy_sha256":
        actual_policy_sha,

    "frozen_semantic_support_threshold":
        FROZEN_THRESHOLD,

    "total":
        total,

    "r4": {

        "autonomous":
            r4_auto,

        "escalated":
            r4_esc,

        "autonomous_correct":
            r4_correct,

        "autonomous_errors":
            r4_errors,

        "coverage":
            r4_auto / total,

        "autonomous_accuracy":
            r4_correct / r4_auto,

        "selective_risk":
            r4_errors / r4_auto,
    },

    "r5": {

        "additional_escalations":
            additional,

        "r4_errors_captured":
            captured_errors,

        "correct_cases_escalated":
            correct_escalated,

        "error_detection_recall":
            error_capture_recall,

        "verification_precision":
            (
                verification_precision
                if not np.isnan(
                    verification_precision
                )
                else None
            ),

        "verification_false_positive_rate":
            verification_fpr,

        "correct_autonomous_preservation":
            correct_preservation,

        "autonomous":
            r5_auto,

        "escalated":
            r5_esc,

        "autonomous_correct":
            r5_correct,

        "autonomous_errors":
            r5_errors,

        "coverage":
            r5_coverage,

        "autonomous_accuracy":
            r5_accuracy,

        "selective_risk":
            r5_risk,

        "wrong_autonomous_over_total":
            wrong_autonomous_rate,

        "additional_escalation_rate_over_r4_autonomous":
            additional_escalation_rate,
    },

    "heldout_retuning_performed":
        False,

    "post_heldout_retuning_allowed":
        False,
}

summary_path = (
    R5_RESULTS /
    "R5_10_FROZEN_HELDOUT_SUMMARY.json"
)

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        summary,
        f,
        indent=2,
        sort_keys=True
    )

# --------------------------------------------------------------------------------------------------------------
# 24. Hash frozen evidence immediately
# --------------------------------------------------------------------------------------------------------------

print("\nFROZEN EVIDENCE HASHES")
print("-" * 110)

evidence_files = [
    decisions_path,
    comparison_path,
    residual_path,
    uc_path,
    summary_path,
]

for p in evidence_files:

    print(
        f"{p.name:<55} "
        f"{sha256_file(p)}"
    )

print("\nPolicy SHA256:")
print(actual_policy_sha)

print("\nFrozen threshold:")
print(FROZEN_THRESHOLD)

print("\nPOST-HELDOUT RETUNING: PROHIBITED")

print("\nPASS - R5-10B first valid frozen held-out evaluation completed")
print("=" * 110)
```
