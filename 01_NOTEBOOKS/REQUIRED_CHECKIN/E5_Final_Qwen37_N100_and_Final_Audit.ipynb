```
import json
import importlib.util
from pathlib import Path
import pandas as pd

base = Path("/content/e5_model_smoke")

models = {
    "Qwen3.7-Max": base / "qwen37_n40" / "e5_summary.json",
    "MiniMax-M3": base / "minimax_m3_n40" / "e5_summary.json",
    "DeepSeek-V4-Pro": base / "deepseek_v4_pro_n40" / "e5_summary.json",
}

rows = []

for name, path in models.items():
    if not path.exists():
        print("MISSING:", path)
        continue

    with open(path) as f:
        s = json.load(f)

    rows.append({
        "model": name,
        "n": s.get("n_intents"),
        "accuracy": s.get("acc_overall"),
        "macro_f1": s.get("macro_f1"),
        "known": s.get("acc_known"),
        "unseen": s.get("acc_unseen"),
        "generic": s.get("acc_generic"),
        "ambiguous": s.get("acc_ambiguous"),
        "oov_gap": s.get("oov_gap"),
        "slot_recall": s.get("slot_recall"),
        "slot_fp": s.get("slot_false_positive"),
        "contract": s.get("contract_valid"),
        "retry": s.get("retry_rate"),
        "parse_failures": s.get("parse_failures"),
        "api_errors": s.get("api_errors"),
        "llm_calls": s.get("llm_calls_per_intent"),
        "tool_calls": s.get("tool_calls_per_intent"),
        "latency_s": s.get("latency_s_per_intent"),
        "cache_hit": s.get("cache_hit_rate"),
        "tokens_in": s.get("tokens_in"),
        "tokens_out": s.get("tokens_out"),
    })

df = pd.DataFrame(rows)

print("=" * 120)
print("CONTROLLED E5 MODEL COMPARISON — N=40")
print("=" * 120)

print(df.to_string(index=False))

print("\n" + "=" * 120)
print("ENVIRONMENT / PAPER-READINESS CHECK")
print("=" * 120)

print("LIME installed:",
      "✓ YES" if importlib.util.find_spec("lime") else "✗ NO")

# Check current E1 summary for actual SHAP backend
e1 = Path(
    "/content/xagentic_work/xagentic-slicing/"
    "experiment/results/e1_summary.json"
)

if e1.exists():
    with open(e1) as f:
        e1s = json.load(f)

    print("Current E1 SHAP backend:",
          e1s.get("shap_backend",
                  e1s.get("backend", "CHECK SUMMARY STRUCTURE")))
else:
    print("Current E1 summary: MISSING")

print("\n" + "=" * 120)
print("AUTOMATIC FLAGS")
print("=" * 120)

for _, r in df.iterrows():

    flags = []

    if r["parse_failures"] and r["parse_failures"] > 0:
        flags.append(f"{int(r['parse_failures'])} parse failures")

    if r["api_errors"] and r["api_errors"] > 0:
        flags.append(f"{int(r['api_errors'])} API errors")

    if r["cache_hit"] and r["cache_hit"] > 0:
        flags.append(f"cache hit {100*r['cache_hit']:.2f}%")

    if r["retry"] and r["retry"] > 0:
        flags.append(f"retry {100*r['retry']:.1f}%")

    if not flags:
        flags.append("clean")

    print(f"{r['model']:20s}: " + ", ".join(flags))
```
